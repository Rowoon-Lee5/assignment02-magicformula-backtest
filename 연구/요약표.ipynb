# 요약표: 전략 × 가중 × 기간
따라하기 01~04와 연구 노트북이 `결과_월수익률/`에 남긴 월수익률을 모아 한 표로 만든다. 숫자는 모두 `평가.요약()`이 계산한다.
- **기간**: 전체(그 노트북이 돌린 기간), 1990-04~2004-03, 2004-04~2017-03, 2017-04~ (보유 월 기준). 24개월 미만 구간은 뺀다.
- **평균(월 초과)**: 월평균 초과수익률(포트폴리오 − 무위험. 롱숏(LS)은 수익률 그대로). **평균 t(NW)**: 그 평균의 뉴이-웨스트 t.
- **샤프**: 연환산. **MDD**: 최대 낙폭. **Carhart α(월)**: 시장·SMB·HML·MOM 4요인 알파, **α t(NW)**: 그 t.
- 무위험이자율은 노트북마다 원문 설정을 따른다(01·04 국고3년, 02·03·연구 콜금리). 시장지수는 코스피 가격지수(I.001)라 배당 포함 가격(TR)을 쓴 전략은 알파가 배당만큼 높게 나온다.
- 전략 이름 끝의 Q1·LS는 분위 포트폴리오, '상위30'은 30종목 포트폴리오다.

줄임말: EW(동일가중, Equal Weight), VW(시가총액가중, Value Weight), LS(롱숏: 좋은 분위를 사고 나쁜 분위를 판 수익률),
MDD(최대 낙폭, Maximum Drawdown), IR(정보비율, Information Ratio), CAPM(자본자산가격결정모형), FF3(파마-프렌치 3요인), Carhart(카하트 4요인: FF3 + 모멘텀),
OLS(보통최소제곱), NW(뉴이-웨스트: 자기상관·이분산을 감안한 t), TR(배당 포함 수정주가).
