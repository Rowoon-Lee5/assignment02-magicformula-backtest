```python
# 공통 준비: 패키지 불러오기, 표시 설정, 결과 폴더, 칸 매직 두 개
import json, os, sys, time, types
import numpy as np
import pandas as pd
from IPython.core.magic import register_cell_magic

pd.set_option("display.float_format", "{:.4f}".format)  # 따라하기 노트북과 같은 표시 설정(표시만)
pd.set_option("display.width", 220)
pd.set_option("display.max_columns", 40)
pd.set_option("display.max_rows", 300)
for d in ["결과", "결과/결과_월수익률", "결과/결과_전략수익률"]:  # 결과는 노트북 옆 '결과' 폴더에만 씀(data 폴더는 읽기만)
    Path(d).mkdir(parents=True, exist_ok=True)


@register_cell_magic
def 모듈(line, cell):
    """칸 맨 위 '%%모듈 이름 별칭': 그 칸의 코드를 '이름'이라는 모듈(코드 묶음)로 실행하고 노트북에서는 '별칭'으로 부른다.
    왜: 저장소에서는 공통/의 파일 6개가 서로 `import 데이터읽기 as 읽기`처럼 불러 쓴다. 파일 하나로 내야 해서 그 코드를 칸에 그대로 넣고,
    이 매직이 칸 코드를 모듈로 만들어 sys.modules에 올려 그 import 줄이 그대로 돌게 한다(계산 코드는 한 글자도 안 바꿈)."""
    이름, *별칭 = line.split()
    m = types.ModuleType(이름)
    sys.modules[이름] = m
    exec(compile(cell, f"<{이름}>", "exec"), m.__dict__)
    get_ipython().user_ns[별칭[0] if 별칭 else 이름] = m


@register_cell_magic
def 절(line, cell):
    """칸 맨 위 '%%절 번호': 맨 위 칸의 실행할_절에 그 번호가 있으면(또는 '전부') 칸 코드를 그대로 실행, 없으면 건너뜀.
    왜: 저장소 노트북 칸을 글자 그대로 옮기면서 절마다 따로 돌릴 수 있게 하려고(칸 코드를 if 안에 넣으면 들여쓰기가 바뀜). 오류는 그대로 멈춤."""
    번호 = line.split()[0]
    if 실행할_절 == "전부" or any(번호 == s.strip() or 번호.startswith(s.strip() + ".") for s in str(실행할_절).split(",")):
        get_ipython().run_cell(cell).raise_error()
    else:
        print(f"{번호}절 건너뜀(실행할_절 = {실행할_절!r})")


import importlib.metadata as md
print("Python", sys.version.split()[0], "/", ", ".join(f"{p} {md.version(p)}" for p in ["pandas", "numpy", "scipy", "statsmodels", "matplotlib", "scikit-learn", "pyarrow"]))
print("data 폴더:", DATA.resolve(), "있음" if DATA.is_dir() else "없음 → 노트북 옆에 data 폴더를 두세요")
```

## 1. 공통 함수 (저장소 `공통/` 모듈 6개, 글자 그대로)
계산은 모두 여기 함수가 한다. 3~8절은 설정(dict)을 정하고 `포.실행(설정)`을 불러 결과를 본다. 순서: 데이터읽기(원본 파일 읽기) → 지표 → 종목거르기 → 포트폴리오(기준일 → 거르기 → 지표 → 점수 → 고르기 → 비중 → 매수보유 → 월수익률) → 평가(성과표·요인 회귀, t는 OLS) → 점검(손으로 답을 아는 작은 예로 계산 확인, 'ok'가 찍혀야 함).
- 각 칸 맨 위 `%%모듈 이름 별칭`은 0절 칸 매직(그 칸을 모듈로 실행). 함수마다 무엇을 왜 하는지는 그 함수의 설명 글(docstring)과 줄 끝 주석에 있다. 원문에 말이 없어 정한 기본값과 그 영향은 저장소 `_검증/기본값_영향_20261106.md`.
- 제출본에서 바꾼 줄: `데이터읽기`의 파일 경로 줄(→ `data` 폴더 아래 같은 이름 하위 폴더. 저장소 안 `_검증/새목록`은 `data/저장소_목록`으로 복사, 저장소 `_캐시/변환`은 같은 파일이 `data/통합_전체_202611`에 있어 `변환캐시 = DATA`), `평가`의 결과 폴더 줄(→ 노트북 옆 `결과/결과_월수익률`)과 git 기록 줄(`import shutil`을 더해 git이 없는 컴퓨터에서도 돌게, 기록 칸에만 쓰임). 나머지는 저장소와 같다.
- data에는 이 노트북이 읽는 파일을 넣었다(`data_파일목록.csv`). 이 노트북에서 쓰지 않는 거래량·상장주식수(`종목거르기.회전율`) 파일은 없으므로, 회전율 거르기를 켜려면 그 파일을 더 받아야 한다.

```python
%%모듈 데이터읽기 읽기
"""새로 통째로 받은 파일 읽기(추가 지시 1: 예전에 나눠 받아 합친 data\\통합_202611은 쓰지 않음, 값 대조용으로만).
경로는 이 파일 맨 위에서만 정한다. 읽은 표는 @cache로 아이템코드만 키로 들고 있으므로, 노트북에서 `데이터읽기.DATA`를 바꾸면 커널을 다시 시작해야 한다(감사 #17).
값은 원본 그대로 돌려준다. 고치거나 채우지 않는다."""
from functools import cache
from pathlib import Path

import numpy as np
import pandas as pd
import pyarrow.parquet as pq

from __main__ import DATA as 뿌리  # (제출본에서 바꿈) 노트북 맨 위 칸의 DATA = Path("data")
DATA = 뿌리 / "통합_전체_202611"  # DataGuide 달력기준 일간, I.00P, 1988-01-04~2026-09-30
XLSX = 뿌리 / "DataGuide_전체_202611"  # X01 지수, X03 상장일, X04 상장폐지일(원본 xlsx)
# 한 번에 받은 원본이라 그대로 씀(추가 지시 1 예외)
RF_FILE = 뿌리 / "DataGuide_1990s_202610" / "무위험이자율후보_DataGuide_시장금리전체아이템35종_월간_19880101-20260930.xlsx"
FACTOR_FILE = 뿌리 / "DataGuide_추가_202610" / "요인_팩터모델과시장지수_일간_1990-2026.xlsx"
새목록 = 뿌리 / "저장소_목록"  # B2(2026-11-04): 예전 데이터로 만든 목록을 새 데이터에서 같은 규칙으로 다시 만든 것(저장소 _검증/새목록에서 data/저장소_목록으로 복사)
ADD10 = FACTOR_FILE.parent  # DataGuide_추가_202610: '쓰지 말 것' 목록 csv(README_데이터), 코드 재사용 16종목 csv
CD_FILE = 뿌리 / "교수제공" / "08.[액셀실습]TSMom(K200).xlsx"  # 교수님 수업 자료, CD1m(추가 지시 10)
KRX_FILE = 뿌리 / "DataGuide_추가_202611" / "R12_통합지수_KRXTMI_KRX300_KRX300TR_KRX100_코스닥TR_I31000040F_일별_19991228-20260930.xlsx"

# 일간 아이템코드
TR, 수정주가, 종가, 거래대금, 거래량, 시총, 주식수, 정지구분 = "S410007700", "S410000700", "S41000060F", "S410000900", "S41000080F", "S410001250", "S41000100F", "S410002600"
시장구분, 업종, WICS, 관리여부, 정지여부, 결산월, 발표일 = "CP10000300", "CP10001090", "CP10000600", "CP10001660", "CP10001650", "CP10002200", "FP56000900"


변환캐시 = DATA  # (제출본에서 바꿈) 저장소에서는 data에 parquet가 없던 분기 R40~·결산 R32~를 저장소 _캐시/변환에 두었으나, 지금은 같은 이름 파일이 data\통합_전체_202611에 모두 있고 SHA-256도 같아(2026-11-05) 같은 폴더를 가리킴


def 목록(pat, 분기=False):
    """분기=False: 결산(연간)·일반 파일만, True: '분기3M' 파일만(같은 아이템코드로 결산·분기 두 파일이 있음, 추가 지시 F 2026-11-04)."""
    fs = [f for f in DATA.glob(f"통합_일간_{pat}.parquet") if not f.name.endswith(".tmp.parquet") and ("_분기" in f.name) == 분기]  # 변환 중인 .tmp 파일은 안 봄
    # data 폴더에 같은 파일이 있으면 그것, 없으면 저장소 캐시
    fs += [f for f in 변환캐시.glob(f"통합_일간_{pat}.parquet") if not f.name.endswith(".tmp.parquet") and ("_분기" in f.name) == 분기 and f.name not in {g.name for g in fs}]
    return fs


def 있음(item, 분기=False):
    return bool(목록(f"{item}_*", 분기))


def 파일(item, 분기=False):
    f = 목록(f"{item}_*", 분기)
    if not f:
        raise FileNotFoundError(f"{item} 파일이 {DATA}에 아직 없음 → 파일 오면 다시 실행")
    (f,) = f
    return f


@cache
def 공통끝():
    """통합_확인기록 F6: 항목마다 값이 끝나는 날이 다를 수 있다. 가격·시가총액·시장구분 일간 파일에 값이 있는 마지막 날 중 가장 이른 날 하나로 자른다.
    재무(M·FP)·결산월 일간은 아직 끝나지 않은 회계연도가 빈칸이라(예: 2026년 회계연도) 끝 날짜 판정에서 뺀다."""
    ends = [(t := pq.read_table(파일(i))).slice(max(t.num_rows - 260, 0)).to_pandas().dropna(how="all").index.max()
            for i in (TR, 수정주가, 종가, 시총, 시장구분) if 있음(i)]  # 결산월 등 회계연도 항목·드문드문한 항목은 안 봄(감사 M2)
    assert pd.notna(ends).all(), ends
    return min(ends)


@cache
def 일간(item):
    """통합_일간_{item}_*.parquet. 행=날짜(공통끝까지), 열=종목코드. 글자 값(시장구분·업종 등)은 범주형으로 읽는다(값은 같고 메모리만 줄임).
    6,293종목 파일(A0147B0 없음)도 그대로 읽는다. 다른 파일과 맞출 때 그 칸은 빈칸(reindex)."""
    return pq.read_table(파일(item)).to_pandas(strings_to_categorical=True).loc[:공통끝()]


@cache
def 상장중():
    """시장구분이 KS·KQ인 칸 = True(bool 표). 범주형 6천여 열 isin을 기준일마다 다시 하지 않게 한 번만(추가 지시 I, 2026-11-04)."""
    return 일간(시장구분).isin(["KS", "KQ"])


@cache
def 상장가격(item):
    """F2·F3: 시장구분(CP10000300)이 코스피·코스닥('KS'·'KQ')인 날의 가격만 계산에 쓴다.
    상장 전(이전상장 전 가격)·상장폐지 뒤(장외 K-OTC 등) 가격은 계산에서 빈칸으로 본다. 원본 값은 그대로.
    가격 0도 빈칸으로 본다: 0원은 거래 가격일 수 없고 수익률이 −100%·무한대가 된다(상장 구간 0원 칸 TR 4,748·수정주가 2,253, 대부분 1998~2001년 우선주 '우B',
    TR은 한국ANKOR유전 A152550 2012~2022년 2,652칸. `_검증/전처리_영향_결과.txt`, 진행기록 2026-11-03 08:15)."""
    p = 일간(item)
    return p.where(일간(시장구분).reindex(index=p.index, columns=p.columns).isin(["KS", "KQ"]) & (p > 0))


@cache
def 쓰지말것(item):
    """README_데이터.md '쓰지 말 것' 칸 → 표(코드, 시작, 끝). 값은 그대로 두고, 이 구간 가격을 쓰는 편입·수익률 계산만 피한다.
    A008080 2004-01~2013-09(D-05), 수정주가 병합·감자 미반영 7칸의 그 달(D27), 배당 포함 가격(TR)을 쓸 때 TR 이상칸의 앞날~그날(D-06·U3, 새 데이터로 다시 만든 목록).
    병합미반영은 예전 7칸 그대로: 새 데이터에 판정 조건 ③(DataGuide 1개월대비수익률)이 없어 새 목록(3,582줄)은 조건이 빠진 후보일 뿐."""
    d = ADD10
    m = pd.read_csv(d / "수정주가_병합미반영_7칸.csv", encoding="utf-8-sig", parse_dates=["날짜(월말)"])
    t = [("A008080", "2004-01-01", "2013-09-30", "D-05")] + [(c, x.replace(day=1), x, "D27") for c, x in zip(m["코드"], m["날짜(월말)"])]
    if item == TR:
        a = pd.read_csv(새목록 / "수정주가TR_이상칸.csv", encoding="utf-8-sig")  # 같은 규칙으로 새 데이터에서 다시 만든 232칸(예전 107칸 모두 포함, `_검증/B2_목록다시.py`·`새목록/차이표.md`)
        t += [(c, x, y, "TR이상칸:" + k) for c, x, y, k in zip(a["코드"], a["앞_날짜"], a["날짜"], a["유형"])]
    return pd.DataFrame(t, columns=["코드", "시작", "끝", "근거"]).astype({"시작": "datetime64[ns]", "끝": "datetime64[ns]"})


def 재무항목():
    """새 폴더의 재무(주재무 결산) 일간 파일 아이템코드: M(재무)·FP(발표일 등). FM(컨센서스)은 회계연도 값이 아니라 뺀다."""
    return sorted(f.name.split("_")[2] for p in ("M", "FP") for f in 목록(f"{p}*"))


검사 = {}  # 연간재무(): 항목마다 '결산일 값 ≠ 그 회계연도 첫 영업일 값'이라 빈칸으로 둔 행 수


@cache
def 연간재무():
    """달력기준 일간 재무 → (코드, 회계연도) 행(예전 통합_연간재무와 같은 모양: 코드·회계연도·CP10002200·아이템코드 열, 천원).
    달력기준 일간 값은 그 회계연도 기간 내내 그 회계연도 값이다(발표 전부터 들어감 → 그대로 쓰면 미래 정보).
    그래서 결산일 = 결산월 이력(CP10002200)이 그 달인 달의 마지막 영업일, 그날 값을 그 회계연도 값으로 뽑는다. 쓸 수 있는 날은 포트폴리오.재무가능일이 정한다.
    회계연도 = 결산일 연도 − (결산월 ≤ 5)(예전 연간재무 표기와 같음: 1~5월 결산은 회계연도 다음 해에 끝남).
    확인: 결산일 값이 그 회계연도 첫 영업일(직전 결산일 다음 달 첫 영업일, 첫 행은 11개월 전 달 첫 영업일) 값과 다르면
    그 칸은 그 회계연도 값이라고 믿을 수 없어 빈칸으로 둔다(값이 결산월과 다른 날 바뀌는 종목 = 다음 회계연도 값이 들어갈 수 있음, 감사 H1).
    둘 중 하나가 빈칸이면 확인 못 해 값 그대로 둔다. 빈칸으로 둔 수는 읽기.검사[아이템코드]."""
    m = 일간(결산월)
    g = m.index.to_series().groupby(m.index.to_period("M"))
    last, first = g.max(), g.min()  # 달마다 마지막·첫 영업일
    e = m.loc[last].apply(lambda x: pd.to_numeric(x.astype(object))).stack().rename("CP10002200").reset_index()
    e.columns = ["날짜", "코드", "CP10002200"]
    e = e[e["CP10002200"] == e["날짜"].dt.month].sort_values(["코드", "날짜"]).reset_index(drop=True)  # 그날 결산월 = 그 달 → 결산일
    e["회계연도"] = e["날짜"].dt.year - (e["CP10002200"] <= 5)
    pm = e["날짜"].dt.to_period("M")
    시작 = (pm.groupby(e["코드"]).shift(1).fillna(pm - 12) + 1).map(first)  # 그 회계연도 첫 영업일(데이터 전이면 빈칸)
    need = pd.DatetimeIndex(sorted(set(last) | set(first)))
    for item in 재무항목():  # 일간 전체(항목당 약 500MB)를 들고 있지 않도록 달 첫·마지막 영업일 줄만 읽는다
        t = pq.read_table(파일(item))
        d = t.take(pd.Index(t.column("날짜").to_pandas()).get_indexer(need)).to_pandas().reindex(columns=m.columns)  # 6,293종목 파일의 없는 종목은 빈칸
        assert d.index.equals(need), f"{item}: 결산월 파일과 날짜가 다름"
        c, x = d.columns.get_indexer(e["코드"]), d.to_numpy(float)
        a = x[d.index.get_indexer(e["날짜"]), c]
        ok = 시작.notna().to_numpy()
        b = np.full(len(e), np.nan)
        b[ok] = x[d.index.get_indexer(시작[ok]), c[ok]]
        bad = ~np.isnan(a) & ~np.isnan(b) & (a != b)
        e[item], 검사[item] = np.where(bad, np.nan, a), int(bad.sum())
    return e.drop(columns="날짜")


분기흐름 = ["M000908004", "M000904001", "M000909001", "M000906001", "M000701032"]  # 지배순이익·매출·영업CF·영업이익·EBITDA: 분기(3개월) 값 → 4분기 합 등
분기잔액 = ["M000903003", "M000903001", "M000902001", "M000901001", "M000901030"]  # 자본(지배)·자본총계·부채·자산·현금: 분기말 값


@cache
def 분기재무():
    """분기(3개월, Fiscal) 달력기준 일간 재무 → (코드, 분기말) 행. 값은 그 분기 동안 같고 분기 첫날 바뀐다(R40 삼성전자 확인).
    분기말 = 결산월 이력(CP10002200)과 3개월 간격인 달(결산월, −3, −6, −9)의 마지막 영업일. 그날 값을 그 분기 값으로, 그 분기 첫 영업일 값과 다르면
    믿을 수 없어 빈칸(연간재무와 같은 확인, 빈칸 수는 검사['분기:'+코드]). FP56000900(분기) = 그 분기 보고서 발표일. 받은 분기 파일만 열로 넣는다."""
    m = 일간(결산월)
    g = m.index.to_series().groupby(m.index.to_period("M"))
    last, first = g.max(), g.min()
    e = m.loc[last].apply(lambda x: pd.to_numeric(x.astype(object))).stack().rename("CP10002200").reset_index()
    e.columns = ["날짜", "코드", "CP10002200"]
    e = e[(e["날짜"].dt.month - e["CP10002200"]) % 3 == 0].sort_values(["코드", "날짜"]).reset_index(drop=True)
    시작 = (e["날짜"].dt.to_period("M") - 2).map(first)
    need = pd.DatetimeIndex(sorted(set(e["날짜"]) | set(시작.dropna())))
    for item in [i for i in 분기흐름 + 분기잔액 + ["FP56000900"] if 있음(i, 분기=True)]:
        t = pq.read_table(파일(item, 분기=True))
        d = t.take(pd.Index(t.column("날짜").to_pandas()).get_indexer(need)).to_pandas().reindex(columns=m.columns)
        assert d.index.equals(need), f"{item}: 분기 파일 날짜가 다름"
        c, x = d.columns.get_indexer(e["코드"]), d.to_numpy(float)
        a, b = x[d.index.get_indexer(e["날짜"]), c], x[d.index.get_indexer(시작), c]
        bad = ~np.isnan(a) & ~np.isnan(b) & (a != b)
        e[item], 검사["분기:" + item] = np.where(bad, np.nan, a), int(bad.sum())
    return e.rename(columns={"날짜": "분기말"})


@cache
def 종목정보():
    """X03 상장일·X04 상장폐지일(특정시점 최근, 원본 xlsx). 코드명은 지금 이름."""
    r = lambda f, c: pd.read_excel(next(XLSX.glob(f)), header=None, skiprows=6, names=["코드", "코드명", c], dtype={c: "Int64"}).set_index("코드")
    a, b = r("X03_*.xlsx", "상장일"), r("X04_*.xlsx", "상장폐지일")
    assert a["코드명"].equals(b["코드명"]), "X03·X04 종목 목록이 다름"
    return a.join(b["상장폐지일"])


@cache
def CD1m():
    """교수님 제공 엑셀(시트 TS모멘텀_m) B열 날짜(월말)·D열 CD1m(연 %) → 월 수익률(소수) = 연 % / 1200.
    엑셀 수식 =D6*(1/12)과 같은 월 단리. 값은 파일 그대로(1990-01~1994-12는 파일에 14로 들어 있음)."""
    x = pd.read_excel(CD_FILE, sheet_name="TS모멘텀_m", header=None)
    assert str(x.iloc[4, 3]).startswith("CD1m"), "D열 머리글이 CD1m이 아님"
    s = x.iloc[5:, 3].astype(float)
    s.index = pd.to_datetime(x.iloc[5:, 1])
    return s[s.index.notna()].rename_axis(None) / 1200


def 무위험(item):
    """DataGuide 금리(연 %, 월말) → 월 수익률(소수). 행=월말 날짜. item="CD1m"이면 교수님 제공 CD1m 한 계열(추가 지시 10)."""
    if item == "CD1m":
        return CD1m()
    x = pd.read_excel(RF_FILE, header=None)
    col = x.columns[x.iloc[14].eq(item).to_numpy()][0]  # 15행 = 아이템코드
    s = x.iloc[17:][col].astype(float)
    s.index = pd.to_datetime(x.iloc[17:, 5])
    return s.rename_axis(None) / 1200


@cache
def 국고3년일간():
    """X02 E11.02.011.001 국고채 3년 일간(연 %, 원본 xlsx 그대로). 우동호 외(2023) p.513 무위험 '3년 만기 국고채 수익률'의 일간 판(01 일간 정의용)."""
    x = pd.read_excel(next(XLSX.glob("X02_*.xlsx")), header=None)
    return pd.Series(x.iloc[14:, 1].astype(float).to_numpy(), pd.to_datetime(x.iloc[14:, 0])).dropna()


컨센서스파일 = {"EPS_FY1": "FM30011505_R17", "EPS_FY2": "FM30011505_R18", "추정기관수_FY1": "FM30012910_R19"}  # 같은 아이템코드(FM30011505)가 FY1·FY2 두 파일이라 R번호로 고름


@cache
def 컨센서스(이름):
    """애널리스트 추정치 일간(DataGuide 컨센서스, 최근 3개월 추정치. EPS는 지배주주 기준, 원/주, 수정 주식 수 기준). 행=날짜, 열=종목코드. 값은 그대로."""
    (f,) = 목록(f"{컨센서스파일[이름]}_*")
    return pq.read_table(f).to_pandas().loc[:공통끝()]


def 지수표(f, 코드행, 시작행):
    x = pd.read_excel(f, header=None)
    lv = x.columns[1:][x.iloc[코드행 + 3, 1:].eq("I31000040F").to_numpy()]  # 아이템코드 = 코드 3행 아래, 종가지수만
    t = x.iloc[시작행:][lv].set_axis(x.loc[코드행, lv], axis=1)
    t.index = pd.to_datetime(x.iloc[시작행:, 0])
    return t.astype(float).rename_axis(index=None, columns=None)


@cache
def 지수():
    """시장지수(X01: 코스피 I.001·코스피TR I.030·코스피200 동일가중 I.401·코스닥 I.201·코스닥TR I.273)와
    요인(FACTOR_FILE: 3FM.2B3.SMB·HML, 3FM.2M3.MOM)의 종가지수(일간). 열=코드. 같은 코드가 두 파일에 있으면 X01을 쓴다."""
    a = 지수표(next(XLSX.glob("X01_*.xlsx")), 8, 14)
    b = 지수표(FACTOR_FILE, 8, 14)
    return a.join(b.drop(columns=a.columns, errors="ignore"), how="outer")


@cache
def KRX지수():
    """R12 KRX 통합지수(KRX TMI 등, 1999-12-28~). 확장 연구의 참고 벤치마크."""
    return 지수표(KRX_FILE, 8, 14)
```

```python
%%모듈 지표 지표
"""지표마다 함수 1개: s → 종목별 값(Series). 공식대로 계산한 값 그대로(분모가 0 이하여도 손대지 않음).
s["fin"] = 그날 쓸 수 있는 회계연도 재무(천원, 열=아이템코드), s["mcap"] = 시가총액(백만원), s["ret"] = 직전 12개월 일간수익률.
비율 지표(PBR·PER·PSR·POCF·EV/EBITDA)는 적자·자본잠식 종목이 '싼 쪽'으로 줄 서지 않도록 역수(수익률 꼴)로 둔다 → 방향 '높게'."""
import numpy as np
import pandas as pd

import 데이터읽기 as 읽기

# 세전이익 = 세전계속사업이익(K-IFRS 법인세비용차감전순이익은 계속사업 기준, 중단사업손익은 세후로 따로 표시 → 들어가지 않음)
세전이익, 이자비용, 무형자산 = "M000907001", "M000906020", "M000901024"

유동자산, 현금, 유동부채, 영업이익 = "M000901002", "M000901030", "M000902003", "M000906001"
자산, 매출, 매출원가, 지배순이익, 부채, 자본, 지배자본 = "M000901001", "M000904001", "M000905001", "M000908004", "M000902001", "M000903001", "M000903003"
영업CF, EBITDA_, 순이익, 매출총이익 = "M000909001", "M000701032", "M000908001", "M000904007"


# 판본마다 값이 달라 어느 쪽도 고르지 않는 칸 → 그 회계연도에 빈칸으로 본다(값은 그대로). 호텔신라 FY2023 매출원가·매출총이익(R29, 통합_확인기록 F1, 회원 결정 2026-11-02)
빈칸으로볼칸 = {("A008770", 2023): {"M000905001", "M000904007"}}


def 값(s, code):
    x = s["fin"][code]
    bad = [c for (c, y), items in 빈칸으로볼칸.items() if code in items and s["fin"]["회계연도"].get(c) == y]
    return x.mask(x.index.isin(bad))


def 기업가치(s):
    """우동호 외(2023) 3.2절: 시가총액 + 부채총액 − 현금 (천원)."""
    return s["mcap"] * 1000 + 값(s, 부채) - 값(s, 현금)


# 마법공식
def EBIT(s):
    """우동호 외(2023) 3.2절 ②: 법인세차감전이익 + 이자비용."""
    return 값(s, 세전이익) + 값(s, 이자비용)


def BP_장부원문_6개월(s):
    """Novy-Marx(2010) 각주 3 장부가 = 자본 + 이연법인세 − 우선주: 자본총계(지배) + 이연법인세부채(M000911019) − 우선주자본금(M000903008),
    시가는 6개월 전 시가총액(BP_6개월과 같음). 원문(PDF 13쪽 각주 3) 'plus deferred taxes, minus preferred stock, when available' → 이연법인세·우선주 값이 빈 종목은 그 항만 뺌(0으로 셈).
    자본총계(지배)는 Compustat SEQ(우선주 포함 주주지분)에 가장 가까운 항목으로 고름(같은지는 확인 안 함). 자본(지배)이 빈 종목은 AT − LT(각주 3 마지막 단계). 두 항목 파일이 없으면 빈칸."""
    if not {"M000911019", "M000903008"} <= set(s["fin"]):
        return pd.Series(np.nan, s["codes"])
    m = 읽기.일간(읽기.시총).loc[:s["t"] - pd.DateOffset(months=6)].iloc[-1].reindex(s["codes"])
    eq = 값(s, 지배자본).fillna(값(s, 자산) - 값(s, 부채))  # 각주 3 'SEQ … or else … or else total assets minus total liabilities (AT - LT)'(중간 CEQ+PSTX는 대응 항목 없어 건너뜀, 2026-11-06)
    return (eq + 값(s, "M000911019").fillna(0) - 값(s, "M000903008").fillna(0)) / (m * 1000)


def EY(s):
    return EBIT(s) / 기업가치(s)


def 투하자본(s):
    return 값(s, 유동자산) - 값(s, 유동부채) + 값(s, 자산) - 값(s, 무형자산)


def ROC(s):
    """우동호 외(2023): EBIT / (유동자산 − 유동부채 + 총자산 − 무형자산)."""
    return EBIT(s) / 투하자본(s)


def EBIT_잠정(s):
    """01 확인판(추가 지시 F, 회원 결정 2026-11-04 15:13): 세전이익 = 세전계속사업이익 잠정치(FP56000300), 이자비용 = 지금 값(최초치 없음).
    잠정치가 없는 종목은 빈칸(확정치로 채우면 종목마다 다른 시점 값이 섞임, `_검증/추가로_받을것.md` 9절). 파일이 없으면 전부 빈칸."""
    if "FP56000300" not in s["fin"]:
        return pd.Series(np.nan, s["codes"])
    return 값(s, "FP56000300") + 값(s, 이자비용)


def EY_잠정(s):
    return EBIT_잠정(s) / 기업가치(s)


def ROC_잠정(s):
    return EBIT_잠정(s) / 투하자본(s)


def GPA(s):
    """우동호 외(2023) p.513 IMF: 매출총이익 / 총자산(Novy-Marx 2013 GP/A: 매출총이익 = 매출액 − 매출원가). 매출총이익 M000904007."""
    return 값(s, 매출총이익) / 값(s, 자산)


# 밸류 (역수 꼴)
def BP(s):
    """PBR 역수: 자본총계(지배) / 시가총액."""
    return 값(s, 지배자본) / (s["mcap"] * 1000)


def BP_6개월(s):
    """BP의 시가총액을 6개월 전(그날 또는 그 전 마지막 거래일) 값으로. Novy-Marx(2010) 각주 3 'lagged six months'(04 진단용)."""
    m = 읽기.일간(읽기.시총).loc[:s["t"] - pd.DateOffset(months=6)].iloc[-1].reindex(s["codes"])
    return 값(s, 지배자본) / (m * 1000)


def EP(s):
    """PER 역수: 당기순이익(지배) / 시가총액."""
    return 값(s, 지배순이익) / (s["mcap"] * 1000)


def SP(s):
    """PSR 역수: 매출액 / 시가총액."""
    return 값(s, 매출) / (s["mcap"] * 1000)


def CFP(s):
    """POCF 역수: 영업활동현금흐름 / 시가총액."""
    return 값(s, 영업CF) / (s["mcap"] * 1000)


def EBITDA_EV(s):
    """EV/EBITDA 역수: EBITDA / 기업가치."""
    return 값(s, EBITDA_) / 기업가치(s)


def EBITDA_EV_차입금(s):
    """02 확인판(회원 결정 2026-11-04 15:51): EV = 시가총액 + 단기차입금(M000902009) + 유동성장기부채(M000902044) − 현금. NH 밸류 원문에 EV 정의가 없어
    본문 EV(시총 + 부채총계 − 현금)는 그대로 두고 비교만. 두 항목은 결산 파일뿐이라 연간 재무로만 돈다. 빈칸은 빈칸(0으로 안 채움)."""
    return 값(s, EBITDA_) / (s["mcap"] * 1000 + 값(s, "M000902009") + 값(s, "M000902044") - 값(s, 현금))


def 영업이익_EV(s):
    """EV/영업이익 역수(칼라일 '인수자 배수'): 영업이익(발표기준) M000906002 / 기업가치."""
    return 값(s, "M000906002") / 기업가치(s)


# 퀄리티
def ROE(s):
    return 값(s, 지배순이익) / 값(s, 지배자본)


def ROE_양자본(s):
    """확인판: 지배자본 > 0인 종목만(자본이 음수면 ROE 부호가 뒤집혀 좋은 쪽에 줄 섬). 값은 그대로."""
    return ROE(s).where(값(s, 지배자본) > 0)


def ROA(s):
    """당기순이익(연결 전체, M000908001) / 자산총계."""
    return 값(s, 순이익) / 값(s, 자산)


def 영업CF자산(s):
    """영업활동현금흐름 / 자산총계(현금 기준 수익성)."""
    return 값(s, 영업CF) / 값(s, 자산)


def 영업이익률(s):
    return 값(s, 영업이익) / 값(s, 매출)


def 부채비율(s):
    return 값(s, 부채) / 값(s, 자본)


def 부채비율_양자본(s):
    """확인판: 자본총계 > 0인 종목만."""
    return 부채비율(s).where(값(s, 자본) > 0)


def 총자산회전율(s):
    return 값(s, 매출) / 값(s, 자산)


# 위험
def 일년상장(s):
    """직전 12개월 내내 시장구분이 코스피·코스닥이던 종목(F2: 상장일은 이전상장 전 날짜라 쓰지 않음, 코드 재사용 공백도 여기서 걸림)."""
    return s["일년상장"]


def 변동성(s):
    """NH(2018-10): 직전 12개월 일간수익률 표준편차. 원문에 최소 관측 기준이 없어 12개월 내내 상장된 종목만. 기간 안 빈칸은 빼고 계산."""
    return s["ret"].std().where(일년상장(s))


def 변동성_원문(s):
    """따라하기 03(NH 2018-10 p.23 '일간수익률의 12개월 변동성'): 직전 12개월 창 안에 있는 일간수익률로 표준편차('직전 12개월 내내 상장' 조건 없음, 원문에 없어 뺌 05:47).
    값이 모자라 계산이 안 되는 종목(수익률 2개 미만)만 빈칸 = 그 달에서 빠짐(기본 ①). '쓰지 말 것' 칸은 빈칸(재료와 같음)."""
    return s["ret"].std()


def 변동성_창(s, 개월=12, 주기="일", 상장조건=True):
    """NH(2018-10) p.23 그림 33·34: 직전 '개월' 동안 일간('일')·주간('주', 금요일 끝)·월간('월', 월말) 수익률 표준편차.
    상장조건=True(기본, 연구용): 변동성()과 같은 규칙 — 창 내내 시장구분이 코스피·코스닥이고 창 첫날 상장 가격이 있는 종목만. 상장조건=False(따라하기 03): 창 안 값으로만(변동성_원문과 같음).
    '쓰지 말 것' 칸은 빈칸. 개월=12·주기='일'이면 변동성()(상장조건=True)·변동성_원문()(False)과 같은 값."""
    t, c = s["t"], s["codes"]
    a = t - pd.DateOffset(months=개월)
    p = 읽기.상장가격(s["가격"]).loc[a:t, c].copy()
    for k, x, y in 읽기.쓰지말것(s["가격"])[["코드", "시작", "끝"]].itertuples(index=False):
        if k in p:
            p.loc[x:y, k] = np.nan
    ok = 읽기.상장중().loc[a:t].reindex(columns=c, fill_value=False).all() & p.iloc[0].notna()
    if 주기 != "일":
        p = p.resample({"주": "W-FRI", "월": "ME"}[주기]).last()
    sd = p.pct_change(fill_method=None).iloc[1:].std()
    return sd.where(ok) if 상장조건 else sd  # 상장조건=False: 따라하기 03(변동성_원문과 같은 규칙, 창 안 값으로만)


# 애널리스트 추정치(NH 2018-07 p.19 그림 25·26)
def EP_선행(s):
    """12개월 선행 E/P = 12개월 선행 EPS / 수정주가. 12개월 선행 EPS = FY1 × (12 − m)/12 + FY2 × m/12, m = 그날의 달(NH 2018-07 p.19는 '12개월 Forward'라는 말만 있어 가장 단순한 식 = 우리가 고른 값, 기본값 목록 02-24).
    한계(G02 2026-11-06): DataGuide FY1은 1월 초가 아니라 대부분 직전 회계연도 실적이 나온 뒤(2~3월) 다음 해로 넘어감 → 1·2월(드물게 3월)에는 이미 끝난 회계연도에 큰 가중이 붙음
    (3분위 같은 칸 비율 1월 74%, 2월 88~94%, 3~12월 94~100%; 넘어감을 감지하는 판은 `_검증/1106/G02_그림25_26_셀.py` 보고, 쓰지 않음). 결산월이 12월이 아닌 종목도 같은 m.
    추정치는 DataGuide '최근 3개월' 컨센서스(원문 '애널리스트의 최근 3개월 EPS 추정치'). FY1·FY2 중 하나라도 없으면 빈칸."""
    t, c = s["t"], s["codes"]
    f1, f2 = (읽기.컨센서스(k).loc[:t].iloc[-1].reindex(c) for k in ("EPS_FY1", "EPS_FY2"))
    m = t.month
    return (f1 * (12 - m) / 12 + f2 * m / 12) / 읽기.일간(읽기.수정주가).loc[:t].iloc[-1].reindex(c)
```

```python
%%모듈 종목거르기 종목거르기
"""종목 거르기. 필터마다 함수 1개: (s, c, 옵션) → 남는 종목코드(Index).
s = 포트폴리오.재료()가 만든 그날 자료, c = 지금까지 남은 종목. 값이 빈 종목은 그 필터에서 빠진다(빈칸은 빈칸)."""
import numpy as np
import pandas as pd

import 데이터읽기 as 읽기
import 지표


def 그날(item, t, c):
    """그날 값. 월간 자료(관리종목 등)는 그날 이전 가장 최근 기준일 행을 그대로 쓴다."""
    d = 읽기.일간(item)
    i = d.index.searchsorted(t, side="right") - 1  # t 이전 마지막 행 위치(reindex(ffill)과 같음, 범주형 6천여 열 reindex가 느려 위치로 찾음 2026-11-04)
    return (d.iloc[i] if i >= 0 else pd.Series(np.nan, d.columns)).reindex(c)


def 시장(s, c, 시장들):
    """시장구분 CP10000300: 'KS' 유가증권, 'KQ' 코스닥."""
    return c[그날(읽기.시장구분, s["t"], c).isin(시장들)]


def 쓰지말것(s, c):
    """README_데이터 '쓰지 말 것' 구간(읽기.쓰지말것, (시작, 끝])에 기준일이 들어가는 종목 제외 = 그날 값이 의심인 종목(회원 결정 2026-11-02). 값은 그대로.
    기준일까지 알 수 있는 것만 본다(2026-11-04 고침: 전에는 [기준일, 다음 리밸런싱일]을 봐서 미래 정보였음, README 한계 '판정 창').
    보유 중에 걸리는 칸은 포트폴리오.쓰지말것칸이 그 칸 수익률만 바꾼다."""
    b = 읽기.쓰지말것(s["가격"])
    return c[~c.isin(b.loc[(b["시작"] < s["t"]) & (b["끝"] >= s["t"]), "코드"])]


def 종목빼기(s, c, 코드들):
    """확인판: 지정 종목을 뺀다(A070480 이상값 확인, 회원 결정 2026-11-02)."""
    return c[~c.isin(코드들)]


def 보통주(s, c):
    """수업 노트북: 코드 끝자리 0 = 보통주."""
    return c[c.str.endswith("0")]


def 스팩(s, c):
    return c[~읽기.종목정보()["코드명"].reindex(c).str.contains("스팩", na=False)]


def 금융업(s, c, 낱말=("은행", "보험", "증권", "금융", "금고"), 이름="넓게"):
    """그날 거래소 업종(CP10001090)에 낱말이 있으면 제외. 업종은 1988~2026 기간에 있지만 빈칸인 종목이 있다(수는 기록의 확인:업종빈칸).
    업종이 빈 종목은 이름('(주)'를 뗀 지금 이름)에 은행·보험·증권·금융·금고·캐피탈·카드·저축·신금·리츠·스팩이 있거나 '지주'로 끝나면 제외
    (스팩 = 기업인수목적회사, 거래소 업종이 있는 날은 '금융'으로 들어 있어 업종 빈칸 날만 이름으로 잡음, 감사 2026-11-03 #15)
    (수업 노트북 이름 규칙을 넓힘. 금융지주·카드·캐피탈·저축은행이 빠지던 것 막기, '리스'는 엑큐리스 같은 이름이 걸려 뺌).
    업종이 빈 날 그날 WICS(CP10000600, 2000년부터)가 '금융'이어도 제외(2026-11-04: 이름 규칙이 놓친 경매·해외개발 펀드, 지금 이름이 바뀐 스팩 시절 26종목, `_검증/B5_금융업이름_결과.txt`).
    이름='원문': 따라하기용(우동호 p.512 금융 7업종 = 은행·보험·증권·투자금융·신용금고·종합금융·기타금융). 이름 규칙을 7업종에 드는 낱말(은행·보험·증권·금융·금고·캐피탈·카드·저축·신금)만으로
    (리츠·스팩·'지주'로 끝남은 7업종이 아니라 뺌, 회원 05:57). 거래소 업종 낱말과 업종 빈 날 WICS '금융'은 같음."""
    krx = 그날(읽기.업종, s["t"], c)
    name = 읽기.종목정보()["코드명"].reindex(c).str.replace(r"\(주\)|㈜", "", regex=True).str.strip()
    wics = 그날(읽기.WICS, s["t"], c).astype(object).eq("금융") if 읽기.있음(읽기.WICS) else False
    규칙 = "은행|보험|증권|금융|금고|캐피탈|카드|저축|신금" + ("" if 이름 == "원문" else "|리츠|스팩|지주$")
    fin = krx.str.contains("|".join(낱말), na=False) | (krx.isna() & (name.str.contains(규칙, na=False) | wics))
    return c[~fin]


def 투자기구(s, c):
    """선박투자회사·부동산투자회사(CR리츠 포함)·리츠 제외. 일반기업과 재무제표 형태가 달라 금융업을 빼는 이유와 같음(회원 결정 2026-11-01).
    거래소 업종이 '일반서비스' 등으로 들어 있어 이름(지금 이름)으로 판정: 선박투자회사·부동산투자가 들어감, '리츠'로 끝남, 리츠코크렙."""
    name = 읽기.종목정보()["코드명"].reindex(c).str.replace(r"\(주\)|㈜", "", regex=True).str.strip()
    return c[~name.str.contains("선박투자회사|부동산투자|리츠$|리츠코크렙", na=False)]


def 결산월(s, c, 월=12):
    """그날 쓰는 재무 행의 결산월(회계연도별 이력 CP10002200)."""
    return c[s["fin"]["CP10002200"].reindex(c) == 월]


def 거래정지(s, c, 아이템=읽기.정지구분):
    """일간 S410002600('거래정지') 또는 일간 CP10001650('정지', 10,061일 파일)."""
    return c[~그날(아이템, s["t"], c).isin(["거래정지", "정지"])]


def 관리종목(s, c):
    return c[그날(읽기.관리여부, s["t"], c).ne("관리")]


def 동전주(s, c, 하한=1000):
    """종가가 하한 미만인 종목 제외(수업: 1,000원 미만)."""
    return c[그날(읽기.종가, s["t"], c) >= 하한]


def 자본잠식(s, c):
    return c[s["fin"]["M000903001"].reindex(c) > 0]


def 최근20일(item, t, c):
    return 읽기.일간(item).loc[:t, c].iloc[-20:]


def 회전율(s, c, 하위=0.1):
    """최근 20거래일 평균 거래량 / 그날 상장주식수, 하위 비율 제외(수업: 10%)."""
    turn = 최근20일(읽기.거래량, s["t"], c).mean() / 그날(읽기.주식수, s["t"], c)
    return c[turn > turn.quantile(하위)]


def 거래대금(s, c, 하한=1e9, 창="20일"):
    """평균 거래대금(S410000900, 원) ≥ 하한(NH: 10억 원). 창='20일': 최근 20거래일, '월': 기준일이 든 달의 거래일(NH p.26 '월평균 거래대금')."""
    x = 최근20일(읽기.거래대금, s["t"], c) if 창 == "20일" else 읽기.일간(읽기.거래대금).loc[f"{s['t']:%Y-%m}":s["t"], c]
    return c[x.mean() >= 하한]


def 지표있음(s, c, 지표들):
    """지표 값이 모두 있는 종목만(±inf는 없는 것으로 봄). Novy-Marx p.21 '500 largest … for which GP/A and B/M are both available'처럼 시총상위 앞에 둔다."""
    v = pd.DataFrame({m: getattr(지표, m)(s).reindex(c) for m in 지표들}).replace([np.inf, -np.inf], np.nan)
    return c[v.notna().all(axis=1).to_numpy()]


def 시총상위(s, c, n=300):
    return s["mcap"].reindex(c).nlargest(n).index


def 시총분위(s, c, 분위, k=5, 기준="남은"):
    """그날 시가총액 k분위 중 한 분위만(1 = 가장 작은 분위). 기준='남은': 앞 단계까지 남은 종목으로 나눔(기본),
    '상장전체': 그날 가격 있는 KS·KQ 전 종목의 분위 경계로 남은 종목을 나눔(원문 p.510·514 '시가총액 기준 5분위'의 모집단이 적혀 있지 않아 민감도용)."""
    m = s["mcap"].reindex(c).dropna()
    if 기준 == "상장전체":
        a = s["mcap"].reindex(시장(s, s["codes"], ["KS", "KQ"])).dropna()
        b = a.quantile([i / k for i in range(1, k)]).to_numpy()
        return m.index[(np.searchsorted(b, m.to_numpy(), side="right") + 1) == 분위]
    q = pd.qcut(m.rank(method="first"), k, labels=range(1, k + 1))
    return m.index[q == 분위]


def 적용(s, 목록):
    """목록 순서대로 거른다. 항목은 '이름' 또는 ('이름', {옵션}). 단계별 남은 종목 수도 돌려준다."""
    c, n = s["codes"], {"가격있음": len(s["codes"])}
    for f in 목록:
        name, kw = (f, {}) if isinstance(f, str) else f
        c = globals()[name](s, c, **kw)
        n[name] = len(c)
    return c, n
```

```python
%%모듈 포트폴리오 포
"""포트폴리오 만들기와 보유. 노트북은 설정(dict)만 바꿔 `실행(설정)`을 부른다.
흐름(수업 노트북과 같음): 날짜×종목 표 → 월 기준일 → 기준일마다 거르기·지표·고르기·비중 → 다음 리밸런싱까지 매수보유 → 월수익률."""
from collections import defaultdict
from functools import cache

import numpy as np
import pandas as pd
from scipy.optimize import minimize
from sklearn.covariance import LedoitWolf

import 데이터읽기 as 읽기
import 종목거르기
import 지표

기본설정 = dict(
    시장=["KS", "KQ"],                 # 시장구분 CP10000300
    가격=읽기.TR,                       # 수익률 계산 가격: 읽기.TR(배당 포함) | 읽기.수정주가
    기간=("2004-04", "2006-12"),        # (첫 리밸런싱 월, 마지막 기준일 월). 설정에 "끝": "마지막거래일"을 넣으면 끝 달 마지막 거래일을 한 칸 더함
    기준일="첫거래일",                  # 월 기준일: '첫거래일' | '마지막거래일'
    리밸런싱=12,                        # 몇 달마다 다시 고르나: 1 | 3 | 12 (사이에는 매수보유)
    재무시점=("결산후개월", 3),          # ('결산후개월', k): 결산일+k개월부터(발표일 안 봄) | ('발표일', k): 실적보고서발표일 다음 거래일부터(없으면 결산일+k일)
    제외=["보통주", "금융업", "투자기구", "거래정지"],  # 종목거르기 함수 이름, 또는 ('이름', {옵션})
    지표=[("ROC", "높게"), ("EY", "높게")],  # (지표 함수, 방향[, 단계별일 때 남길 비율])
    합산="순위합산",                    # '순위합산' | 'z합산' | '단계별'
    고르기=("상위N", 30),               # ('상위N', n) | ('분위', k): Q1이 가장 좋은 쪽
    비중=["동일"],                      # '동일' | '시총' | '시총상한' | '위험균형' | '최소분산'
    비중상한=0.1,                       # 시총상한에서만 씀
    거래비용=dict(수수료=0.00015, 거래세=0.0025, 슬리피지=0.005),  # 매수·매도마다 수수료+슬리피지, 매도에 거래세. 월비용=x를 넣으면 매달 x를 뺌(NH 현실판)
    전량교체=False,                     # True: 리밸런싱 때 전부 팔고 새로 삼(겹치는 종목도 비용) | False: 비중 바뀐 만큼만
    무위험="E11.02.003.001",            # 콜금리(1일물 중개거래). 1988년부터 빈 달 없는 것은 이것과 무담보콜(E11.02.003.002) 둘
    시장지수="I.001",                   # 코스피
)

# 우동호 외(2023) 설정. 연구 노트북(연구_틀·이로운 3개·확장연구)의 비교 기준 = master 때 동작 그대로(감독 05:50 지시: 회원 결정은 따라하기에만).
우동호2023 = {**기본설정, "이름": "우동호2023_MF", "기간": ("2004-04", "2022-04"),
              "제외": ["보통주", "금융업", "투자기구", ("결산월", {"월": 12}), "거래정지"], "무위험": "E11.02.003.008", "가격": 읽기.수정주가, "전량교체": True}
# 따라하기 01(원문대로, 2026-11-06). 원문 p.512: 금융 7업종(은행·보험·증권·투자금융·신용금고·종합금융·기타금융) 제외, 12월 결산, '거래중지 중이면 표본에서 제외',
# '직전년도에 공시된 재무제표' → 회계연도 t−1 결산만(없으면 그 종목은 그해 빠짐). 투자기구(리츠·선박) 제외는 원문에 없어 뺌(회원 05:57).
우동호2023_원문 = {**우동호2023, "제외": ["보통주", ("금융업", {"이름": "원문"}), ("결산월", {"월": 12}), "거래정지"], "재무연도": "직전년도"}


def 기준일(cfg):
    d = 읽기.일간(cfg["가격"]).index.to_series()
    g = d.groupby(d.dt.to_period("M"))
    x = (g.min() if cfg["기준일"] == "첫거래일" else g.max()).loc[cfg["기간"][0]:cfg["기간"][1]]
    if cfg.get("끝") == "마지막거래일":  # 마지막 구간을 끝 달 마지막 거래일에서 끊음(우동호 p.512 '다음연도 3월 마지막 거래일까지')
        x = pd.concat([x, g.max().loc[x.index[-1:]]])
    return pd.DatetimeIndex(x.to_numpy())


@cache
def 재무가능일(how):
    """연간재무 행마다 결산일·발표일과 '쓸 수 있는 날'. 결산월은 회계연도별 이력(CP10002200).
    결산월 1~5월은 회계연도 다음 해에 끝난다(DataGuide 표기). 발표일이 결산일보다 이르면 믿지 않는다."""
    f = 읽기.연간재무().dropna(subset=["CP10002200"])
    m = f["CP10002200"].astype(int)
    end = pd.to_datetime((f["회계연도"] + (m <= 5)).astype(int).astype(str) + m.astype(str).str.zfill(2), format="%Y%m") + pd.offsets.MonthEnd(0)
    kind, k = how
    if "FP56000900" not in f:  # 발표일 파일이 아직 없으면 발표일을 안 보는 '결산후개월'만 돈다(빈 발표일로 대신하지 않음)
        assert kind == "결산후개월", "발표일 FP56000900 파일이 아직 없음 → 파일 오면 다시 실행"
        return f.assign(결산일=end, 가능일=end + pd.DateOffset(months=k))
    pub = pd.to_datetime(f["FP56000900"].astype("Int64").astype(str), format="%Y%m%d", errors="coerce")
    pub = pub.where(pub > end)
    하루 = pd.Timedelta(days=0 if kind == "발표당일" else 1)  # '발표당일': 발표한 날부터(03 진단용)
    if kind == "발표월말":  # NH(2018) 02 p.14·03 p.18 '분기/연간 재무제표가 발표되는 월의 월말에 반영': 발표한 날부터 → 월말 리밸런싱이면 그 달 마지막 거래일까지 발표분을 그 달 말에 씀.
        # 가능일 = 발표일(달 안 비교는 쓰는 쪽이 함). 02·03은 발표재무()가 t가 든 달력 달 발표분까지 씀(마지막 거래일 뒤 휴일 발표분 포함 = 최대 2일 미래 정보, V03 HIGH-1 2026-11-06)
        return f.assign(결산일=end, 발표일=pub, 가능일=pub.fillna(end + pd.Timedelta(days=k)))
    ok = end + pd.DateOffset(months=k) if kind == "결산후개월" else (pub + 하루).fillna(end + pd.Timedelta(days=k))
    return f.assign(결산일=end, 발표일=pub, 가능일=ok)


def 발표지연(키="CP10002200"):
    """12월 결산 연간 보고서의 결산일→실적보고서발표일 일수(발표일 > 결산일만)와 발표일 있는 비율. '발표일' 규칙의 k를 고른 근거.
    키="CP10002200": 결산월 이력이 12인 회계연도. 키="결산월": 정적 결산월 12(결산월을 바꾼 회사도 12월 31일로 계산 → 꼬리가 길어짐)."""
    f = 읽기.연간재무()
    f = f[(f[키] == 12) & f["M000901001"].notna()]  # 자산총계가 있는 행 = 보고서가 있는 회사-연도
    end = pd.to_datetime(f["회계연도"].astype(int).astype(str) + "1231")
    pub = pd.to_datetime(f["FP56000900"].astype("Int64").astype(str), format="%Y%m%d", errors="coerce")
    d = (pub - end).dt.days.where(pub > end)
    구간 = pd.cut(f["회계연도"], [1986, 1994, 1999, 2004, 2010, 2026], labels=["1987~94", "1995~99", "2000~04", "2005~10", "2011~26"])
    표 = lambda x: pd.Series({"행": len(x), "발표일 있음": x.notna().mean(), "건수": x.count()} | ({} if x.count() == 0 else {
        "중앙값(일)": x.median(), "95%(일)": x.quantile(0.95), "99%(일)": x.quantile(0.99),
        **{f"{k}일 초과": (x > k).sum() / x.count() for k in (90, 150, 180)}}))
    return pd.concat({**{g: 표(x) for g, x in d.groupby(구간, observed=False)}, "전체": 표(d)}, axis=1).T


def 재무(t, how, 연도=None):
    """그날 쓸 수 있는 가장 최근 회계연도 행(종목마다 한 행). 연도=None: 결산일이 2년보다 오래된 행은 안 씀(묵은 값 막기, ponytail).
    연도='직전년도': 결산일이 t 전 해(t.year − 1)인 행만(우동호 p.512 '직전년도', Novy-Marx 각주 4 '다음 해 6월 말'). 연도='최근': 상한 없이 가장 최근."""
    f = 재무가능일(how)
    f = f[f["가능일"] <= t]
    if 연도 == "직전년도":
        f = f[f["결산일"].dt.year == t.year - 1]
    elif 연도 is None:
        f = f[f["결산일"] >= t - pd.DateOffset(years=2)]
    return f.sort_values(["회계연도", "결산일"]).drop_duplicates("코드", keep="last").set_index("코드")  # F9: 같은 회계연도 두 결산월이면 늦은 결산일


@cache
def 분기가능일(시점="다음날"):
    """분기 행마다 쓸 수 있는 날 = 분기 보고서 발표일(FP56000900 분기) 다음 날. 발표일이 없거나 분기말보다 이르면 쓰지 않는다(분기는 대신할 일수 근거가 없어 빈칸).
    시점="발표월말": 발표한 날부터 = NH(2018) '발표되는 월의 월말에 반영'(월말 리밸런싱이면 그 달 마지막 거래일까지 발표분이 그 달 말에 들어감. 그 뒤 휴일 발표분은 다음 리밸런싱, 2026-11-05)."""
    f = 읽기.분기재무()
    pub = pd.to_datetime(f["FP56000900"].astype("Int64").astype(str), format="%Y%m%d", errors="coerce") if "FP56000900" in f else pd.Series(pd.NaT, f.index)
    pub = pub.where(pub > f["분기말"])
    return f.assign(발표일=pub, 가능일=pub if 시점 == "발표월말" else pub + pd.Timedelta(days=1))


def 분기재무(t, 흐름="4분기합", 시점="다음날"):
    """그날 쓸 수 있는 분기 행으로 종목마다 한 줄(열 = 아이템코드, 천원). NH(2018) p.14·p.18 '분기/연간 재무제표가 발표되는 월의 월말에 반영'.
    흐름 항목(읽기.분기흐름): '4분기합' = 쓸 수 있는 가장 최근 4개 분기(연속: 첫 분기말과 마지막 분기말이 9개월 차이) 합, '최근분기x4' = 최근 분기 × 4.
    두 방식 모두 원문에 없음(민감도). 잔액 항목(읽기.분기잔액) = 최근 분기말 값. 분기말이 2년보다 오래된 행은 안 씀(연간과 같음)."""
    f = 분기가능일(시점)
    f = f[(f["가능일"] <= t) & (f["분기말"] >= t - pd.DateOffset(years=2))].sort_values(["코드", "분기말"])
    끝 = f.groupby("코드").tail(1).set_index("코드")
    흐름열 = [c for c in 읽기.분기흐름 if c in f]
    if 흐름 == "4분기합":
        f4 = f.groupby("코드").tail(4)
        n, 처음 = f4.groupby("코드")["분기말"].count(), f4.groupby("코드")["분기말"].min()
        연속 = (n == 4) & ((끝["분기말"].dt.to_period("M") - 처음.dt.to_period("M")).map(lambda x: x.n) == 9)
        v = f4.groupby("코드")[흐름열].sum(min_count=4).where(연속, np.nan)
    else:
        v = 끝[흐름열] * 4
    out = 끝.drop(columns=흐름열).join(v)
    return out.assign(회계연도=np.nan)  # 지표.값의 '빈칸으로 볼 칸'(회계연도 기준)은 분기에 안 씀


@cache
def 발표행():
    """보고서마다 한 행: 분기 보고서(1·2·3분기, 분기 3개월 값) + 연간 사업보고서(결산 값). 각 행의 가능일 = 그 보고서 발표일(FP56000900).
    결산월 분기는 분기 행 대신 연간 행(사업보고서가 발표하는 값). 발표일이 없거나 기간 끝보다 이른 보고서는 안 씀(대신할 일수 근거 없음)."""
    q = 분기가능일("발표월말")
    q = q[q["분기말"].dt.month != q["CP10002200"]].dropna(subset=["가능일"])
    y = 재무가능일(("발표월말", 0)).dropna(subset=["발표일"]).rename(columns={"결산일": "분기말"})
    cols = [c for c in 읽기.분기흐름 + 읽기.분기잔액 if c in q and c in y]
    return pd.concat([q[["코드", "분기말", "가능일", "CP10002200", *cols]].assign(보고서="분기"),
                      y[["코드", "분기말", "발표일", "CP10002200", "회계연도", *cols]].assign(가능일=lambda d: d["발표일"], 보고서="연간")], ignore_index=True)


def 발표재무(t, 상한=None):
    """NH(2018) 02 p.14·03 p.18 '주요 재무제표 수치는 분기/연간 재무제표가 발표되는 월의 월말에 반영', 02 p.26 '확정 재무데이터':
    t까지 발표된 가장 최근 보고서(분기 또는 연간) 값 그대로(4분기 합·×4 같은 가공 없음, 회원 결정 2026-11-06). 같은 날 발표면 기간 끝이 늦은 보고서.
    단 영업CF·EBITDA 분기 3개월 값은 DataGuide가 누적값에서 환산·계산한 값(V02 MEDIUM-2).
    상한 = 년 수를 주면 기간 끝이 그보다 오래된 보고서는 안 씀(기본 없음).
    '발표되는 월의 월말' = t가 든 달력 달에 발표된 것까지(마지막 거래일 뒤 같은 달 휴일 발표분 포함, 원문 문장대로 V03 HIGH-1 2026-11-06).
    그 휴일 발표분은 t 종가에는 아직 몰랐던 값 = 최대 2일 미래 정보(1990~2026 전체 1,749행, R1 재감사; 한계, _검증/기본값_목록_20261106.md)."""
    월말 = t + pd.offsets.MonthEnd(0)
    assert (월말 - t).days <= 10, f"발표재무는 월말 리밸런싱일에만: {t:%Y-%m-%d}"  # 달 중간 t면 그 달 뒤쪽 발표분이 새어 들어감
    f = 발표행()
    f = f[f["가능일"] <= 월말]
    if 상한:
        f = f[f["분기말"] >= t - pd.DateOffset(years=상한)]
    return f.sort_values(["가능일", "분기말"]).drop_duplicates("코드", keep="last").set_index("코드")


def 재무표(t, cfg):
    """설정의 재무주기대로 그날 쓸 재무(종목마다 한 행). ('발표값',) = 발표재무, ('분기', 흐름) = 분기재무, 그 밖 = 연간 재무()."""
    kind = cfg.get("재무주기", ("연간",))
    if kind[0] == "발표값":
        return 발표재무(t, cfg.get("재무상한"))
    if kind[0] == "분기":
        return 분기재무(t, kind[1], "발표월말" if cfg["재무시점"][0] == "발표월말" else "다음날")
    return 재무(t, cfg["재무시점"], cfg.get("재무연도"))


def 다음리밸런싱(t, cfg):
    g = 기준일(cfg)
    return g[min(g.get_loc(t) + cfg["리밸런싱"], len(g) - 1)] if t in g else t


def 재료(t, cfg):
    """기준일 t의 자료: 그날 상장 가격(읽기.상장가격) 있는 종목, 재무, 시가총액, 직전 12개월 일간수익률(빈칸은 빈칸, '쓰지 말 것' 칸은 빈칸으로 봄).
    다음 = 다음 리밸런싱일, 일년상장 = 직전 12개월 내내 시장구분 KS·KQ이고 그 창 첫날 상장 가격이 있음
    (시장구분이 첫 가격보다 몇 년 먼저 KS로 들어 있는 우선주 등이 짧은 수익률로 통과하던 것 막기, 감사 2026-11-03 #3·#14)."""
    p = 읽기.상장가격(cfg["가격"]).loc[:t]
    codes = p.columns[p.iloc[-1].notna()]
    a = t - pd.DateOffset(years=1)
    p = p.loc[a:, codes].copy()
    for c, x, y in 읽기.쓰지말것(cfg["가격"])[["코드", "시작", "끝"]].itertuples(index=False):
        if c in p:
            p.loc[x:y, c] = np.nan
    return dict(t=t, 다음=다음리밸런싱(t, cfg), 가격=cfg["가격"], codes=codes, fin=재무표(t, cfg).reindex(codes),
                mcap=읽기.일간(읽기.시총).loc[t, codes], ret=p.pct_change(fill_method=None).iloc[1:],
                일년상장=읽기.상장중().loc[a:t].reindex(columns=codes, fill_value=False).all() & p.iloc[0].notna())


def 거르기(t, cfg):
    """재료 → 시장 → '쓰지 말 것' → 설정의 제외 순서로 거른다. (재료, 남은 종목, 단계별 수)."""
    s = 재료(t, cfg)
    return (s, *종목거르기.적용(s, [("시장", {"시장들": cfg["시장"]}), "쓰지말것"] + cfg["제외"]))


def 점수(v, 지표들, how):
    """v: 종목×지표 값. 높을수록 좋은 점수를 돌려준다. 지표 하나라도 빈 종목은 빠진다."""
    x = v.dropna() * [1 if m[1] == "높게" else -1 for m in 지표들]
    if how == "순위합산":
        return -x.rank(ascending=False).sum(axis=1)
    if how == "z합산":
        return ((x - x.mean()) / x.std()).sum(axis=1)
    for m, _, keep in 지표들:  # 단계별: 앞 지표부터 상위 keep만 남기고, 마지막 지표로 줄 세움
        x = x[x[m] >= x[m].quantile(1 - keep)]
    return x[m]


def 고르기(sc, how):
    kind, n = how
    if kind == "상위N":
        return {f"상위{n}": sc.nlargest(n).index}
    if kind == "상하N":  # 위 n(Q1)·가운데(Q2)·아래 n(Q3): Novy-Marx(2010) p.21 '150 stocks with the highest … 150 … lowest'
        a = sc.sort_values(ascending=False).index
        return {"Q1": a[:n], "Q2": a[n:len(a) - n], "Q3": a[len(a) - n:]}
    if len(sc) < n:  # 0종목이거나 분위 수보다 적으면 빈 묶음(구간 수익률 빈칸, 감사 #11; 02 '최근 분기 × 4' 2002-03-29에 1종목뿐 → 5분위 못 나눔, 2026-11-04)
        return {f"Q{i}": sc.index[:0] for i in range(1, n + 1)}
    q = pd.qcut(sc.rank(method="first", ascending=False), n, labels=[f"Q{i}" for i in range(1, n + 1)])
    return {g: sc.index[q == g] for g in q.cat.categories}


# 비중: (종목, 재료, 설정) → 합 1인 Series
def 동일(c, s, cfg):
    return pd.Series(1 / len(c), c)


def 시총(c, s, cfg):
    m = s["mcap"][c]
    return m / m.sum()


def 시총상한(c, s, cfg):
    cap = cfg["비중상한"]
    if cap * len(c) < 1:  # 종목 수 × 상한 < 1이면 상한을 지킬 수 없음 → 빈 비중(그 구간 수익률 빈칸)
        return pd.Series(dtype=float)
    w = 시총(c, s, cfg)
    while w.max() > cap + 1e-12:
        over = w >= cap
        w[over] = cap
        w[~over] *= (1 - cap * over.sum()) / w[~over].sum()
    return w


def 이력(c, s):
    """위험 비중용 직전 12개월 수익률. 12개월 내내 상장된 종목만(지표.변동성과 같은 기준)."""
    r = s["ret"][c]
    return r.loc[:, 지표.일년상장(s)[c].to_numpy()]


def 위험균형(c, s, cfg):
    sd = 이력(c, s).std()
    iv = 1 / sd.where(sd > 0)  # 변동성 0(가격이 1년 내내 같음)은 역수가 없어 빈칸 → 비중에서 빠짐(감사 #10)
    return iv / iv.sum()


def 시총상한10(c, s, cfg):
    return 시총상한(c, s, {**cfg, "비중상한": 0.1})


def 시총상한30(c, s, cfg):
    return 시총상한(c, s, {**cfg, "비중상한": 0.3})


최소분산기록 = []  # 최소분산을 풀 때마다 (기준일, 종목 수, SLSQP 수렴 여부, 메시지). README 한계 '최소분산 수렴 미확인'(2026-11-04)


def 최소분산(c, s, cfg, 개월=36):
    """공매도 없는 최소분산. 직전 36개월 월수익률(설정의 기준일 간격) 공분산을 Ledoit-Wolf로 수축.
    36개월 수익률이 다 있는 종목만 쓴다(나머지는 빠짐 — 편입 수는 보유 비중 > 0 종목 수로 확인)."""
    g = 기준일({**cfg, "기간": (None, None)})
    g = g[g <= s["t"]][-(개월 + 1):]
    r = 쓰지말것칸(읽기.상장가격(cfg["가격"]).loc[g, c].pct_change(fill_method=None), cfg["가격"]).iloc[1:].dropna(axis=1)  # '쓰지 말 것' 칸은 R과 같게(README 한계 '최소분산 공분산')
    if len(g) <= 개월 or r.shape[1] < 2:
        return pd.Series(dtype=float)
    S, n = LedoitWolf().fit(r.to_numpy()).covariance_, r.shape[1]
    o = minimize(lambda w: w @ S @ w, np.full(n, 1 / n), jac=lambda w: 2 * S @ w, bounds=[(0, 1)] * n,
                 constraints={"type": "eq", "fun": lambda w: w.sum() - 1}, method="SLSQP")
    최소분산기록.append(dict(기준일=s["t"], 종목=n, 수렴=bool(o.success), 메시지=o.message))
    w = o.x.clip(0)
    return pd.Series(w / w.sum(), r.columns)


def 거래비용(old, new, c):
    """비중 변화로 매수·매도 금액(포트폴리오 대비 비율)을 구해 비용률을 돌려준다."""
    d = new.sub(old, fill_value=0)
    buy, sell = d.clip(lower=0).sum(), -d.clip(upper=0).sum()
    return buy * (c["수수료"] + c["슬리피지"]) + sell * (c["수수료"] + c["거래세"] + c["슬리피지"])


def 기간수익률(p, grid):
    """기준일 간 수익률. 구간 안에서 상장폐지되면 마지막 가격으로 판 것으로 본다(가격 값은 바꾸지 않고 마지막 값을 찾기만 함)."""
    return p.ffill().loc[grid] / p.loc[grid].shift(1) - 1


def 쓰지말것칸(R, item):
    """기준일 간 수익률 R(행 = 구간 끝)에서 '쓰지 말 것' 구간(시작, 끝]에 걸린 칸만 바꾼다(README_데이터 권고 D-06·U3·D27):
    TR 이상칸은 그 구간 가격수익률(수정주가)로, 병합·감자 미반영(D27)·A008080(D-05)은 빈칸(그 종목·달을 뺌). 다른 칸은 그대로."""
    R, g = R.copy(), R.index
    prev = pd.DatetimeIndex([pd.NaT, *g[:-1]])
    b = 읽기.쓰지말것(item)
    for c, x, y, k in b[b["코드"].isin(R.columns)][["코드", "시작", "끝", "근거"]].itertuples(index=False):
        hit = (g > x) & (prev < y)
        if k.startswith("TR이상칸"):  # 걸린 종목 한 열만 계산(전 종목 ffill을 최소분산마다 다시 해 느렸음, 지시 H 2026-11-04). 열마다 계산이라 값은 같음
            R.loc[hit, c] = 기간수익률(읽기.상장가격(읽기.수정주가)[c], g).loc[hit]
        else:
            R.loc[hit, c] = np.nan
    return R


def 보유(R, W, 비용, 전량=False, 회전=None):
    """R: 기준일 간 수익률(행=구간 끝 기준일). W: {리밸런싱일: 비중}. 리밸런싱 사이엔 매수보유(비중이 값 따라 변함).
    수익률이 빈 종목(폐지 다음 구간부터)은 그달 계산에서 빠진다 = 남은 종목에 값 비례로 다시 넣은 것과 같다. 비용은 리밸런싱 뒤 첫 달에 뺀다."""
    out, old, dates = [], pd.Series(dtype=float), sorted(W)
    for t0, t1 in zip(dates, dates[1:] + [R.index[-1]]):
        w = W[t0]
        r = R.loc[(R.index > t0) & (R.index <= t1), w.index]
        g = (1 + r.fillna(0)).cumprod()
        v = g.shift(1, fill_value=1) * w
        pr = (1 + (v * r).sum(axis=1) / v.where(r.notna()).sum(axis=1)) * (1 - 비용.get("월비용", 0)) - 1
        c = 거래비용(old, w[:0], 비용) + 거래비용(w[:0], w, 비용) if 전량 else 거래비용(old, w, 비용)
        if 회전 is not None:
            회전[t0] = w.sub(old, fill_value=0).clip(lower=0).sum()  # 한쪽 회전율 = 산 비중 합(처음 매수 = 1). 전량교체 비용과 별개
        i = pr.first_valid_index()  # 첫 달이 빈칸(보유 종목 모두 값 없음)이면 값 있는 첫 달에 뺀다(한계 15번, 2026-11-04). 구간 내내 빈칸이면 비용도 없음
        if i is not None:
            pr.loc[i] = (1 - c) * (1 + pr.loc[i]) - 1
        살아있음 = r.iloc[-1].notna()  # 마지막 달 수익률이 빈 종목(폐지 등)은 pr에서 이미 빠져 남은 종목에 다시 넣은 셈 → 다음 매도비용·회전율에서 뺀다(한계 21·40번, 2026-11-04)
        old = (g.iloc[-1] * w)[살아있음] / (g.iloc[-1] * w)[살아있음].sum()
        out.append(pr)
    return pd.concat(out)


def 실행(cfg):
    """설정대로 돌려 dict(수익률=월수익률 표(열=(비중, 그룹)), 기록=리밸런싱일별 종목 수, 보유=비중, 설정).
    cfg["전략들"] = {이름: dict(지표=..., 합산=...)}을 주면 같은 거르기 결과로 여러 전략을 한 번에 고른다(열 = (이름, 비중, 그룹),
    ('유니버스', '시총', '전체') = 거른 종목 전부 시가총액 가중 = 확장 연구 자체 벤치마크)."""
    cfg = {**기본설정, **cfg}
    전략들 = cfg.get("전략들") or {None: dict(지표=cfg["지표"], 합산=cfg["합산"])}
    grid = 기준일(cfg)
    R = 쓰지말것칸(기간수익률(읽기.상장가격(cfg["가격"]), grid), cfg["가격"])
    W, 기록 = defaultdict(dict), []
    for t in grid[:-1][::cfg["리밸런싱"]]:
        s, c, n = 거르기(t, cfg)
        v = pd.DataFrame({m: getattr(지표, m)(s).reindex(c) for m in dict.fromkeys(m[0] for e in 전략들.values() for m in e["지표"])})
        v = v.mask(np.isinf(v))  # 분모 0 → ±inf는 값이 없는 것과 같아 빈칸으로 본다(inf가 1등으로 줄 서는 것 막기)
        if None not in 전략들:
            m = s["mcap"].reindex(c).dropna()
            W[("유니버스", "시총", "전체")][t] = m / m.sum()
        편입 = {}
        for 이름, e in 전략들.items():
            sc = 점수(v[[m[0] for m in e["지표"]]], e["지표"], e["합산"])
            groups = 고르기(sc, cfg["고르기"])
            for b in cfg["비중"]:
                for g, cc in groups.items():
                    k = (b, g) if 이름 is None else (이름, b, g)
                    W[k][t] = globals()[b](cc, s, cfg) if len(cc) else pd.Series(dtype=float)  # 0종목이면 그 구간 수익률 빈칸
                    편입[":".join(["보유수", *map(str, k)])] = int((W[k][t] > 0).sum())  # 위험균형·최소분산은 과거 수익률이 모자란 종목이 빠져 편입 수보다 적을 수 있음(감사 #2)
            편입 |= {"빈칸으로 빠짐" if 이름 is None else f"빈칸으로 빠짐:{이름}": len(c) - len(v[[m[0] for m in e["지표"]]].dropna())}  # 단계별에서 남길 비율로 잘린 수는 넣지 않음(감사 #12)
            편입 |= {f"편입:{g}" if 이름 is None else f"편입:{이름}:{g}": len(cc) for g, cc in groups.items()}
        기록.append({"기준일": t, **n, **{f"빈칸:{k}": x for k, x in v.isna().sum().items()}, **편입,
                   "확인:업종빈칸": 종목거르기.그날(읽기.업종, t, c).isna().sum() if 읽기.있음(읽기.업종) else pd.NA,  # 업종을 안 쓰는 02·03이 업종 파일 때문에 멈추지 않게(감사 #13)
                   "확인:WICS금융": 종목거르기.그날(읽기.WICS, t, c).eq("금융").sum() if 읽기.있음(읽기.WICS) else pd.NA})  # 금융업 판정은 거래소 업종, WICS는 확인용(2000년 이후만 있음, 파일 없으면 빈칸 = 확인 안 함)
    회전 = defaultdict(dict)
    수익률 = pd.DataFrame({k: 보유(R, w, cfg["거래비용"], cfg["전량교체"], 회전[k]) for k, w in W.items()})
    return dict(수익률=수익률, 기록=pd.DataFrame(기록).set_index("기준일"), 보유=W, 설정=cfg, 종목수익률=R, 회전율=pd.DataFrame(회전))



def 큰수익률(R, 하한=3):
    """기준일 간 종목 수익률이 하한 초과(기본 3 = +300%, 이전 보고의 '한 달 3배 초과')인 종목-월. 원본 그대로 공개용(고치지 않음)."""
    x = R.stack()
    return x[x > 하한].rename("수익률")


def 이상값보유(결과, 하한=3):
    """열(비중, 그룹)마다 큰수익률 종목-월을 보유한 건수. 0보다 크면 그 열은 '이상값 영향'."""
    big, out = 큰수익률(결과["종목수익률"], 하한), {}
    for k, W in 결과["보유"].items():
        d = pd.DatetimeIndex(sorted(W))
        out[k] = sum(c in W[d[d < m][-1]].index for m, c in big.index if (d < m).any() and m <= 결과["수익률"].index[-1])
    return pd.Series(out, name="이상값 보유 건수")
```

```python
%%모듈 평가 평가
"""성과 평가: 누적·연환산 수익률, 표준편차, 샤프, MDD(최대낙폭), 정보비율, CAPM·FF3·Carhart 알파(t는 OLS. 2026-11-06 회원 결정으로 Newey-West t를 뺌, 원문에 없었음).
'초과평균'은 설명변수 없는 회귀 = 월평균 초과수익률과 그 t.
타겟 페이퍼 성과표 지표(서브에이전트가 원문 PDF로 확인, 진행기록 2026-11-03 04:50): 우동호 외(2023) 표 2~7(p.514–516) = 누적·MDD·표준편차·샤프·정보비율·젠센·3요인·4요인 α(t)
→ 위 칸들. Novy-Marx(NBER w15940, 2010년 4월판) 표 = 월평균 초과수익률·α·요인 적재값(β)·adj.-R² → '초과평균 α(월)'·'Carhart β:요인'·'Carhart adj R²' 칸."""
import shutil  # (제출본에서 더함) git 있는지 보기
import subprocess
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import statsmodels.api as sm

import 데이터읽기 as 읽기

plt.rcParams.update({"font.family": "Malgun Gothic", "axes.unicode_minus": False})

모형 = {"초과평균": [], "CAPM": ["MKT"], "FF3": ["MKT", "SMB", "HML"], "Carhart": ["MKT", "SMB", "HML", "MOM"]}


def 요인(grid, cfg):
    """기준일 간 요인 수익률(DataGuide 종가지수 비율). RF = 구간 시작 시점 금리/12. MKT = 시장지수 − RF."""
    r = 읽기.지수().reindex(grid).pct_change(fill_method=None)
    rf = 읽기.무위험(cfg["무위험"]).reindex(grid, method="ffill").shift(1)
    시장 = r[cfg["시장지수"]]
    return pd.DataFrame({"시장": 시장, "RF": rf, "MKT": 시장 - rf, "SMB": r["3FM.2B3.SMB"],
                         "HML": r["3FM.2B3.HML"], "MOM": r["3FM.2M3.MOM"]})


def 회귀(y, X):
    """알파(월)와 t(OLS), OLS 결과(β·adj R²용)."""
    d = X.assign(_y=y).dropna()
    Xc = sm.add_constant(d.drop(columns="_y"), has_constant="add")
    ols = sm.OLS(d["_y"], Xc).fit()
    if ols.rsquared > 1 - 1e-10:  # 시장 자기 자신 회귀(잔차 ≈ 0): t는 계산 오차라 빈칸(결과 감사 LOW 2026-11-03)
        return ols.params["const"], np.nan, ols
    return ols.params["const"], ols.tvalues["const"], ols


def 성과(r, F, 롱숏=False):
    """r: 월수익률. 롱숏이면 초과수익 = r 그대로(무비용 포트폴리오), 아니면 r − RF."""
    r = r.dropna()
    F = F.loc[r.index]
    ex = r if 롱숏 else r - F["RF"]
    w = (1 + r).cumprod()
    a = r - F["시장"]
    out = {"개월": len(r), "누적": w.iloc[-1] - 1, "연환산": w.iloc[-1] ** (12 / len(r)) - 1,
           "표준편차(연)": r.std() * np.sqrt(12), "샤프": ex.mean() / ex.std() * np.sqrt(12),
           "MDD": (w / w.cummax().clip(lower=1) - 1).min(), "정보비율": a.mean() / a.std() * np.sqrt(12) if a.std() else np.nan}  # 시장 자신은 a = 0
    for name, cols in 모형.items():
        out[f"{name} α(월)"], out[f"{name} t(OLS)"], ols = 회귀(ex, F[cols])
    out |= {f"Carhart β:{c}": ols.params[c] for c in cols} | {"Carhart adj R²": ols.rsquared_adj}  # 마지막 모형(Carhart)의 적재값
    return out


def 샤프차이(a, b):
    """두 월수익률의 (평균/표준편차) 차이 검정. Jobson·Korkie(1981) 식을 Memmel(2003)이 고친 식(두 원문 확인 안 함):
    z = (SR_a − SR_b) / √{[2(1−ρ) + ½(SR_a² + SR_b² − 2·SR_a·SR_b·ρ²)] / T}. 같은 달만 쓴다. p = 양측 정규."""
    from scipy.stats import norm
    d = pd.concat([a, b], axis=1).dropna()
    x, y = d.iloc[:, 0], d.iloc[:, 1]
    sa, sb, r, T = x.mean() / x.std(), y.mean() / y.std(), x.corr(y), len(d)
    z = (sa - sb) / np.sqrt((2 * (1 - r) + 0.5 * (sa ** 2 + sb ** 2 - 2 * sa * sb * r ** 2)) / T)
    return pd.Series({"월수": T, "a": sa, "b": sb, "차이": sa - sb, "z": z, "p": 2 * norm.sf(abs(z))})


def 평균차이(a, b):
    """두 월수익률 차이(a − b)의 평균과 t(OLS: 상수항만 있는 회귀의 보통 t), p(양측 정규)."""
    from scipy.stats import norm
    d = (a - b).dropna()
    m, t, _ = 회귀(d, pd.DataFrame(index=d.index))
    return pd.Series({"월수": len(d), "평균 차이(월)": m, "t(OLS)": t, "p": 2 * norm.sf(abs(t))})


def 성과표(R, F, 롱숏열=()):
    """R: 열마다 월수익률. 표 한 장(행=전략)."""
    return pd.DataFrame({k: 성과(R[k], F, k in 롱숏열) for k in R}).T


def 누적그림(R, 제목, 로그=True, 퍼센트=False, 오른쪽=None, 시작100=False):
    """누적수익률. 흑백으로도 구분되게 선 모양·표식을 다르게. 세로축은 로그(README 한계 '선형 축', 2026-11-04).
    퍼센트=True: 원문 그림처럼 (누적 가치 − 1) × 100 %, 보통 축. 시작100=True면 누적 가치 × 100 %(우동호 Fig. 1: 세 선이 약 100%에서 시작, 원문사양_01). 오른쪽 = {이름: 지수 값 Series}를 주면 오른쪽 축에 그 값을 그대로 그림(NH 그림 26·32 'KOSPI(우)')."""
    fig, ax = plt.subplots(figsize=(9, 4.5))
    styles = [(ls, mk) for mk in (None, "o", "s", "^", "x") for ls in ("-", "--", ":", "-.")]  # 20가지(선 9개 그림에서 둘이 겹치던 것, 결과 감사 LOW)
    for (ls, mk), k in zip(styles * (len(R.columns) // len(styles) + 1), R):  # 20개 넘으면 모양 반복(열은 빠지지 않음)
        w = (1 + R[k].dropna()).cumprod()
        ax.plot((w if 시작100 else w - 1) * 100 if 퍼센트 else w, color="black" if ls == "-" else "dimgray", ls=ls, marker=mk, markevery=6, ms=4, lw=1.2, label=str(k))
    if 퍼센트:
        ax.set(title=제목, ylabel="누적 가치(%, 시작=100)" if 시작100 else "누적 수익률(%)")
    else:
        ax.set(title=제목, ylabel="누적 가치(시작=1, 로그 축)" if 로그 else "누적 가치(시작=1)", yscale="log" if 로그 else "linear")
    h, l = ax.get_legend_handles_labels()
    if 오른쪽:
        ax2 = ax.twinx()
        for k, v in 오른쪽.items():
            ax2.plot(v, color="gray", lw=0.8, alpha=0.7, label=f"{k}(우)")
        ax2.set_ylabel("지수(우)")
        h2, l2 = ax2.get_legend_handles_labels()
        h, l = h + h2, l + l2
    ax.legend(h, l, fontsize=8, ncol=2)
    ax.grid(alpha=0.3)
    return fig


def 일간정의(r, 시장, 무위험일):
    """우동호 외(2023) 표 2~7의 SD·Sharpe·IR·MDD를 원문 숫자가 나오는 정의로(원문에 정의 없음, KOSPI 원문 숫자로 역산 `_검증/01_정의역산.py`):
    일간 수익률 r, SD = 일 표준편차 × √N(N = 전 기간 일 수), Sharpe = 일 초과수익 평균/표준편차 × √N, IR = (r − 시장) 평균/표준편차 × √N, MDD = 일간 가치의 최대 낙폭.
    무위험일 = 그날 쓸 일 무위험수익률(직전 날 값을 씀)."""
    r = r.dropna()
    N, w = len(r), (1 + r).cumprod()
    e = r - 무위험일.reindex(r.index)
    a = r - 시장.reindex(r.index)
    return {"일 수": N, "누적": w.iloc[-1] - 1, "MDD": (w / w.cummax().clip(lower=1) - 1).min(), "SD": r.std() * np.sqrt(N),
            "Sharpe": e.mean() / e.std() * np.sqrt(N), "IR": a.mean() / a.std() * np.sqrt(N) if a.std() else 0.0}


결과폴더 = Path("결과") / "결과_월수익률"  # (제출본에서 바꿈) 노트북 옆 결과 폴더
구간들 = {"전체": (None, None), "1990-04~2004-03": ("1990-04", "2004-03"), "2004-04~2017-03": ("2004-04", "2017-03"), "2017-04~": ("2017-04", None)}


def 구간(r, a, b):
    """보유 월로 자른다. 기준일 간 수익률의 보유 월 = (구간 끝 − 15일)의 달(첫거래일·마지막거래일 기준일 모두 맞음)."""
    m = (r.index - pd.Timedelta(days=15)).to_period("M")
    return r[(m >= pd.Period(a or "1900-01", "M")) & (m <= pd.Period(b or "2999-12", "M"))]


def 저장(R, F, 이름, 롱숏=(), 이상값=None):
    """요약표용: 월수익률(열 = (전략, 가중))과 요인, 이상값 = {열: 포.이상값보유 건수}, 만든 코드의 git 커밋(커밋 안 된 변경 파일 목록 포함)을 결과_월수익률/이름.pkl로 남긴다."""
    결과폴더.mkdir(exist_ok=True)
    git = lambda *a: subprocess.run(["git", *a], cwd=결과폴더.parent, capture_output=True, text=True, encoding="utf-8").stdout.strip() if shutil.which("git") else ""  # 한글 파일 이름(utf-8)을 cp949로 읽다 깨지지 않게. (제출본에서 더함) git이 없는 컴퓨터면 빈칸(기록용 칸, 계산 안 씀)
    pd.to_pickle(dict(R=R, F=F, 롱숏=list(롱숏), 이상값=dict(이상값 or {}), 커밋=git("rev-parse", "HEAD"), 변경=git("status", "--porcelain", "--", "공통", "따라하기", "연구")), 결과폴더 / f"{이름}.pkl")


def 요약(최소개월=24):
    """결과_월수익률의 모든 전략 × 가중 × 기간: 평균(월 초과수익)과 t(OLS), 샤프, MDD, Carhart 알파와 t(OLS), 이상값 보유(전 기간 건수, 저장 때 안 준 파일은 빈칸). 최소개월 미만 구간은 뺀다."""
    rows = {}
    for f in sorted(결과폴더.glob("*.pkl")):
        d = pd.read_pickle(f)
        if d["F"] is None:  # 따라하기 02·03은 원문에 없는 요인을 안 만듦(05:47-3) → 옮김 노트북이 결과_월수익률/요인/<같은 이름>.pkl 에 저장
            d["F"] = pd.read_pickle(결과폴더 / "요인" / f.name)
        for k, r in d["R"].items():
            for g, (a, b) in 구간들.items():
                x = 구간(r.dropna(), a, b)
                if len(x) >= 최소개월:
                    o = 성과(x, d["F"], k in d["롱숏"])
                    rows[(f.stem, *k, g)] = {"보유 첫 달": f"{x.index[0] - pd.Timedelta(days=15):%Y-%m}", "보유 끝 달": f"{x.index[-1] - pd.Timedelta(days=15):%Y-%m}",
                                             "개월": o["개월"], "평균(월 초과)": o["초과평균 α(월)"], "평균 t(OLS)": o["초과평균 t(OLS)"], "샤프": o["샤프"],
                                             "MDD": o["MDD"], "Carhart α(월)": o["Carhart α(월)"], "α t(OLS)": o["Carhart t(OLS)"],
                                             "이상값 보유": d.get("이상값", {}).get(k, np.nan)}
    return pd.DataFrame(rows).T.rename_axis(["파일", "전략", "가중", "기간"])
```

```python
%%모듈 점검 점검
"""계산 점검(손으로 답을 아는 작은 예). 실행: python 점검.py → 'ok'."""
import numpy as np
import pandas as pd

import 포트폴리오 as 포

d = pd.to_datetime(["2020-01-31", "2020-02-28", "2020-03-31", "2020-04-30"])
R = pd.DataFrame({"A": [np.nan, 0.10, 0.10, np.nan], "B": [np.nan, -0.10, np.nan, 0.20]}, index=d)
c0 = dict(수수료=0, 거래세=0, 슬리피지=0)

# 매수보유: 2월 (0.1−0.1)/2 = 0. 3월 B 빈칸 → A만: 0.1. 4월 A 빈칸 → B만(값 0.45 그대로 유지): 0.2.
r = 포.보유(R, {d[0]: pd.Series(0.5, ["A", "B"])}, c0)
assert np.allclose(r, [0.0, 0.1, 0.2]), r

# 비용: 처음 매수 100% × (수수료+슬리피지). 매도 땐 거래세까지.
assert np.isclose(포.거래비용(pd.Series(dtype=float), pd.Series(1.0, ["A"]), dict(수수료=0.001, 거래세=0.01, 슬리피지=0.002)), 0.003)
assert np.isclose(포.거래비용(pd.Series(1.0, ["A"]), pd.Series(1.0, ["B"]), dict(수수료=0.001, 거래세=0.01, 슬리피지=0.002)), 0.003 + 0.013)

# 전량교체: 같은 종목이어도 전부 팔고 다시 삼 → 매도(수수료+거래세+슬리피지) + 매수(수수료+슬리피지)
R1 = pd.DataFrame({"A": [np.nan, 0.0, 0.0]}, index=d[:3])
W1 = {d[0]: pd.Series(1.0, ["A"]), d[1]: pd.Series(1.0, ["A"])}
c1 = dict(수수료=0.001, 거래세=0.01, 슬리피지=0.002)
assert np.isclose(포.보유(R1, W1, c1, 전량=True).iloc[1], -(0.013 + 0.003))

# 폐지 종목 몫엔 매도비용 없음(한계 21·40번): A·B 반반, 2월 B 폐지(빈칸) → 3월 리밸런싱(A 100%)에서 팔 것은 A뿐 → 전량교체 비용 = A 매도 1 × 0.013 + 매수 1 × 0.003
R2 = pd.DataFrame({"A": [np.nan, 0.0, 0.0, 0.0], "B": [np.nan, np.nan, np.nan, np.nan]}, index=d)
assert np.isclose(포.보유(R2, {d[0]: pd.Series(0.5, ["A", "B"]), d[2]: pd.Series(1.0, ["A"])}, c1, 전량=True).loc[d[3]], -(0.013 + 0.003))
# 첫 달 빈칸이면 비용은 값 있는 첫 달에(한계 15번): 2월 A 빈칸, 3월 0 → 3월에 매수비용 0.003
R3 = pd.DataFrame({"A": [np.nan, np.nan, 0.0]}, index=d[:3])
r = 포.보유(R3, {d[0]: pd.Series(1.0, ["A"])}, c1)
assert np.isnan(r.loc[d[1]]) and np.isclose(r.loc[d[2]], -0.003), r

# 월비용: 수익 0인 달마다 −0.2%
assert np.allclose(포.보유(R1, {d[0]: pd.Series(1.0, ["A"])}, dict(수수료=0, 거래세=0, 슬리피지=0, 월비용=0.002)), -0.002)

# 상장폐지 구간: 마지막 가격(11)으로 정산 → 0.1, 그다음 구간은 빈칸
p = pd.DataFrame({"A": [10, 11, np.nan, np.nan]}, index=pd.to_datetime(["2020-01-31", "2020-02-10", "2020-02-28", "2020-03-31"]))
r = 포.기간수익률(p, p.index[[0, 2, 3]])["A"]
assert np.isclose(r.iloc[1], 0.1) and np.isnan(r.iloc[2]), r

# 시총상한: 합 1, 최대 = 상한
w = 포.시총상한(pd.Index(list("abcdefghij") + ["k"]), {"mcap": pd.Series([100] + [1] * 10, list("abcdefghij") + ["k"])}, {"비중상한": 0.2})
assert np.isclose(w.sum(), 1) and np.isclose(w.max(), 0.2), w

# 고르기: 분위 Q1 = 점수 가장 높은 쪽
g = 포.고르기(pd.Series(range(10), list("abcdefghij")), ("분위", 5))
assert list(g["Q1"]) == ["i", "j"] and list(g["Q5"]) == ["a", "b"], g
# 종목이 분위 수보다 적으면 모든 분위가 빈 묶음(감사 2 H1)
g = 포.고르기(pd.Series([1.0, 2.0, 3.0], list("abc")), ("분위", 5))
assert list(g) == ["Q1", "Q2", "Q3", "Q4", "Q5"] and all(len(x) == 0 for x in g.values()), g

# 순위합산: 둘 다 1등인 종목이 최고
v = pd.DataFrame({"x": [3, 2, 1], "y": [1, 2, 3]}, index=["p", "q", "r"])
assert 포.점수(v, [("x", "높게"), ("y", "낮게")], "순위합산").idxmax() == "p"

# '쓰지 말 것' 구간: A008080 2004-01~2013-09에 기준일이 들면 뺀다(다음 리밸런싱일은 안 봄 = 미래 정보 안 씀)
import 종목거르기, 지표
c = pd.Index(["A008080", "A005930"])
ts = lambda a, b: dict(가격="수정주가", t=pd.Timestamp(a), 다음=pd.Timestamp(b))
assert list(종목거르기.쓰지말것(ts("2013-04-01", "2014-04-01"), c)) == ["A005930"]
assert list(종목거르기.쓰지말것(ts("2013-10-01", "2014-10-01"), c)) == ["A008080", "A005930"]
assert list(종목거르기.쓰지말것(ts("2003-12-01", "2004-12-01"), c)) == ["A008080", "A005930"]  # 구간이 보유 중에 시작 → 거르지 않음
# 보유 중 걸린 칸만 빈칸(D-05): 구간 끝 9-30에 걸리는 9월·10월 구간만
m = pd.to_datetime(["2013-08-01", "2013-09-02", "2013-10-01", "2013-11-01"])
q = 포.쓰지말것칸(pd.DataFrame({"A008080": 0.1, "A005930": 0.1}, index=m), "수정주가")
assert q["A008080"].isna().tolist() == [False, True, True, False] and q["A005930"].notna().all(), q

# 빈칸으로 볼 칸: 호텔신라 FY2023 매출원가만 빈칸, FY2024와 다른 항목은 그대로
f = pd.DataFrame({"회계연도": [2023, 2023], "M000905001": [0.0, 5.0], "M000904001": [9.0, 9.0]}, index=["A008770", "A000001"])
assert 지표.값({"fin": f}, "M000905001").isna().tolist() == [True, False] and 지표.값({"fin": f}, "M000904001").notna().all()
assert 지표.값({"fin": f.assign(회계연도=2024)}, "M000905001").notna().all()

# 큰수익률: 기본 하한 3(+300%) 초과만
assert 포.큰수익률(pd.DataFrame({"A": [3.0, 3.5]}, index=d[:2])).index.tolist() == [(d[1], "A")]

# 회전율: A 100% → B 100%면 한쪽 회전율 1, 처음 매수도 1
h = {}
포.보유(R1.assign(B=0.0), {d[0]: pd.Series(1.0, ["A"]), d[1]: pd.Series(1.0, ["B"])}, c0, 회전=h)
assert np.allclose(list(h.values()), [1, 1]), h

# 보유 월로 자르기: 첫거래일 기준 (2017-03-02, 2017-04-03] = 3월 보유, 월말 기준 (2017-03-31, 2017-04-28] = 4월 보유
import 평가
r = pd.Series(1.0, pd.to_datetime(["2004-05-03", "2017-04-03", "2017-05-02"]))
assert list(평가.구간(r, "2004-04", "2017-03").index.strftime("%Y-%m-%d")) == ["2004-05-03", "2017-04-03"]
assert list(평가.구간(pd.Series(1.0, pd.to_datetime(["2017-03-31", "2017-04-28"])), "2017-04", None).index.month) == [4]

# 빈 비중(0종목) 구간은 수익률 빈칸
assert 포.보유(R1, {d[0]: pd.Series(dtype=float)}, c0).isna().all()
assert len(평가.구간(r, None, None)) == 3
# 일간(달력기준) 재무 → 회계연도 값(읽기.연간재무): 작은 가짜 파일로. A = 12월 결산, B = 3월 결산, C = 3월 결산인데 값이 1월에 바뀜(믿을 수 없어 빈칸)
import tempfile
from pathlib import Path
import 데이터읽기 as 읽기
dd = pd.bdate_range("2009-01-01", "2011-12-31")
y = lambda a, b, c: pd.Series(np.select([dd <= a, dd <= b], [1.0, 2.0], 3.0), dd)
fin = pd.DataFrame({"A": y("2009-12-31", "2010-12-31", None), "B": y("2010-03-31", "2011-03-31", None), "C": y("2009-12-31", "2010-12-31", None)})
ym = pd.DataFrame({"A": 12.0, "B": 3.0, "C": 3.0}, index=dd)
옛 = 읽기.DATA, 읽기.공통끝, 읽기.변환캐시
with tempfile.TemporaryDirectory() as tmp:
    읽기.DATA, 읽기.공통끝, 읽기.변환캐시 = Path(tmp), lambda: dd[-1], Path(tmp)  # 시험 폴더만 보게(저장소 캐시의 실제 파일 빼기)
    ym.rename_axis("날짜").to_parquet(Path(tmp) / "통합_일간_CP10002200_결산월.parquet")
    fin.rename_axis("날짜").to_parquet(Path(tmp) / "통합_일간_M000901001_자산.parquet")
    a = 읽기.연간재무().set_index(["코드", "회계연도"])["M000901001"]
    읽기.DATA, 읽기.공통끝, 읽기.변환캐시 = 옛
    읽기.일간.cache_clear(); 읽기.연간재무.cache_clear()
assert a[("A", 2009)] == 1 and a[("A", 2010)] == 2 and a[("A", 2011)] == 3, a
assert a[("B", 2008)] == 1 and a[("B", 2009)] == 1 and a[("B", 2010)] == 2, a  # 3월 결산: 2009-03 결산 = FY2008(첫 행은 데이터 전이라 확인 못 하고 그대로), 2011-03 결산 = FY2010
assert np.isnan(a[("C", 2009)]) and np.isnan(a[("C", 2010)]) and 읽기.검사["M000901001"] == 2, (a, 읽기.검사)
# 분기재무: 4분기 합은 쓸 수 있는 연속 4분기만, 잔액은 최근 분기말 값, 발표일 다음 날부터
q = pd.DataFrame({"코드": ["A"] * 5 + ["B"] * 3, "분기말": pd.to_datetime(["2019-12-31", "2020-03-31", "2020-06-30", "2020-09-30", "2020-12-31", "2020-03-31", "2020-09-30", "2020-12-31"]),
                  "M000908004": [9, 1, 2, 3, 4, 1, 2, 3], "M000903003": [90, 10, 20, 30, 40, 10, 20, 30]})
q["가능일"] = q["분기말"] + pd.Timedelta(days=40)
원래분기 = 포.분기가능일
포.분기가능일 = lambda 시점="다음날": q  # 분기재무가 시점을 넘김(2026-11-05)
z = 포.분기재무(pd.Timestamp("2021-03-01"))
assert z.loc["A", "M000908004"] == 1 + 2 + 3 + 4 and z.loc["A", "M000903003"] == 40 and np.isnan(z.loc["B", "M000908004"]), z  # B는 2020-06 분기가 없어 연속 아님
assert 포.분기재무(pd.Timestamp("2021-01-15")).loc["A", "M000908004"] == 9 + 1 + 2 + 3  # 2020-12 분기는 아직 발표 전
assert 포.분기재무(pd.Timestamp("2021-03-01"), "최근분기x4").loc["A", "M000908004"] == 16
포.분기가능일 = 원래분기
# 재무시점 '발표월말'(2026-11-05): 가능일 = 발표일 → 월말 기준일이면 그 달 마지막 거래일까지 발표분만. 토요일 2007-03-31 발표분은 2007-03-30(금) 리밸런싱에 안 씀
m, d = 포.분기가능일("발표월말"), 포.분기가능일()
assert (m["가능일"].dropna() == m["발표일"].dropna()).all() and (d["가능일"].dropna() == d["발표일"].dropna() + pd.Timedelta(days=1)).all()
sat = m[m["발표일"] == pd.Timestamp("2007-03-31")]
assert len(sat) > 0 and (sat["가능일"] > pd.Timestamp("2007-03-30")).all(), len(sat)
y = 포.재무가능일(("발표월말", 150)).dropna(subset=["발표일"])
assert (y["가능일"] == y["발표일"]).all()
print("ok")
```

## 2. 엑셀 → parquet 다시 만들기 (기본은 끔)
3~8절은 `data\통합_전체_202611`의 parquet(엑셀을 빠르게 읽으려고 바꾼 파일)를 읽는다. 그 parquet는 `data\DataGuide_전체_202611`의 DataGuide 원본 엑셀에서 아래 함수(데이터를 받은 원래 data 폴더의 감독 변환기 `DataGuide_추가_202611\_tools\_to_parquet.py`의 `labels`·`read_rows`·`to_frame`, 글자 그대로. 변환기 파일은 제출 data에 없음)로 만들었다.
- `엑셀_변환 = True`로 바꾸면 원본 엑셀마다 parquet를 다시 만들어 노트북 옆 `다시만든_parquet` 폴더에 쓰고(data 폴더는 안 건드림), data의 parquet와 값이 칸마다 같은지(`DataFrame.equals`: 날짜·종목·값·빈칸 위치·형식 모두) 찍는다. 엑셀이 커서(파일당 0.2~0.4GB) 파일당 수 분, 전체 수 시간 걸린다.
- 원래 변환기는 저장 뒤 독립 검사기(xlsx XML 직접 읽기)로 한 번 더 대조했다(PASS인 파일만 씀, 기록은 원래 data 폴더의 `DataGuide_추가_202611\_tools\_audit`, 제출 data에는 없음). 여기서는 다시 만든 것을 data의 parquet와 값으로 대조한다.
- 읽는 규칙(값은 원본 그대로): '코드' 줄이 있는 시트 중 줄이 가장 많은 시트, 코드 줄 빈칸·중복·날짜 중복·역순·숫자 글자 섞임이면 멈춤, 마지막 날짜 뒤 날짜 없는 줄의 값은 빼고 알림(R24·R25 A015890 한 칸).

```python
%%절 2
엑셀_변환 = False  # True로 바꾸면 원본 엑셀에서 parquet를 다시 만들어 data의 parquet와 같은지 확인(오래 걸림)
import datetime as dt  # 아래 to_frame이 날짜 칸 형식을 볼 때 씀

# ↓ 감독 변환기 _to_parquet.py의 읽기 함수 세 개(글자 그대로). 원본 엑셀(행 = 날짜, 열 = 종목)을 표로 읽는다.
def labels(rows):
    return {str(r[0]).strip(): i for i, r in enumerate(rows[:30]) if r and r[0] not in (None, "")}


def read_rows(f):
    wb = CalamineWorkbook.from_path(str(f))
    best = None
    for name in wb.sheet_names:
        rows = wb.get_sheet_by_name(name).to_python(skip_empty_area=False)
        if "코드" in labels(rows) and (best is None or len(rows) > len(best)):
            best = rows
    assert best is not None, "'코드' 줄이 있는 시트 없음"
    return best


def to_frame(rows):
    lab = labels(rows)
    head = rows[lab["코드"]][1:]
    n = len(head)
    while n and head[n - 1] in (None, ""):
        n -= 1                                              # 오른쪽 끝 빈칸만 잘라냄
    codes = head[:n]
    assert all(c not in (None, "") for c in codes), "코드 줄 가운데 빈칸 → 열 밀림 위험, 멈춤"
    assert len(codes) == len(set(codes)), "코드 중복"
    items = set(rows[lab["아이템코드"]][1:n + 1])
    assert len(items) == 1, f"아이템코드가 하나가 아님: {items}"
    start = max(lab[k] for k in ("코드", "아이템코드", "아이템명", "집계주기") if k in lab) + 1
    body = rows[start:]
    last = max(i for i, r in enumerate(body) if r and r[0] not in (None, ""))
    for r in body[:last]:                                   # 날짜 줄 사이에 날짜 없는 줄의 값이 있으면 멈춤
        if r and r[0] in (None, ""):
            assert all(v in (None, "") for v in r[1:]), "날짜 줄 사이 빈 날짜 줄에 값 있음"
    tail = {codes[j - 1]: sum(1 for r in body[last + 1:] if j < len(r) and r[j] not in (None, ""))  # 마지막 날짜 뒤 날짜 없는 줄의 값: 날짜가 없어 시계열에 넣을 수 없음 → 빼고 기록
            for j in range(1, n + 1) if any(j < len(r) and r[j] not in (None, "") for r in body[last + 1:])}
    if tail:
        print(f"주의: 마지막 날짜 뒤 날짜 없는 줄에 값 {tail} (종목: 값 칸 수) → parquet에서 뺌", flush=True)
    data = [r for r in body if r and r[0] not in (None, "")]
    assert all(isinstance(r[0], (dt.date, dt.datetime)) for r in data), "날짜 칸이 날짜 형식이 아님(일련번호 등)"
    dates = pd.to_datetime([r[0].date() if isinstance(r[0], dt.datetime) else r[0] for r in data])
    assert dates.is_unique and dates.is_monotonic_increasing, "날짜 중복 또는 역순"
    vals = [[(None if v == "" else v) for v in (list(r[1:n + 1]) + [None] * (n + 1 - len(r)))] for r in data]
    kinds = {type(v) for row in vals for v in row if v is not None}
    if kinds <= {int, float}:
        df = pd.DataFrame(vals, index=pd.Index(dates, name="날짜"), columns=codes, dtype="float64")
    elif kinds <= {str}:
        df = pd.DataFrame(vals, index=pd.Index(dates, name="날짜"), columns=codes).astype("string")
    else:
        raise AssertionError(f"숫자·글자 섞임 {kinds} → 멈춤")
    return items.pop(), df


if 엑셀_변환:
    from python_calamine import CalamineWorkbook  # 엑셀을 빨리 읽는 패키지(read_rows가 씀). 이 칸을 켤 때만 필요
    다시 = Path("다시만든_parquet")
    다시.mkdir(exist_ok=True)
    for f in sorted((DATA / "DataGuide_전체_202611").glob("*_I00P_일간_*.xlsx")):  # 일간 아이템 엑셀(X01~X04는 3~8절이 엑셀을 바로 읽음)
        name = f.name.split("_I00P_")[0]  # 예: R08_부채총계_M000902001_결산 (원래 변환기와 같은 이름 규칙)
        (old,) = list((DATA / "통합_전체_202611").glob(f"통합_일간_*_{name}.parquet"))
        t0 = time.time()
        item, df = to_frame(read_rows(f))
        new = 다시 / f"통합_일간_{item}_{name}.parquet"
        df.to_parquet(new)
        같음 = pd.read_parquet(new).equals(pd.read_parquet(old))
        print(f"{'같음' if 같음 else '다름!'} {new.name}: {df.shape}, {time.time() - t0:.0f}초", flush=True)
else:
    print("엑셀 → parquet 다시 만들기: 꺼짐(엑셀_변환 = False). 3~8절은 data\\통합_전체_202611의 parquet를 읽음")
```

줄임말: EW(동일가중, Equal Weight), VW(시가총액가중, Value Weight), LS(롱숏: 좋은 분위를 사고 나쁜 분위를 판 수익률),
MDD(최대 낙폭, Maximum Drawdown), IR(정보비율, Information Ratio), CAPM(자본자산가격결정모형), FF3(파마-프렌치 3요인), Carhart(카하트 4요인: FF3 + 모멘텀),
OLS(보통최소제곱).

**원문에 없는 추가는 이 노트북에서 뺐다**(2026-11-06 회원 결정 "원문에 말이 없으면 빼. 원문에 있으면 원문대로 해."): VW, 확인판(발표일 + 150일), 5분위·LS, 월간 × √12 성과 칸, 월 누적 대 일간 누적 표, 단계별 표·이상값·기록 표, 원문 차이 진단 → `연구/따라하기에서_옮김_01.ipynb`(맨 앞에 옮긴 목록).

### 3. 설정
원문대로 '4월 첫 거래일, 직전 연도 결산자료'. 12월 결산이면 결산일 + 3개월(3월 말)이 지나면 쓰고(발표일은 보지 않음), 회계연도 t−1 결산만 쓴다(`재무연도="직전년도"`, 원문 p.512 '직전년도에 공시된 재무제표'). 설정 = `포.우동호2023_원문`(금융업 = 원문 7업종 판정, 투자기구 거르기 없음). 아래 칸이 설정에 이 값이 들어 있는지 확인한다(보통주만·수정주가 포함).

```python
%%절 3
IMF = {"이름": "우동호2023_IMF", "지표": [("GPA", "높게"), ("EY", "높게")]}
설정 = {"MF": {**포.우동호2023_원문, "기간": 기간, "끝": "마지막거래일"}, "IMF": {**포.우동호2023_원문, **IMF, "기간": 기간, "끝": "마지막거래일"}}
for k, v in 설정.items():  # 원문 p.512 '직전년도', '수정주가', 금융 7업종. 보통주만(회원 결정 2026-11-06). 투자기구 거르기 없음(원문에 없음, 회원 05:57)
    assert v["재무연도"] == "직전년도" and v["가격"] == 읽기.수정주가 and "보통주" in v["제외"] and v["비중"] == ["동일"], k
    assert ("금융업", {"이름": "원문"}) in v["제외"] and "투자기구" not in v["제외"], k
pd.DataFrame(설정).astype(str)
```

### 4. 본문: 상위 30종목 EW (원문 표 2)
4월 첫 거래일마다 거르기 → 지표 계산 → 순위 합산 → 상위 30종목을 동일비중으로 사서 1년 보유한다(원문 p.512).
- **요인**: 과제 공지대로 DataGuide 요인을 쓴다(`평가.요인`: SMB = 3FM.2B3.SMB, HML = 3FM.2B3.HML, MOM = 3FM.2M3.MOM). 원문 p.513은 SMB = 시가총액 하위 30% − 상위 30%, HML = 장부가치/시장가치 상위 30% − 하위 30%, MOM = 과거 11개월 수익률 상위 30% − 하위 30%로 직접 구성해, 원문 3·4요인 α와 정의가 다르다. 시장 요인 = KOSPI − 무위험(국고 3년).
- **누적·알파**는 월수익률(원문 p.514 '월별로 집계'), t는 OLS.
- **SD·Sharpe·IR·MDD**는 같은 보유 비중을 일간 수익률에 넣어 잰다(`평가.일간정의`: 일 표준편차 × √N, 샤프 = 일 초과수익 평균/표준편차 × √N, IR = KOSPI 대비 같은 꼴, MDD = 일간 가치 최대 낙폭. 원문에 정의가 없고 KOSPI 원문 숫자가 이 정의에서 나옴, 9절 KOSPI 줄).
- KOSPI 행의 알파·t는 시장을 자기 자신에 회귀한 값이라 뜻이 없어 빈칸으로 둔다(원문 표도 빈칸).

```python
%%절 3
결과 = {k: 포.실행(v) for k, v in 설정.items()}
g = 포.기준일(설정["MF"])
F = 평가.요인(g, 설정["MF"])  # 시장 = KOSPI(I.001), RF = 국고 3년(E11.02.003.008) 연 %/1200 직전 달 값, MKT = 시장 − RF. SMB·HML·MOM = DataGuide 요인(과제 공지)
R = pd.DataFrame({k: o["수익률"][("동일", "상위30")] for k, o in 결과.items()})
R["KOSPI"] = F["시장"]
R = R.dropna(how="all")

# 일간: 같은 보유 비중을 일간 수익률(월간 실행과 같은 '쓰지 말 것' 규칙)에 넣는다
가격 = 설정["MF"]["가격"]
날 = 읽기.상장가격(가격).loc[g[0]:g[-1]].index
Rd = 포.쓰지말것칸(포.기간수익률(읽기.상장가격(가격), 날), 가격)
국고 = 읽기.국고3년일간()  # 원문 p.513 '3년 만기 국고채 수익률'(연 %)
무위험일 = 국고.reindex(날.union(국고.index)).ffill().reindex(날).shift(1) / 100 / 365  # 직전 날 값
시장일 = 읽기.지수()[설정["MF"]["시장지수"]].reindex(날).pct_change(fill_method=None).iloc[1:]
일 = {k: 포.보유(Rd, o["보유"][("동일", "상위30")], 설정[k]["거래비용"], 설정[k]["전량교체"]) for k, o in 결과.items()} | {"KOSPI": 시장일}

열 = ["Return", "MDD", "SD", "Sharpe", "IR", "Jensen α", "Jensen t", "3F α", "3F t", "4F α", "4F t"]


def 원문꼴(R월, R일):
    """원문 표 2~7 칸: 누적·알파·t(OLS)는 월수익률, SD·Sharpe·IR·MDD는 일간."""
    월 = 평가.성과표(R월, F)
    일표 = pd.DataFrame({k: 평가.일간정의(R일[k], 시장일, 무위험일) for k in R월}).T
    return pd.DataFrame({"Return": 월["누적"], "MDD": 일표["MDD"], "SD": 일표["SD"], "Sharpe": 일표["Sharpe"], "IR": 일표["IR"],
                         "Jensen α": 월["CAPM α(월)"], "Jensen t": 월["CAPM t(OLS)"], "3F α": 월["FF3 α(월)"], "3F t": 월["FF3 t(OLS)"],
                         "4F α": 월["Carhart α(월)"], "4F t": 월["Carhart t(OLS)"]}).astype(float)


표 = 원문꼴(R, 일)
표.loc["KOSPI", ["Jensen α", "Jensen t", "3F α", "3F t", "4F α", "4F t"]] = float("nan")  # 시장을 자기 자신에 회귀한 값(수치 잡음)이라 빈칸(원문 표도 빈칸)
print("월 수", int(R["MF"].notna().sum()), "/ 일 수", len(일["MF"].dropna()))
표
```

### 5. 누적수익률 그림 (원문 Fig. 1)
원문 Fig. 1처럼 보통(선형) 축, 선은 MF·IMF·KOSPI 셋(p.513). 원문 세 선이 2004-04에 약 100%에서 시작해 세로값 = 누적 가치(1 + 누적수익률) × 100%로 그린다(`_검증/1106/원문사양_01.md` A3). 첫 점은 2004-04 첫 거래일 100%. 선 모양은 원문 범례대로 IMF 실선, MF 파선, KOSPI 점선. 로그 축 그림은 확장연구(`연구/따라하기에서_옮김_01.ipynb`).

```python
%%절 3
그림R = pd.concat([pd.DataFrame(0.0, index=[g[0]], columns=["IMF", "MF", "KOSPI"]), R[["IMF", "MF", "KOSPI"]]])  # 첫 점 = 2004-04 첫 거래일 100%
평가.누적그림(그림R, "우동호 외(2023) Fig. 1 재현: 상위 30종목 EW", 퍼센트=True, 시작100=True);  # 열 순서 = 선 모양: IMF 실선, MF 파선, KOSPI 점선(원문 범례 '——IMF', '-----MF', '······KOSPI')
```

### 6. 시가총액 5분위별 (원문 표 3~7)
4월 첫 거래일에 본문 거르기를 한 뒤, 그 전략의 지표 자료가 있는 종목(MF = ROC·EY, IMF = GPA·EY; 원문 '표본', p.512·513)만 4월 첫 거래일 시가총액으로 5분위로 나누고(1분위 = 가장 작음, 같은 수), 분위마다 따로 상위 30종목을 고른다(원문 p.514 '시가총액 하위 20%' 등). 분위를 나누는 모집단은 원문에 없어 기본값(11절). 다른 모집단 판은 확장연구.
월 누적과 일간 누적이 같은지 보는 표(원문에 없음)는 확장연구(`연구/따라하기에서_옮김_01.ipynb` 15절)로 옮겼다.

```python
%%절 3
# 분위 모집단 = 그 전략 지표 자료가 있는 종목(원문 '표본', p.512·513). 2026-11-05 2차 진단 1
규모결과 = {f"{k}_시총{q}분위": 포.실행({**설정[k], "제외": 설정[k]["제외"] + [("지표있음", {"지표들": [m[0] for m in 설정[k]["지표"]]}), ("시총분위", {"분위": q})]}) for q in range(1, 6) for k in ["MF", "IMF"]}
규모 = pd.DataFrame({n: o["수익률"][("동일", "상위30")] for n, o in 규모결과.items()})
규모일 = {n: 포.보유(Rd, o["보유"][("동일", "상위30")], o["설정"]["거래비용"], o["설정"]["전량교체"]) for n, o in 규모결과.items()}
규모표 = 원문꼴(규모, 규모일)
규모표
```

### 7. 원문 vs 우리 (원문 표 2~7)
원문 숫자는 표 2~7(p.514–516)을 옮겨 적은 것이고, 우리 숫자는 위 4·6절 표에서 코드가 가져온다. 누적·MDD·표준편차는 소수(980.35% → 9.8035). 두 칸의 단위는 같다(SD·Sharpe·IR·MDD 모두 일간 정의, 4절).

**원문 판정**(두 번째 표, `01_원문판정.csv`): 원문 본문 글(p.514–517)이 말한 방향·크기만 대조한다. '유의'는 원문 별표 수준을 그대로 OLS t에 쓴다: 1% = |t| ≥ 2.58, 5% = 1.96, 10% = 1.645(원문 표 5 MF 젠센 0.0049*(1.68)을 p.515가 '10%수준에서 유의'라고 씀). '유의하지 않다' = |t| < 1.645. 원문이 말하지 않은 p 기준 판정(경계 등)은 하지 않는다. KOSPI 대비 문장(p.515–516 4분위 '좋은 성과를 나타내지 못하거나 오히려 나쁜 성과', 개선된 마법공식 '여전히 기존 마법공식 및 벤치마크 보다 좋은 성과', p.516–517 1·2분위 '누적수익률, 샤프비율, 정보비율 등 모든 면에서 월등')은 방향만 본다(비교한 숫자는 '원문 값'·'우리 값' 칸, KOSPI = 같은 기간 표 2 줄).

```python
%%절 3
원문표 = pd.DataFrame({  # 우동호 외(2023) 표 2~7
    ("전체", "MF"): [9.8035, -0.5034, 0.9479, 2.42, 1.48, 0.0068, 2.32, -0.0050, -1.28, -0.0053, -1.34],
    ("전체", "IMF"): [10.2791, -0.4962, 0.8857, 2.58, 1.57, 0.0071, 2.47, -0.0028, -0.73, -0.0034, -0.87],
    ("전체", "KOSPI"): [2.1239, -0.5454, 0.8349, 1.34, 0.00] + [None] * 6,
    ("시총1분위", "MF"): [44.2427, -0.5760, 1.1082, 3.52, 2.72, 0.0137, 3.88, 0.0139, 3.51, 0.0141, 3.55],
    ("시총1분위", "IMF"): [101.3052, -0.6203, 1.1828, 4.06, 3.31, 0.0179, 4.39, 0.0167, 3.62, 0.0168, 3.64],
    ("시총2분위", "MF"): [17.6391, -0.5155, 1.0596, 2.78, 1.94, 0.0095, 2.76, 0.0097, 2.70, 0.0096, 2.69],
    ("시총2분위", "IMF"): [28.4832, -0.5343, 1.1032, 3.13, 2.32, 0.0120, 3.13, 0.0127, 3.25, 0.0126, 3.24],
    ("시총3분위", "MF"): [6.3007, -0.4963, 0.9663, 1.99, 1.07, 0.0049, 1.68, 0.0044, 1.46, 0.0044, 1.45],
    ("시총3분위", "IMF"): [9.8866, -0.4799, 0.9275, 2.47, 1.51, 0.0068, 2.38, 0.0066, 2.22, 0.0066, 2.21],
    ("시총4분위", "MF"): [2.0623, -0.5528, 0.9325, 1.10, 0.14, 0.0009, 0.33, 0.0005, 0.18, 0.0004, 0.13],
    ("시총4분위", "IMF"): [2.7631, -0.5583, 0.9242, 1.32, 0.35, 0.0020, 0.68, 0.0031, 0.99, 0.0028, 0.90],
    ("시총5분위", "MF"): [3.3954, -0.4398, 0.7890, 1.59, 0.47, 0.0022, 1.13, 0.0004, 0.20, -0.0003, -0.15],
    ("시총5분위", "IMF"): [3.7548, -0.3496, 0.6953, 1.82, 0.54, 0.0028, 1.47, 0.0018, 0.86, 0.0011, 0.57],
}, index=열).T
우리 = {("전체", k): 표.loc[k] for k in ["MF", "IMF", "KOSPI"]}
우리 |= {(f"시총{q}분위", k): 규모표.loc[f"{k}_시총{q}분위"] for q in range(1, 6) for k in ["MF", "IMF"]}
우리표 = pd.DataFrame(우리).T[열].astype(float)
print("우리 = 본문 설정 / 기간", 기간, "/ 월 수", int(R["MF"].notna().sum()), "/ 일 수", len(일["MF"].dropna()))
대조 = pd.concat({"원문": 원문표, "우리": 우리표}, axis=1).swaplevel(axis=1)[열]
대조.to_csv(평가.결과폴더 / "01_원문대조.csv", encoding="utf-8-sig")  # 결과요약 노트북이 읽음
display(대조)

# 원문 판정: (구간, 전략, 대조 칸, 원문 방향, 원문 유의, 원문 쪽·글)
수준 = {"1%": 2.58, "5%": 1.96, "10%": 1.645}
주장 = [("전체", k, a, "+", "5%", "p.514 '젠센의 알파의 계수 값은 … 5% 수준에서 유의한 (+)'") for k in ["MF", "IMF"] for a in ["Jensen"]]
주장 += [("전체", k, a, "−", "아님", "p.514 '3요인 모형과 4요인 모형에서 알파의 계수는 (-)의 값 … 통계적으로 유의하지 않았다'") for k in ["MF", "IMF"] for a in ["3F", "4F"]]
주장 += [(f"시총{q}분위", k, a, "+", "1%", f"p.{514 if q == 1 else 515} {q}분위 '1%수준에서 통계적으로 유의한 (+)'") for q in [1, 2] for k in ["MF", "IMF"] for a in ["Jensen", "3F", "4F"]]
주장 += [("시총3분위", "MF", "Jensen", "+", "10%", "p.515 '0.0049로서 10%수준에서 통계적으로 유의한 (+)'")]
주장 += [("시총3분위", "MF", a, "+", "아님", "p.515 '모두 0.0044로 (+)의 값을 나타냈지만 통계적으로 유의하진 않았다'") for a in ["3F", "4F"]]
주장 += [("시총3분위", "IMF", a, "+", "5%", "p.515 '5%수준에서 통계적으로 유의한 (+)', '5% 수준에서 통계적으로 유의하였다'") for a in ["Jensen", "3F", "4F"]]
주장 += [("시총4분위", k, a, "+", "아님", "p.515 '(+)의 값을 나타냈지만 통계적으로 유의하진 않았다'") for k in ["MF", "IMF"] for a in ["Jensen", "3F", "4F"]]
주장 += [("시총5분위", "MF", a, s, "아님", "p.516 '각각 0.0022, 0.0004로 (+) … 4요인 … –0.0003로 오히려 (-) … 모두 통계적으로 유의하지도 않았다'") for a, s in [("Jensen", "+"), ("3F", "+"), ("4F", "−")]]
주장 += [("시총5분위", "IMF", a, "+", "아님", "p.516 '(+)의 값을 나타냈지만 통계적으로 유의하진 않았다'") for a in ["Jensen", "3F", "4F"]]
판정 = []
for 구간, k, a, 방향, 유의, 글 in 주장:
    α, t = 우리표.loc[(구간, k), f"{a} α"], 우리표.loc[(구간, k), f"{a} t"]
    우리방향 = "+" if α > 0 else "−"
    우리유의 = next((n for n, x in 수준.items() if abs(t) >= x), "아님")
    맞음 = 우리방향 == 방향 and (abs(t) < 수준["10%"] if 유의 == "아님" else abs(t) >= 수준[유의])
    판정.append({"구간": 구간, "전략": k, "항목": f"{a} α", "원문 글": 글, "원문 방향": 방향, "원문 유의": 유의,
               "원문 α": 원문표.loc[(구간, k), f"{a} α"], "원문 t": 원문표.loc[(구간, k), f"{a} t"], "우리 α": α, "우리 t(OLS)": t,
               "우리 방향": 우리방향, "우리 유의(가장 강한 수준)": 우리유의, "원문과 같음": "같음" if 맞음 else "다름"})
for 구간 in ["전체"] + [f"시총{q}분위" for q in range(1, 6)]:  # p.514·515·517 '개선된 마법공식은 … 모두에서 기존의 마법공식보다 뛰어난 성과'
    a, b = 우리표.loc[(구간, "IMF"), "Return"], 우리표.loc[(구간, "MF"), "Return"]
    판정.append({"구간": 구간, "전략": "IMF vs MF", "항목": "Return(누적)", "원문 글": "p.517 '전체종목 … 시가총액 1분위에서 5분위까지 … 모두에서 기존의 마법공식보다 뛰어난 성과'",
               "원문 방향": "IMF > MF", "원문 α": 원문표.loc[(구간, "IMF"), "Return"], "원문 t": 원문표.loc[(구간, "MF"), "Return"], "우리 α": a, "우리 t(OLS)": b,
               "우리 방향": "IMF > MF" if a > b else "IMF ≤ MF", "원문과 같음": "같음" if a > b else "다름"})
# p.515–517 KOSPI 대비 문장(방향만). 비교한 숫자는 '원문 값'·'우리 값' 칸에 글로(KOSPI = 같은 기간 표 2 줄)
K = {"원문": 원문표.loc[("전체", "KOSPI")], "우리": 우리표.loc[("전체", "KOSPI")]}
값 = lambda 쪽, 구간, k, 칸들: ", ".join(f"{c} {(원문표 if 쪽 == '원문' else 우리표).loc[(구간, k), c]:.4f} vs KOSPI {K[쪽][c]:.4f}" for c in 칸들)
for k in ["MF", "IMF"]:  # 4분위: 누적·샤프로 보아 KOSPI보다 좋지 않음(원문 IMF는 누적은 KOSPI보다 크고 샤프 1.32 < 1.34 → '못하거나'를 '둘 다 앞서지는 않음'으로 읽음)
    좋음 = all(우리표.loc[("시총4분위", k), c] > K["우리"][c] for c in ["Return", "Sharpe"])
    판정.append({"구간": "시총4분위", "전략": f"{k} vs KOSPI", "항목": "Return·Sharpe",
               "원문 글": "p.515–516 '누적수익률과 샤프비율 등으로 확인하였을 때 4분위에서의 마법공식 및 개선된 마법공식은 벤치마크인 KOSPI 대비 좋은 성과를 나타내지 못하거나 오히려 나쁜 성과'",
               "원문 방향": "KOSPI보다 좋지 않음(누적·샤프 둘 다 앞서지는 않음)", "원문 값": 값("원문", "시총4분위", k, ["Return", "Sharpe"]), "우리 값": 값("우리", "시총4분위", k, ["Return", "Sharpe"]),
               "우리 방향": "누적·샤프 둘 다 KOSPI보다 큼" if 좋음 else "KOSPI보다 좋지 않음", "원문과 같음": "다름" if 좋음 else "같음"})
q4 = {n: 우리표.loc[("시총4분위", n), "Return"] for n in ["IMF", "MF"]} | {"KOSPI": K["우리"]["Return"]}
판정.append({"구간": "시총4분위", "전략": "IMF vs MF·KOSPI", "항목": "Return(누적)",
           "원문 글": "p.516 '하지만 개선된 마법공식은 여전히 기존 마법공식 및 벤치마크 보다 좋은 성과를 나타냈다'(4분위. 원문 표 6 IMF 샤프 1.32 < KOSPI 1.34라 누적으로 읽음)",
           "원문 방향": "IMF > MF, IMF > KOSPI", "원문 값": f"IMF {원문표.loc[('시총4분위', 'IMF'), 'Return']:.4f}, MF {원문표.loc[('시총4분위', 'MF'), 'Return']:.4f}, KOSPI {K['원문']['Return']:.4f}",
           "우리 값": ", ".join(f"{n} {v:.4f}" for n, v in q4.items()),
           "우리 방향": "IMF > MF, IMF > KOSPI" if q4["IMF"] > max(q4["MF"], q4["KOSPI"]) else "아님", "원문과 같음": "같음" if q4["IMF"] > max(q4["MF"], q4["KOSPI"]) else "다름"})
for 구간 in ["시총1분위", "시총2분위"]:  # p.516–517 1·2분위 'KOSPI 대비 누적수익률, 샤프비율, 정보비율 등 모든 면에서 월등하게 좋은 투자 성과'(정보비율은 KOSPI 대비라 KOSPI 줄 = 0)
    for k in ["MF", "IMF"]:
        좋음 = all(우리표.loc[(구간, k), c] > K["우리"][c] for c in ["Return", "Sharpe", "IR"])
        판정.append({"구간": 구간, "전략": f"{k} vs KOSPI", "항목": "Return·Sharpe·IR",
                   "원문 글": "p.516–517 '시가총액 하위 1분위 및 2분위, 즉 소형주들에서는 KOSPI 대비 누적수익률, 샤프비율, 정보비율 등 모든 면에서 월등하게 좋은 투자 성과'",
                   "원문 방향": "KOSPI보다 셋 다 큼", "원문 값": 값("원문", 구간, k, ["Return", "Sharpe", "IR"]), "우리 값": 값("우리", 구간, k, ["Return", "Sharpe", "IR"]),
                   "우리 방향": "KOSPI보다 셋 다 큼" if 좋음 else "셋 중 하나 이상 KOSPI 이하", "원문과 같음": "같음" if 좋음 else "다름"})
판정 = pd.DataFrame(판정)  # Return 줄은 '원문 α'·'우리 α' 칸 = IMF 누적, '원문 t'·'우리 t(OLS)' 칸 = MF 누적
판정.to_csv(평가.결과폴더 / "01_원문판정.csv", encoding="utf-8-sig", index=False)
print("원문 판정", len(판정), "줄 중 같음", int(판정["원문과 같음"].eq("같음").sum()))
판정
```

### 8. 표본: 원문 표 1(Number·Average return·Standard deviation) vs 우리
원문 3.1절의 표본 조건: 유가증권 + 코스닥, 은행·보험·증권·투자금융·신용금고·종합금융·기타금융 제외, 12월 결산, 필요한 자료 입수 가능, 기준일 거래정지 제외(p.512). '우리: 순위 매김' = 리밸런싱마다 거르기 뒤 MF 지표(ROC·EY)가 있어 순위를 매긴 종목 수.
우리 설정에는 원문에 없는 `보통주`(코드 끝자리 0, 회원 결정 "보통주만 해"), `쓰지말것`(README_데이터의 '쓰지 말 것' 구간) 단계도 있다. 투자기구(리츠·선박·부동산투자) 제외는 원문에 없어 뺐다(회원 05:57). 850에 맞추려고 조건을 새로 만들지 않는다. 단계별 남은 수는 확장연구.
평균수익률·표준편차: 원문은 내는 법을 적지 않았다. 우리는 순위를 매긴 종목마다 4월 첫 거래일 → 다음 해 3월 마지막 거래일 1년 수익률(수정주가, 중간 폐지는 마지막 상장 가격)의 단면 평균·표본 표준편차(%)로 잰다(기본값 ②, 11절). 그 1년 구간에 '쓰지 말 것' 칸이 걸린 종목은 뺀다(본문 수익률과 같은 규칙). 결과 `01_표1대조.csv`.

```python
%%절 3
표1 = pd.Series([850, 900, 972, 1033, 1098, 1129, 1196, 1282, 1351, 1382, 1434, 1484, 1568, 1638, 1744, 1815, 1880, 1955],
               index=range(2004, 2022), name="원문 표1")
기록 = 결과["MF"]["기록"]
표1통계 = pd.DataFrame({"원문 평균(%)": [40.88, 70.36, 11.60, 13.59, -26.56, 49.97, 14.02, 12.67, 14.41, 5.91, 32.31, 16.21, 3.40, 13.94, -8.43, -24.18, 98.69, 5.76],
                      "원문 표준편차(%)": [106.23, 187.52, 53.91, 67.75, 34.07, 87.05, 61.06, 61.35, 57.63, 43.52, 93.78, 61.97, 42.11, 62.35, 36.58, 42.81, 123.51, 49.53]}, index=표1.index)
P = 읽기.상장가격(가격)
Pf, d = P.ffill(), P.index.to_series()
마지막 = d.groupby(d.dt.to_period("M")).max()
금지 = 읽기.쓰지말것(가격)  # README_데이터 '쓰지 말 것' 구간(본문 수익률 계산과 같은 규칙)
우리1 = {}
for t in 기록.index:  # 리밸런싱일
    s, c, _ = 포.거르기(t, 설정["MF"])
    v = pd.DataFrame({m: getattr(지표, m)(s).reindex(c) for m, _ in 설정["MF"]["지표"]}).replace([np.inf, -np.inf], np.nan).dropna()
    끝날 = 마지막[pd.Period(f"{t.year + 1}-03", "M")]
    r = (Pf.loc[끝날] / P.loc[t] - 1).reindex(v.index)  # 4월 첫 거래일 → 다음 해 3월 마지막 거래일
    r[r.index.isin(금지.loc[(금지["시작"] < 끝날) & (금지["끝"] > t), "코드"])] = float("nan")  # 그 1년 구간에 '쓰지 말 것' 칸이 걸린 종목은 뺌
    우리1[t.year] = {"우리: 순위 매김": len(v), "우리 평균(%)": r.mean() * 100, "우리 표준편차(%)": r.std() * 100}
표1대조 = pd.concat([표1, 표1통계, pd.DataFrame(우리1).T], axis=1).dropna()
표1대조.to_csv(평가.결과폴더 / "01_표1대조.csv", encoding="utf-8-sig")
표1대조
```

#### 남은 차이의 원인 후보
- **표 1 'Number'의 정의**: 원문은 '표본 종목 수', '분석대상 기업의 종목 수는 850개'(p.512)라고만 하고, 어떤 기준으로 셌는지(재료가 모두 있는 수인지, 상장 기간 조건이 있는지)는 적지 않았다.
- **원문 데이터(TS2000)의 종목 범위**: 코스닥 소형주·상장 직후 종목의 재무 입수 범위가 DataGuide와 다를 수 있다.
- **'필요한 자료 입수 가능'의 범위**: 우리는 t−1년 결산만 본다(원문 '직전년도'). 원문이 더 이전 연도까지 요구했는지는 알 수 없다.
- **금융업 범위**: 원문 7개 업종 이름이 거래소 업종(세부)과 1:1로 맞지 않는다. 거래소 업종이 빈 날은 이름 규칙(은행·보험·증권·금융·금고·캐피탈·카드·저축·신금)이나 그날 WICS '금융'으로 판정한다(업종 빈칸 수·WICS 금융 수는 확장연구 8절 기록 표).
- **이자비용 빈칸**: 이자비용이 없는 회사가 0 대신 빈칸으로 들어 있을 수 있다. 빈칸은 채우지 않고 그 종목을 뺀다(회원 결정, README '빈칸 그대로'. 연도별 빠진 수는 확장연구 6절)(빚 없는 회사가 빠지는 쪽으로 치우칠 수 있음).
- **원문에 적히지 않은 조건**: 우선주·관리종목·자본잠식 등이 TS2000 표본에서 빠져 있을 수 있다(확인 불가).
- **상장폐지 종목 포함 여부**: 원문에 서술이 없다.

### 9. 차이 이유 후보
- **데이터 출처**: 원문은 TS2000·DART·NAVER금융(p.512), 우리는 DataGuide. 과제 자료라 바꿀 수 없다(한계). 종목 범위·재무 값이 다를 수 있다(8절).
- **EBIT**: 세전계속사업이익 + 이자비용. 중단사업손익은 들어가지 않는다(원문이 중단사업을 어떻게 했는지 서술 없음).
- **표준편차·샤프·정보비율·MDD 정의**: 원문에 정의가 없다(p.513 식 (1)·(2)는 평균/표준편차 꼴만). KOSPI 원문 숫자(표준편차 83.49%, 샤프 1.34, MDD −54.54%, 정보비율 0.00)가 **일간 수익률, 표준편차 × √(전 기간 일 수), 샤프 = 일 초과수익(국고 3년 금리/365, 직전 날) 평균/표준편차 × √(일 수), MDD = 일간 가치**에서 나와(`_검증/01_정의역산.py`·`_결과.txt`) 이 정의로 잰다(7절 KOSPI 줄로 확인). p.514 '월별로 집계' 문장과는 맞지 않는다. 월간 × √12 판은 확장연구.
- **무위험 일간 /365**: 원문은 '3년 만기 국고채 수익률'(p.513)만 적었다. 거래일마다 연 금리/365를 빼면 1년에 빠지는 무위험이 연 금리보다 작다(거래일 약 250일). /252로 하면 KOSPI 샤프가 1.1444라 원문 1.34와 멀어져(`_검증/01_정의역산_결과.txt`) 원문 숫자가 나오는 /365를 쓴다.
- **누적수익률 계산**: 원문이 거래비용·월별 집계를 어떻게 했는지 자세히 나와 있지 않다. 우리는 리밸런싱 때 전량 매도·매수 비용을 그 뒤 첫 달에 뺀다(마지막 청산 비용은 안 뺌, 11절 기본값).
- **시장지수**: 코스피 가격지수(I.001). 원문은 가격/TR을 적지 않았지만, I.001 일간 종가 2004-04-01 → 2022-03-31 누적이 2.12393으로 원문 KOSPI 212.39%와 같다(`_검증/01_정의역산_결과.txt`). 포트폴리오도 수정주가(배당 미포함)라 정의가 맞는다. 코스닥 종목도 담지만 벤치마크는 코스피뿐이다.
- **요인**: 과제 공지('규모/가치/모멘텀 팩터 수익률은 직접 만들지 말고 DataGuide에서 제공하는 데이터를 사용')대로 DataGuide 요인(3FM.2B3.SMB·HML, 3FM.2M3.MOM, 2×3 요인지수)을 쓴다. 원문 p.513은 시총·BM·과거 11개월 수익률 상위 30% − 하위 30%로 직접 구성해 정의가 다르므로, 3요인·4요인 α 차이의 원인 후보다.
- **보유 중 상장폐지**: 원문 처리 방법이 없다. 우리는 폐지된 구간에 시장구분이 코스피·코스닥인 마지막 날 가격으로 판 것으로 보고, 그 뒤로는 남은 종목에 값 비례로 다시 넣은 것과 같게 계산한다(대가 지급 폐지는 반영 안 됨).

### 11. 원문에 말이 없어 정한 것 (기본값)
원문에 없는데 빼면 계산이 안 되는 칸은 회원 결정(2026-11-06)대로 ① 자료가 없거나 계산이 안 되면 그 종목을 그 달(해)에서 빼고 ② 나머지는 가장 단순한 쪽으로 정했다. 다른 선택의 영향은 이 노트북에서 재지 않는다. 원문에 근거가 있는 줄은 굵게 따로 적었다(**원문에 있음**·**원문 안 어긋남**·**역산(원문 숫자에 맞춰 찾은 정의)**·**제도로 정함**·**해석**).

### 3. 설정
지표만 바꿔 같은 설정으로 돌린다.
- 재무: 원문 p.14 "주요 재무제표 수치는 분기/연간 재무제표가 발표되는 월의 월말에 반영" → 기준일까지 발표된 가장 최근 보고서 값 그대로(`포.발표재무`, 회원 결정 2026-11-06: 4분기 합·×4 같은 가공 없음). 발표일이 없는 보고서는 쓰지 않는다(기본값 ①). 오래된 보고서 상한은 두지 않는다(기본값 ②).
- 보통주만, 수익률은 수정주가(회원 결정 2026-11-06, 원문에 없음).
- 원문 본 실험 조건(p.14)에는 거래비용 언급이 없어 0으로 둔다(기본값 ②; 원문이 비용을 뺐는지는 알 수 없음).

```python
%%절 4
# 재무시점은 발표값에서는 쓰지 않는다(보고서 발표일 = 쓸 수 있는 날, 포.발표행). '보통주'는 관리·정지 다음(현실판 시총 300위 순위도 보통주 안에서 매김)
기본 = {**포.기본설정, "이름": "NH밸류", "시장": ["KS"], "가격": 읽기.수정주가, "기간": 기간, "기준일": "마지막거래일", "리밸런싱": 1,
       "재무주기": ("발표값",), "재무시점": ("발표월말", 0), "제외": [("거래정지", {"아이템": 읽기.정지여부}), "관리종목", "보통주"],
       "고르기": ("분위", 5), "비중": ["동일"], "거래비용": dict(수수료=0, 거래세=0, 슬리피지=0)}
지표명 = {"P/B": "BP", "P/E": "EP", "P/S": "SP", "P/OCF": "CFP", "EV/EBITDA": "EBITDA_EV"}
숏 = {"P/E": "Q4", "P/OCF": "Q4"}  # 원문 p.14: 적자 종목이 많아 1분위 − 4분위
pd.Series(기본).astype(str).to_frame("공통 설정")
```

### 4. 실행: 지표별 5분위 + Multi-Metric

```python
%%절 4
결과 = {n: 포.실행({**기본, "지표": [(m, "높게")]}) for n, m in 지표명.items()}
결과["Multi-Metric"] = 포.실행({**기본, "지표": [(m, "높게") for m in 지표명.values()], "합산": "z합산"})
```

### 5. 롱숏(LS) 누적 (그림 16·17)
원문 그림 16(p.14)·17(p.15): 세로축 누적 수익률(%), 선형 축. 그림 17은 2010년 이후(2009-12 말을 0으로 두고 2010-01 수익률부터 다시 누적). 범례 순서는 원문대로.

```python
%%절 4
LS = pd.DataFrame({n: o["수익률"]["동일"]["Q1"] - o["수익률"]["동일"][숏.get(n, "Q5")] for n, o in 결과.items() if n != "Multi-Metric"})
원문순서 = ["P/B", "EV/EBITDA", "P/E", "P/S", "P/OCF"]  # 원문 그림 16·17 범례 순서
평가.누적그림(LS[원문순서], "그림 16. 측정지표에 따른 KOSPI 밸류 프리미엄(1분위 Long, 5분위 Short; P/E·P/OCF는 4분위 Short)", 퍼센트=True);
if len(LS.loc["2010":]):  # 짧은 시험 기간에는 2010년 이후가 없음
    평가.누적그림(LS.loc["2010":, 원문순서], "그림 17. 측정지표에 따른 KOSPI 밸류 프리미엄(2010년 이후)", 퍼센트=True);
```

### 6. 1분위 Long Only 누적 (그림 18)
원문 그림 18(p.15): 1분위 동일가중 누적 수익률(%), 선형 축, Multi-Metric과 5개 지표.

```python
%%절 4
Q1 = pd.DataFrame({n: o["수익률"]["동일"]["Q1"] for n, o in 결과.items()})
평가.누적그림(Q1[["Multi-Metric", *원문순서]].dropna(how="all"), "그림 18. 측정지표에 따른 1분위 종목 동일가중 수익률", 퍼센트=True);
```

### 7. 리밸런싱 주기 (그림 23·24, p.18)
원문 p.18: "KOSPI 상장종목을 대상으로 P/B 지표를 사용해서 앞서와 동일한 로직으로 1개월, 3개월, 6개월, 12개월로 리밸런싱 주기를 변경". 그림 23 = 네 주기, 그림 24 = 3·6·12개월(P/B 1분위 Long, 5분위 Short, 누적 %).
1개월은 4절 P/B 결과와 같은 설정이라 그대로 쓴다. 3·6·12개월은 첫 리밸런싱(2002-03 말)부터 k달마다 다시 고르고 사이에는 매수보유(시작 달은 원문에 없음, 기본값 ②).

```python
%%절 4
주기 = {"1개월": 결과["P/B"]} | {f"{k}개월": 포.실행({**기본, "지표": [("BP", "높게")], "리밸런싱": k}) for k in (3, 6, 12)}
주기LS = pd.DataFrame({k: o["수익률"]["동일"]["Q1"] - o["수익률"]["동일"]["Q5"] for k, o in 주기.items()})
평가.누적그림(주기LS, "그림 23. P/B 지표 기준 1분위 Long, 5분위 Short 포트폴리오 성과(KOSPI)", 퍼센트=True);
평가.누적그림(주기LS.drop(columns="1개월"), "그림 24. 리밸런싱 주기가 길어질수록 효과가 크게 감소", 퍼센트=True);
```

### 애널리스트 추정치 P/E (원문 p.19 그림 25, p.20 그림 26)
원문 p.19 "2) 애널리스트 추정치 사용": "이에 동일한 로직으로 확정치가 아닌 애널리스트의 최근 3개월 EPS 추정치를 사용해 매월 리밸런싱하는 밸류 포트폴리오의 성과를 점검했다.
3개월 이내 애널리스트 추정치가 존재하는 종목의 수가 대략 300개 정도로 총 3개 그룹으로 구분해 상위 1/3 Long, 하위 1/3 Short 포트폴리오의 성과로 분석했다."

| 항목 | 원문 | 쪽 | 우리 |
|---|---|---|---|
| 추정치 | 애널리스트의 최근 3개월 EPS 추정치 | 19 | DataGuide 컨센서스 '3개월' EPS(지배) FY1·FY2(`읽기.컨센서스`, FM30011505 R17·R18) |
| 지표 | 그림 25·26 범례 '12개월 Fwd P/E' | 19·20 | 역수 꼴 `EP_선행` = 12개월 선행 EPS / 수정주가(본문 P/E도 역수 꼴 E/P라 같은 꼴) |
| 12개월 선행 계산 | 원문에 없음 | — | FY1 × (12 − m)/12 + FY2 × m/12(`지표.EP_선행`). 우리가 고른 값 |
| 묶음 | 3개 그룹, 상위 1/3 Long·하위 1/3 Short(그림 25 제목 '1분위 Long, 3분위 Short') | 19 | `고르기=("분위", 3)`, LS = Q1 − Q3 |
| 유니버스 | '동일한 로직' + 3개월 이내 추정치가 있는 종목 | 19 | 본문 `기본` 거르기 뒤 FY1·FY2 추정치가 모두 있는 종목(없으면 그 달 제외). 원문 조건(p.19 "3개월 이내 애널리스트 추정치가 존재하는 종목") |
| 리밸런싱 | 매월 | 19 | 본문과 같음(월말, 1개월) |
| 그림 25 기간 | 가로축 '10.1~'18 | 19 | 2010-01 말 첫 리밸런싱 ~ 본문 끝(7절 현실판 '09.1과 같은 읽기). 본문 실행에서 2010-02 수익률부터 자름(매월 리밸런싱·비용 0이라 2010-01부터 따로 돌린 것과 같음) |
| 그림 26 기간 | "2002년 이후", 가로축 '02.4~ | 19·20 | 본문 기간 |
| 그림 26 묶음 | 상위 1/3 Long Only 동일가중 | 19 | Q1, 동일가중 |
| 비교선 'P/E' | 확정 P/E(그림 25·26 범례 'P/E') | 19·20 | 본문 `EP`(발표 재무), 같은 3분위. 유니버스는 본문 전체(원문 p.19 "추정치가 존재하는 종목 수가 제한적"이 확정 P/E 쪽이 더 넓다는 뜻으로 읽음) |

원문 판정(방향·크기만): ① 그림 25 "P/E 지표를 이용한 밸류 포트폴리오가 2014년 이후 성과가 부진했던 것과 달리 애널리스트 추정치 P/E를 사용할 경우 꾸준히 효과적이었다"(p.19)
② 그림 26 "2002년 이후 상위 1/3 Long Only 동일가중 포트폴리오 누적 수익률에서 확정 P/E를 사용하는 데에 비해 크게 못 미쳤다"(p.19) ③ "3개월 이내 애널리스트 추정치가 존재하는 종목의 수가 대략 300개"(p.19). ③의 판정 잣대(리밸런싱마다 종목 수 평균을 백 단위로 반올림해 300)는 원문에 없어 우리가 고른 값.

```python
%%절 4
# 시험(FOLLOW_TEST=1)이면 1절에서 평가.결과폴더가 _검증\1106\시험결과\로 바뀌어 있음
추정 = 포.실행({**기본, "지표": [("EP_선행", "높게")], "고르기": ("분위", 3)})
확정 = 포.실행({**기본, "지표": [("EP", "높게")], "고르기": ("분위", 3)})
q3 = lambda o: o["수익률"]["동일"]
LS3 = pd.DataFrame({"12개월 Fwd P/E": q3(추정)["Q1"] - q3(추정)["Q3"], "P/E": q3(확정)["Q1"] - q3(확정)["Q3"]})
Q1_3 = pd.DataFrame({"12개월 Fwd P/E": q3(추정)["Q1"], "P/E": q3(확정)["Q1"]})
추정수 = 추정["기록"][["편입:Q1", "편입:Q2", "편입:Q3"]].sum(axis=1)  # 그 달 추정치가 있어 3분위에 들어간 종목 수
print("추정치 종목 수(리밸런싱마다):", 추정수.describe().round(1).to_dict(), "/ 첫 리밸런싱", f"{추정수.index[0]:%Y-%m-%d}", int(추정수.iloc[0]))
시작25 = "2010-02" if not 시험 else LS3.index[0]  # 시험(짧은 기간)은 2010년 전이라 전체 구간
LS25 = LS3.loc[시작25:]
평가.누적그림(LS25, "그림 25: 12개월 Fwd P/E와 P/E, 1분위 Long·3분위 Short", 퍼센트=True);
평가.누적그림(Q1_3.dropna(how="all"), "그림 26: 12개월 Fwd P/E와 P/E, 상위 1/3 Long Only 동일가중", 퍼센트=True);
```

```python
%%절 4
누25 = (1 + LS25).prod() - 1
후 = LS25.loc["2014":]  # 원문 '2014년 이후'
누후 = (1 + 후).prod() - 1
누26 = (1 + Q1_3).prod() - 1
대조25_26 = pd.DataFrame([
    ["p.19 그림 25", "애널리스트 추정치 P/E LS는 꾸준히 효과적", "(우리가 읽은 기준) 12개월 Fwd P/E LS 누적 > 0",
     f"Fwd {누25['12개월 Fwd P/E']:.3f} ({LS25.index[0]:%Y-%m}~{LS25.index[-1]:%Y-%m})", "같음" if 누25["12개월 Fwd P/E"] > 0 else "다름"],
    ["p.19 그림 25", "P/E는 2014년 이후 부진, 추정치 P/E는 달리 효과적", "(우리가 읽은 기준) 2014-01~ LS 누적: Fwd > 0이고 Fwd > P/E",
     f"2014~ Fwd {누후['12개월 Fwd P/E']:.3f}, P/E {누후['P/E']:.3f}" if len(후) else "구간 없음(시험 기간)",
     ("같음" if 누후["12개월 Fwd P/E"] > 0 and 누후["12개월 Fwd P/E"] > 누후["P/E"] else "다름") if len(후) else "판정 안 함"],
    ["p.19 그림 26", "상위 1/3 Long Only 누적: 추정치 P/E가 확정 P/E에 크게 못 미침", "Fwd Q1 누적 < P/E Q1 누적(크기는 두 누적을 그대로 적음)",
     f"Fwd {누26['12개월 Fwd P/E']:.3f}, P/E {누26['P/E']:.3f} ({Q1_3.dropna(how='all').index[0]:%Y-%m}~{Q1_3.index[-1]:%Y-%m})",
     "같음" if 누26["12개월 Fwd P/E"] < 누26["P/E"] else "다름"],
    ["p.19", "3개월 이내 추정치가 있는 종목 수 대략 300개", "(우리가 읽은 기준) 리밸런싱마다 종목 수 평균을 백 단위로 반올림하면 300",
     f"평균 {추정수.mean():.1f}, 최소 {추정수.min()}, 최대 {추정수.max()}", "같음" if round(추정수.mean(), -2) == 300 else "다름"],
], columns=["원문 쪽·그림", "원문 결론", "판정 기준", "우리", "판정"])
대조25_26.to_csv(평가.결과폴더 /"02_원문대조_그림25_26.csv", encoding="utf-8-sig")
대조25_26
```

### 8. 시가총액별·유동성별·KOSDAQ (그림 28~33, p.21~24)
- 그림 28·29(p.21~22): "KOSPI 전체 종목, 대형주와 중형주를 포함하는 시가총액 상위 300위 종목 및 대형주만을 포함하는 시가총액 상위 100위 이내 종목을 대상으로 P/B, P/E 기준 성과를 검증". 그림 28 주 "P/B 기준 1분위 Long, 5분위 Short", 그림 29 주 "P/E 기준 1분위 Long, 3분위 Short" → 3분위(3그룹) Q1 − Q3로 읽음(해석): 같은 보고서 p.19 "총 3개 그룹으로 구분해 상위 1/3 Long, 하위 1/3 Short"를 그림 25 제목이 "1분위 Long, 3분위 Short"라고 부름(분위 수는 그림 29에 따로 없음, `_검증/1106/원문사양_02.md` 3절 1). 세 유니버스마다 그 안에서 분위를 새로 나눔.
- 그림 30·31(p.23): "월평균 거래대금 10억원 이상, 20억원 이상인 종목으로 한정", 그림 30 = KOSPI 전체, 그림 31 = 시가총액 300위 이내, 둘 다 "P/B 기준 1분위 Long, 5분위 Short". '월평균'은 현실판(9절)과 같게 기준일이 든 달의 거래일 평균 거래대금.
- 그림 32·33(p.24): "적자기업 등이 많은 KOSDAQ 특성상 시가총액 300위 이내 기업의 밸류 지표 상위 1/3 – 하위 1/3 동일가중 수익률", 그림 33 주 "KOSPI, KOSDAQ 시가총액 상위 300위 종목 대상 P/B 기준 상위 1/3 Long, 하위 1/3 Short" → 3분위 Q1 − Q3.
- 나머지(관리·정지 제외, 보통주, 월말 1개월 리밸런싱, 재무 발표값, 기간)는 본 실험과 같음(원문 p.18 "앞서와 동일한 로직"). 시총 순위는 관리·정지·보통주를 거른 뒤 남은 종목에서 매긴다(지표가 있는지 보기 전, 원문에 없음 → 기본값 ②). 시총과 거래대금을 둘 다 거를 때는 시총 먼저(현실판과 같음, 원문에 순서 없음 → 기본값 ②).

`LS판`은 같은 거르기로 P/B·P/E를 한 번에 돌린다(`포.실행`의 `전략들`: 같은 날 같은 남은 종목에서 지표마다 따로 줄 세움 = 지표마다 따로 돌린 것과 같은 계산).

```python
%%절 4
def LS판(숏들, 더제외=(), 시장=("KS",), 분위=5):
    """본 실험 설정에 거르기를 더해 지표마다 1분위 − 숏 분위 월수익률. 숏들 = {지표명 키: 숏 분위}."""
    R = 포.실행({**기본, "시장": list(시장), "제외": 기본["제외"] + list(더제외), "고르기": ("분위", 분위),
                 "전략들": {n: dict(지표=[(지표명[n], "높게")], 합산="순위합산") for n in 숏들}})["수익률"]
    return pd.DataFrame({n: R[(n, "동일", "Q1")] - R[(n, "동일", q)] for n, q in 숏들.items()})
시총 = lambda n: ("시총상위", {"n": n})
대금 = lambda 하한: ("거래대금", {"하한": 하한, "창": "월"})
시총300, 시총100 = LS판({"P/B": "Q5"}, [시총(300)]), LS판({"P/B": "Q5"}, [시총(100)])  # 그림 28: 5분위
셋300 = LS판({"P/B": "Q3", "P/E": "Q3"}, [시총(300)], 분위=3)  # 3분위: 그림 29의 300위 선(P/E), 그림 33의 KOSPI 선(P/B)
그림28 = pd.DataFrame({"KOSPI 전체": LS["P/B"], "KOSPI 시총300위 이내": 시총300["P/B"], "KOSPI 시총100위 이내": 시총100["P/B"]})
그림29 = pd.DataFrame({"KOSPI 전체": LS판({"P/E": "Q3"}, 분위=3)["P/E"], "KOSPI 시총300위 이내": 셋300["P/E"],
                     "KOSPI 시총100위 이내": LS판({"P/E": "Q3"}, [시총(100)], 분위=3)["P/E"]})
평가.누적그림(그림28, "그림 28. KOSPI도 시가총액이 낮을수록 밸류 프리미엄이 높아(P/B 기준, 1분위 Long, 5분위 Short)", 퍼센트=True);
평가.누적그림(그림29, "그림 29. KOSPI 시가총액 낮을수록 밸류 프리미엄이 높아(P/E 기준, 1분위 Long, 3분위 Short)", 퍼센트=True);
```

```python
%%절 4
그림30 = pd.DataFrame({"거래대금 조건X": LS["P/B"], "10억원 이상": LS판({"P/B": "Q5"}, [대금(1e9)])["P/B"], "20억원 이상": LS판({"P/B": "Q5"}, [대금(2e9)])["P/B"]})
그림31 = pd.DataFrame({"거래대금 조건X": 시총300["P/B"], "10억원 이상": LS판({"P/B": "Q5"}, [시총(300), 대금(1e9)])["P/B"],
                     "20억원 이상": LS판({"P/B": "Q5"}, [시총(300), 대금(2e9)])["P/B"]})
평가.누적그림(그림30, "그림 30. 유동성이 낮을수록 밸류 프리미엄이 높아(KOSPI 전체 대상 P/B 기준)", 퍼센트=True);
평가.누적그림(그림31, "그림 31. 대형주, 중형주의 유동성과 밸류 프리미엄(KOSPI 시가총액 300위 이내 대상 P/B 기준)", 퍼센트=True);
```

```python
%%절 4
그림32 = LS판({"P/B": "Q3", "P/E": "Q3"}, [시총(300)], 시장=("KQ",), 분위=3)
그림33 = pd.DataFrame({"KOSDAQ": 그림32["P/B"], "KOSPI": 셋300["P/B"]})  # 원문 그림 33 주: KOSPI·KOSDAQ 시총 300위, P/B 상위 1/3 − 하위 1/3
평가.누적그림(그림32, "그림 32. KOSDAQ 종목에서도 밸류 프리미엄이 확인됨(시총 300위 이내, 상위 1/3 Long, 하위 1/3 Short)", 퍼센트=True);
평가.누적그림(그림33, "그림 33. 2002년 이후 KOSDAQ의 밸류 프리미엄이 KOSPI보다 높아(P/B 기준, 시총 300위 이내)", 퍼센트=True);
```

### 9. 현실판(p.26): 시총 300위 이내 + 거래대금 10억 원 이상, Multi-Metric, 매매비용 월 0.2%
원문(p.26): "이러한 방식으로 매월 전체 포트폴리오의 0.2% 매매비용을 감안해 1개월 단위 리밸런싱한 결과 KOSPI 200 동일가중지수를 Outperform할 수 있었다(그림36 참조)."
그림 36 제목(p.26): "5가지 지표 Multi-Metric 방식으로 선정한 동일가중 포트폴리오 수익률 (1개월 단위 리밸런싱, 매매비용 월 0.2% 가정)".
본 결과는 매달 포트폴리오의 0.2%를 뺀 판이다(그림 36: 세로축 누적 %, 선형). 기간: 원문 그림 36 '09.1 첫 점이 0 아래라 2008-12 말 첫 리밸런싱(첫 수익률 2009-01). I.401은 값이 2009-01-02부터라 그림 36의 I.401 선과 Outperform 대조는 2009-02부터(자료 한계). 벤치마크는 원문대로 KOSPI 200 동일가중지수(X01 `I.401` '코스피 200 동일가중지수', 종가지수, 값은 2009-01-02부터)를 놓는다(2026-11-04 고침, 예전엔 '데이터에 없음'으로 잘못 보고 코스피 I.001). 배당 포함 여부는 X01에 적혀 있지 않다(확인 안 함). 포트폴리오는 수정주가(배당 없음).

```python
%%절 4
# 기간 = 현실기간(2절: 2008-12 말 첫 리밸런싱 ~ 2018-05). 유니버스: 시총 300위·월평균 거래대금 10억
현실설정 = {**기본, "이름": "NH밸류_현실판", "기간": 현실기간, "지표": [(m, "높게") for m in 지표명.values()], "합산": "z합산",
           "제외": 기본["제외"] + [("시총상위", {"n": 300}), ("거래대금", {"창": "월"})]}
월비용 = dict(수수료=0, 거래세=0, 슬리피지=0, 월비용=0.002)
현실 = {"현실판_비용월0.2%(본)": 포.실행({**현실설정, "거래비용": 월비용})}
F200 = 평가.요인(포.기준일({**기본, "기간": (기간[0], 현실기간[1])}), {**기본, "시장지수": "I.401"})  # 원문 벤치마크(p.26) KOSPI 200 동일가중지수
현실표 = pd.DataFrame({k: o["수익률"][("동일", "Q1")] for k, o in 현실.items()} | {"KOSPI200 동일가중(I.401)": F200["시장"]})
본36 = ["현실판_비용월0.2%(본)", "KOSPI200 동일가중(I.401)"]
그림36 = 현실표.loc["2009-02":, 본36].dropna().set_axis(["밸류 포트폴리오(비용 월 0.2%)", "KOSPI 200 동일가중지수(I.401)"], axis=1)  # 대조용: 두 선 모두 값 있는 달(2009-02~)
평가.누적그림(현실표.loc[현실표[본36[0]].first_valid_index():, 본36].set_axis(["밸류 포트폴리오(비용 월 0.2%, 2009-01~)", "KOSPI 200 동일가중지수(I.401, 2009-02~)"], axis=1),
         "그림 36. 5가지 지표 Multi-Metric 동일가중 포트폴리오(1개월 리밸런싱, 매매비용 월 0.2%)", 퍼센트=True);
```

### 10. 원문 vs 우리
원문은 대부분 그림이라 숫자는 그림 19(위험조정수익률)만 비교한다. 나머지는 원문의 문장(주장)과 우리 결과가 같은 방향인지 본다.

```python
%%절 4
그림19 = pd.Series({"Multi-Metric": 0.39, "P/B": 0.34, "P/E": 0.32, "EV/EBITDA": 0.34, "P/S": 0.30, "P/OCF": 0.29})
우리19 = (Q1.mean() / Q1.std())[그림19.index]
그림19표 = pd.DataFrame({"원문 그림19": 그림19, "우리(Q1 월평균/월표준편차)": 우리19})
display(그림19표)
누적LS = (1 + LS).prod() - 1
주장 = pd.DataFrame({"원문 주장": ["모든 지표에서 LS 누적이 양(+)", "P/B·P/E가 가장 효과적(LS 누적 상위 2)", "Multi-Metric 1분위가 단일 지표 1분위보다 위험조정수익률 높음"],
              "우리": [bool((누적LS > 0).all()), sorted(누적LS.nlargest(2).index) == ["P/B", "P/E"], bool(우리19.idxmax() == "Multi-Metric")]})
그림19표.to_csv(평가.결과폴더 / "02_원문대조_그림19.csv", encoding="utf-8-sig")  # 결과요약 노트북이 읽음
주장.to_csv(평가.결과폴더 / "02_원문주장.csv", encoding="utf-8-sig")
주장
```

#### 원문 결론 판정 (원문이 말한 방향·크기만)
원문 문장마다 원문이 말한 방향·크기만 본다(p값 기준 없음). '판정 기준' 칸은 숫자 기준이 없는 원문 말을 우리가 읽은 방법이다. 그림 31 "감소 폭이 제한적"의 잣대(조건X 대비 20억 원 이상 LS 누적 감소율을 그림 30과 비교)는 원문에 없어 우리가 고른 값. 그림 25·26(애널리스트 추정치) 결론은 G02 칸.

원문 안에서 서로 어긋나는 곳(원문 그대로 적음, 그림 값은 눈대중, `_검증/1106/원문사양_02.md`):
- p.14 본문은 "P/E와 P/OCF 지표의 경우 … 1분위 Long, 4분위 Short"인데 그림 16 제목은 "(1분위 Long, 5분위 Short)". 우리는 본문을 따름.
- P/B LS 누적 끝값이 그림 16에서는 약 2,250%, 같은 P/B 1개월 LS인 그림 23·28·30에서는 약 850%.
- 그림 24 제목은 "리밸런싱 주기가 길어질수록 효과가 크게 감소"인데 그림에서는 6개월(약 25%)이 12개월(약 45%)보다 낮게 끝남.
- p.14는 "P/E는 2015년 이후 다소 부진", p.19는 "P/E 지표를 이용한 밸류 포트폴리오가 2014년 이후 성과가 부진". 판정은 p.14 문장(그림 16)대로 2015년으로 나눔.
- 그림 31에서는 10억 원 이상(약 530%)이 조건 X(약 460%)보다 높게 끝남.

```python
%%절 4
누적 = lambda x: (1 + x).prod() - 1
연환산 = lambda x: (1 + x.dropna()).prod() ** (12 / x.dropna().size) - 1 if x.dropna().size else np.nan
판 = lambda ok: "같음" if ok else "다름"
줄 = lambda s: ", ".join(f"{k} {v:.3f}" for k, v in s.items())
누적Q1 = 누적(Q1[list(결과)])
최고단일 = 누적Q1.drop("Multi-Metric").idxmax()
ls10 = 누적(LS.loc["2010":])
pe전, pe후 = 연환산(LS["P/E"].loc[:"2014"]), 연환산(LS["P/E"].loc["2015":])
주기누적 = 누적(주기LS)
그림누적 = {k: 누적(v) for k, v in dict(그림28=그림28, 그림29=그림29, 그림30=그림30, 그림31=그림31, 그림32=그림32, 그림33=그림33, 그림36=그림36).items()}
감소율 = lambda s: (s["거래대금 조건X"] - s["20억원 이상"]) / s["거래대금 조건X"]  # 거래대금 20억 거르기로 줄어든 LS 누적의 비율
결론 = pd.DataFrame([
    ["p.15 그림 18", "Multi-Metric 1분위 누적이 단일 지표 1분위보다 '월등하게 우수'", "Multi 1분위 누적 > 단일 지표 1분위 누적 최고('월등'의 크기 기준은 원문에 없음)",
     f"Multi {누적Q1['Multi-Metric']:.3f}, 단일 최고 {최고단일} {누적Q1[최고단일]:.3f}", 판(누적Q1.idxmax() == "Multi-Metric")],
    ["p.14 그림 17", "2010년 이후로 한정해도 P/B·P/E가 가장 효과적", "2010-01~ LS 누적 상위 2 = P/B·P/E", 줄(ls10.sort_values(ascending=False)), 판(set(ls10.nlargest(2).index) == {"P/B", "P/E"})],
    ["p.14 그림 16", "P/E는 2015년 이후 다소 부진", "P/E LS 연환산: 2015-01~ < ~2014-12", f"~2014 {pe전:.4f}, 2015~ {pe후:.4f}", 판(pe후 < pe전)],
    ["p.18 그림 23·24", "P/B LS는 1개월 리밸런싱이 가장 우수(1·3·6·12개월)", "LS 누적이 1개월에서 가장 큼", 줄(주기누적), 판(주기누적.idxmax() == "1개월")],
    ["p.21 그림 28", "P/B LS 누적: KOSPI 전체 > 시총 300위 이내 > 시총 100위 이내", "세 LS 누적이 이 순서", 줄(그림누적["그림28"]), 판(그림누적["그림28"].is_monotonic_decreasing)],
    ["p.21~22 그림 29", "P/E LS(1분위 − 3분위) 누적: KOSPI 전체 > 시총 300위 이내 > 시총 100위 이내", "세 LS 누적이 이 순서", 줄(그림누적["그림29"]), 판(그림누적["그림29"].is_monotonic_decreasing)],
    ["p.23 그림 30", "거래대금 10억·20억 원 이상으로 한정하면 밸류 프리미엄 감소(유동성 낮을수록 높음)", "KOSPI 전체 P/B LS 누적: 조건X > 10억 > 20억", 줄(그림누적["그림30"]), 판(그림누적["그림30"].is_monotonic_decreasing)],
    ["p.23 그림 31", "시총 300위 이내 대상이면 감소 폭이 제한적", "(조건X − 20억) / 조건X LS 누적 감소율이 시총 300위(그림 31) < KOSPI 전체(그림 30)('제한적' 크기 기준은 원문에 없음)",
     f"감소율 전체 {감소율(그림누적['그림30']):.3f}, 시총300 {감소율(그림누적['그림31']):.3f}", 판(감소율(그림누적["그림31"]) < 감소율(그림누적["그림30"]))],
    ["p.24 그림 32", "KOSDAQ에도 밸류 프리미엄이 존재(P/B·P/E)", "KOSDAQ 시총 300위 P/B·P/E 3분위 LS 누적 > 0", 줄(그림누적["그림32"]), 판((그림누적["그림32"] > 0).all())],
    ["p.24 그림 33", "KOSPI 대비 오히려 KOSDAQ의 밸류 프리미엄이 큼(P/B)", "시총 300위 P/B 3분위 LS 누적: KOSDAQ > KOSPI", 줄(그림누적["그림33"]), 판(그림누적["그림33"]["KOSDAQ"] > 그림누적["그림33"]["KOSPI"])],
    ["p.26 그림 36", "현실판(비용 월 0.2%)이 KOSPI 200 동일가중지수를 Outperform", "같은 달 누적: 밸류 포트폴리오 > I.401", 줄(그림누적["그림36"]),
     판(그림누적["그림36"].iloc[0] > 그림누적["그림36"].iloc[1])],
], columns=["원문 쪽·그림", "원문 결론", "판정 기준", "우리", "판정"])
결론.to_csv(평가.결과폴더 / "02_원문결론.csv", encoding="utf-8-sig")  # 원문 대조 저장
결론
```

### 11. 차이 이유 후보
- **재무 값**: 원문 p.14대로 발표된 보고서를 발표 월 월말에 반영하고, 값은 발표된 값 그대로 쓴다(회원 결정 2026-11-06). 그래서 같은 달에 분기 보고서(3개월 순이익·매출·영업CF·EBITDA)를 낸 종목과 아직 사업보고서(12개월 값)만 있는 종목이 한 줄에 섞인다(주로 2~4월 사업보고서 제출 기간, 그리고 12월 결산이 아닌 회사는 다른 달에도. KS 보통주 실측: 2002-03-29 분기 152·연간 356, 2009-03 분기 71·연간 609, 2012-02 연간 56·분기 671; V02 검토, 휴일 규칙 고치기 전 숫자). 원문이 흐름 항목을 어떻게 했는지는 원문에 없다. 영업CF·EBITDA의 분기 3개월 값은 보고서 숫자 그대로가 아니라 DataGuide가 환산·계산한 값이다.
- **지표 방향**: 원문은 P/B 등 배수 이름을 쓴다. 우리는 역수(B/P 등)로 줄 세워 적자·자본잠식 종목이 '싼 쪽'에 섞이지 않는다(정렬 꼴은 원문에 없음). 원문이 P/E·P/OCF를 Q1−Q4로 한 것은 적자 종목이 Q5에 모이는 것과 같은 효과로 보인다.
- **Z-Score**: 원문의 극단값 처리(윈저라이즈 등)가 나와 있지 않다. 우리는 처리하지 않는다(기본값 ②). 5개 지표가 모두 있는 종목만 쓴다(기본값 ①).
- **EV·EBITDA 정의**: 원문에 정의가 없다. 우리는 EV = 시가총액 + 부채총계 − 현금(회원 결정 2026-11-04), EBITDA = DataGuide M000701032(기본값 ②).
- **현실판 벤치마크**: KOSPI 200 동일가중지수(I.401, 종가지수, 배당 포함 여부는 X01에 없음). 포트폴리오는 수정주가(배당 없음).

### 13. 요약표용 저장
지표별 Q1, LS, 현실판의 월수익률(원문 그림 16·18·36에 대응하는 열)을 `결과_월수익률/02_NH밸류.pkl`에 남긴다(LS는 롱숏이라 초과수익 = 수익률 그대로). 요인(F: 무위험 콜금리·시장 KOSPI I.001)은 원문에 없어 여기서 계산·저장하지 않는다(pkl의 F = None) → `연구/따라하기에서_옮김_02.ipynb` 11절이 `결과_월수익률/요인/02_NH밸류.pkl`에 저장(`평가.요약`이 F = None일 때 읽음).

```python
%%절 4
S = pd.DataFrame({(f"NH밸류 {n} Q1", "동일"): Q1[n] for n in Q1} | {(f"NH밸류 {n} LS", "동일"): LS[n] for n in LS}
                 | {(f"NH밸류 {k}", "동일"): o["수익률"][("동일", "Q1")] for k, o in 현실.items()})
평가.저장(S, None, "02_NH밸류", [k for k in S if k[0].endswith(" LS")])
list(S)
```

줄임말: L/S·LS(롱숏: 1분위를 사고 5분위를 판 수익률, 매달 Q1 − Q5를 복리로 쌓음), Q1~Q5(1~5분위).

```python
%%절 5
# (제출본) 저장소 import 칸에서 0·1절이 대신하는 줄(sys.path·공통 import·0절 import·pandas 표시 설정)을 뺀 나머지
import os
시험 = bool(시험_기간)
if 시험:  # 저장 경로를 시험 폴더로(평가.저장·판정 csv 모두 평가.결과폴더를 씀)
    평가.결과폴더 = Path("결과") / "시험결과"
    평가.결과폴더.mkdir(parents=True, exist_ok=True)
print("시험 모드:", 시험, "/ 결과 폴더:", 평가.결과폴더)
# (제출본에서 바꿈) 이 칸의 FOLLOW_TEST 시험 모드 → 0절 시험_기간·시험 결과 폴더 → 결과/시험결과
```

### 3. 설정
원문 p.18·23 시험 조건 그대로. 원문에 없는 칸(10절)은 기본 방법. 재무시점(`재무시점`)은 `재무주기=("발표값",)`에서 쓰이지 않는다(발표된 달의 월말, 0절).

```python
%%절 5
기본 = {**포.기본설정, "이름": "NH퀄리티로우볼", "시장": ["KS"], "기간": 기간, "기준일": "마지막거래일", "리밸런싱": 3,
       "가격": 읽기.수정주가, "재무주기": ("발표값",),
       "제외": ["보통주", ("거래정지", {"아이템": 읽기.정지여부}), "관리종목"],
       "고르기": ("분위", 5), "비중": ["동일"], "거래비용": dict(수수료=0, 거래세=0, 슬리피지=0)}
지표들 = {"ROE": ("ROE", "높게"), "영업이익률": ("영업이익률", "높게"), "부채비율": ("부채비율", "낮게"),
         "총자산회전율": ("총자산회전율", "높게"), "로우볼": ("변동성_원문", "낮게")}  # 원문 p.23 '일간수익률의12개월변동성하위'(상장 기간 조건 없음)
퀄리티 = ["ROE", "영업이익률", "부채비율", "총자산회전율"]
분위 = [f"Q{k}" for k in range(1, 6)]
코스피 = 읽기.지수()["I.001"].reindex(포.기준일(기본))  # 리밸런싱 기준일(월말) KOSPI 종가지수: 원문 그림 26·32 'KOSPI(우)'
코스피월 = 코스피.pct_change(fill_method=None)
누적값 = lambda r: (1 + r.dropna()).prod() - 1 if r.notna().any() else np.nan  # 값이 없는 구간은 빈칸(0으로 안 보임)
해누적 = lambda r, y: (1 + r[r.index.year == y].dropna()).prod() - 1 if (r.index.year == y).any() else np.nan  # 그해 1~12월 월수익률 누적(그해가 없으면 빈칸)
pd.Series(기본).astype(str).to_frame("공통 설정")
```

### 4. 실행: 지표별 1~5분위 (그림 22~25)
원문 p.19 그림 22~25: 지표별 1~5분위 누적 수익률(%).

```python
%%절 5
결과 = {n: 포.실행({**기본, "지표": [m]}) for n, m in 지표들.items()}
수익 = {n: o["수익률"]["동일"] for n, o in 결과.items()}
누적 = pd.DataFrame({n: (1 + r).prod() - 1 for n, r in 수익.items()}).T
for i, n in enumerate(퀄리티):
    평가.누적그림(수익[n][분위].rename(columns=lambda g: f"{g[1]}분위"), f"그림 {22 + i}. {n} 지표 기준 퀄리티 포트폴리오 성과 추이", 퍼센트=True)
누적
```

### 5. ROE·로우볼 L/S와 KOSPI (그림 26·32), 2008년
- 원문 p.19 그림 26 'ROE 기준퀄리티Long/Short 포트폴리오는시장하락시기방어적인효과를보임'(L/S 좌축 %, KOSPI 우축 P), p.18 '2008년금융위기에도양호한성과를기록하는등, 시장과낮은상관관계를보였다'.
- 원문 p.24 그림 32 'KOSPI 로우볼Long/Short 포트폴리오성과추이'(L/S 좌축 %, KOSPI 우축 P), p.23 '2008년금융위기에로우볼Long/Short 포트폴리오는시장에대한헤지역할을수행'.
- 2008년 = 2008년 1~12월 월수익률 누적. 상관 = L/S 월수익률과 KOSPI 월수익률의 상관(전 기간). 원문은 상관 숫자를 주지 않아 숫자만 보인다.

```python
%%절 5
LS = pd.DataFrame({n: 수익[n]["Q1"] - 수익[n]["Q5"] for n in ["ROE", "로우볼"]})
평가.누적그림(LS[["ROE"]].set_axis(["1분위-5분위 Long/Short(좌)"], axis=1), "그림 26. ROE 기준 퀄리티 Long/Short와 KOSPI", 퍼센트=True, 오른쪽={"KOSPI": 코스피})
평가.누적그림(LS[["로우볼"]].set_axis(["1분위-5분위 Long/Short(좌)"], axis=1), "그림 32. KOSPI 로우볼 Long/Short 포트폴리오 성과 추이", 퍼센트=True, 오른쪽={"KOSPI": 코스피})
표2008 = pd.DataFrame({"2008년 누적": {"ROE L/S": 해누적(LS["ROE"], 2008), "로우볼 L/S": 해누적(LS["로우볼"], 2008), "KOSPI": 해누적(코스피월, 2008)},
                       "KOSPI 월수익률과 상관(전 기간)": LS.corrwith(코스피월).rename({"ROE": "ROE L/S", "로우볼": "로우볼 L/S"})})
표2008
```

### 6. 시가총액 300위 이내 (그림 27·35)
- 원문 p.20 'KOSPI 상장전체종목과시가총액상위300위이내종목의ROE 기준1분위와5분위포트폴리오성과를비교한결과유니버스에따른성과차이가미미했다'(그림 27).
- 원문 p.25 '2002년이후시가총액300위이내종목의로우볼Long/Short 포트폴리오와KOSPI 전체 Long/Short 포트폴리오의누적성과는유사했으나 2007~ 20014년까지 KOSPI 전체유니버스의로우볼팩터가다소우수'(그림 35. '20014년'은 원문 오타로 보고 2014년으로 읽음).
- 300위 = 리밸런싱일 시가총액(S410001250) 상위 300종목, 다른 거르기(보통주·거래정지·관리종목) 뒤(10절).
- L/S 누적 계산법은 원문에 글이 없어 매달 Q1 − Q5 차이를 복리로 쌓는다(5·7·8절도 같음, 10절). 원문 그림 27의 'KOSPI 전체' 끝값(눈대중 약 230%)은 그림 22의 1분위 누적 − 5분위 누적(약 510 − 280)과 맞아 보이고, 같은 ROE L/S인 그림 26(끝 약 0%)과는 원문 안에서 서로 맞지 않는다(`_검증\1106\원문사양_03.md` E절, 눈대중).

```python
%%절 5
대형 = {n: 포.실행({**기본, "지표": [지표들[n]], "제외": 기본["제외"] + [("시총상위", {"n": 300})]})["수익률"]["동일"] for n in ["ROE", "로우볼"]}
대형LS = pd.DataFrame({n: r["Q1"] - r["Q5"] for n, r in 대형.items()})
평가.누적그림(pd.DataFrame({"1분위-5분위 L/S(KOSPI 전체종목)": LS["ROE"], "1분위-5분위 L/S(시가총액 300위 이내)": 대형LS["ROE"]}), "그림 27. ROE 기준 L/S: KOSPI 전체 대 시가총액 300위 이내", 퍼센트=True)
평가.누적그림(pd.DataFrame({"시가총액 300위 이내 L/S": 대형LS["로우볼"], "KOSPI 전체 L/S": LS["로우볼"]}), "그림 35. 로우볼 L/S: KOSPI 전체 대 시가총액 300위 이내", 퍼센트=True)
대형표 = pd.DataFrame({f"{n} L/S": {"KOSPI 전체 누적": 누적값(LS[n]), "300위 이내 누적": 누적값(대형LS[n]),
                                    "KOSPI 전체 2007~2014 누적": 누적값(LS[n].loc["2007":"2014"]), "300위 이내 2007~2014 누적": 누적값(대형LS[n].loc["2007":"2014"])} for n in ["ROE", "로우볼"]})
대형표
```

### 7. 리밸런싱 주기 (그림 29·37)
- 원문 p.21 'KOSPI 대상으로3개월, 6개월단위리밸런싱에따른퀄리티팩터의성과차이를분석한결과차이가거의나타나지않았다'(그림 29: 총자산회전율 1분위·5분위, 3·6개월).
- 원문 p.26 '3개월, 6개월, 12개월단위리밸런싱을하는로우볼Long/Short 포트폴리오성과를비교한결과3개월단위리밸런싱에비해6개월, 12개월단위리밸런싱의누적성과가우수했다. 다만, 6개월과12개월단위리밸런싱포트폴리오는유사한성과'(그림 37).
- 6·12개월도 같은 기간 첫 달(2002-03)부터 센다(10절).

```python
%%절 5
회전 = {k: 수익["총자산회전율"] if k == 3 else 포.실행({**기본, "지표": [지표들["총자산회전율"]], "리밸런싱": k})["수익률"]["동일"] for k in (3, 6)}
볼주기 = {k: 수익["로우볼"] if k == 3 else 포.실행({**기본, "지표": [지표들["로우볼"]], "리밸런싱": k})["수익률"]["동일"] for k in (3, 6, 12)}
평가.누적그림(pd.DataFrame({"1분위(3개월 리밸런싱)": 회전[3]["Q1"], "1분위(6개월 리밸런싱)": 회전[6]["Q1"], "5분위(6개월 리밸런싱)": 회전[6]["Q5"], "5분위(3개월 리밸런싱)": 회전[3]["Q5"]}),
            "그림 29. 리밸런싱 주기에 따른 퀄리티 포트폴리오(총자산회전율)", 퍼센트=True)
볼주기LS = pd.DataFrame({f"{k}개월 리밸런싱": r["Q1"] - r["Q5"] for k, r in 볼주기.items()})
평가.누적그림(볼주기LS, "그림 37. 리밸런싱 주기에 따른 로우볼 Long/Short 누적 성과", 퍼센트=True)
display(pd.DataFrame({f"{k}개월": {"1분위 누적": 누적값(r["Q1"]), "5분위 누적": 누적값(r["Q5"])} for k, r in 회전.items()}).rename_axis("총자산회전율"))
(1 + 볼주기LS).prod().sub(1).rename("로우볼 L/S 누적").to_frame()
```

### 8. 변동성 산출 기간·주기 (그림 33·34)
- 원문 p.23 '최근12개월주가변동성대신최근24개월, 36개월일간수익률의변동성을기준으로로우볼팩터성과를비교한결과주가변동성산출기간에따른영향은거의없었다'(그림 33).
  **원문 안에서 어긋남**: 본문은 '24개월, 36개월'이고 그림 33 범례는 '12개월변동성·36개월변동성·60개월변동성'이다. 본문 문장이 기준(12·24·36개월)이고, 그림 범례도 원문이라 60개월을 더해 12·24·36·60개월을 모두 돌린다(`_검증\1106\원문사양_03.md` C절).
- 원문 p.23 '일간수익률대신주간수익률, 월간수익률기준으로백테스트한결과도유사한결과'(그림 34: Daily·Weekly·Monthly 주가수익률 변동성). 주간·월간은 직전 12개월.
- `지표.변동성_창(s, 개월, 주기, 상장조건=False)`(공통): 창 안에 있는 상장 가격(시장구분이 코스피·코스닥인 날)으로만 표준편차, 창 내내 상장 같은 조건 없음(12개월과 같은 규칙). 주간 = 금요일 끝 가격, 월간 = 월말 가격(10절). 12개월·일간이면 `지표.변동성_원문`과 같은 값.

```python
%%절 5
for k in (24, 36, 60):  # 실행이 지표 이름(문자열)으로 함수를 찾으므로 이 노트북 안에서만 이름을 붙인다(공통 파일은 그대로)
    setattr(지표, f"변동성_{k}개월", lambda s, k=k: 지표.변동성_창(s, k, "일", 상장조건=False))
for p in ("주", "월"):
    setattr(지표, f"변동성_{p}간", lambda s, p=p: 지표.변동성_창(s, 12, p, 상장조건=False))
볼실행 = lambda m: 포.실행({**기본, "지표": [(m, "낮게")]})["수익률"]["동일"]
창LS = pd.DataFrame({"12개월(본문)": LS["로우볼"], **{f"{k}개월": (lambda r: r["Q1"] - r["Q5"])(볼실행(f"변동성_{k}개월")) for k in (24, 36, 60)}})
주기LS = pd.DataFrame({"Daily": LS["로우볼"], **{n: (lambda r: r["Q1"] - r["Q5"])(볼실행(f"변동성_{p}간")) for n, p in (("Weekly", "주"), ("Monthly", "월"))}})
평가.누적그림(창LS, "그림 33. 변동성 산출 기간에 따른 로우볼 Long/Short(본문 12·24·36개월, 범례 60개월)", 퍼센트=True)
평가.누적그림(주기LS, "그림 34. 일간·주간·월간 변동성 기준 로우볼 Long/Short", 퍼센트=True)
pd.concat({"산출 기간": (1 + 창LS).prod() - 1, "수익률 주기": (1 + 주기LS).prod() - 1}).rename("로우볼 L/S 누적").to_frame()
```

### 9. 원문 vs 우리

숫자 기준이 없어 판정하지 않고 숫자만 보이는 문장: '시장과낮은상관관계'(p.18, 5절 상관), 시총 300위 ROE '성과차이가미미'(p.20, 6절), 퀄리티 3·6개월 '차이가거의나타나지않았다'(p.21, 7절), 변동성 산출 기간·주기 '영향은거의없었다'(p.23, 8절), '6개월과12개월…유사한성과'(p.26, 7절), 로우볼 300위·전체 '누적성과는유사'(p.25, 6절).

```python
%%절 5
최고 = {n: 누적.loc[n, 분위].idxmax() for n in 퀄리티}
최저 = {n: 누적.loc[n, 분위].idxmin() for n in 퀄리티}
볼누적 = (1 + 볼주기LS).prod() - 1
변형누적 = pd.concat([(1 + 창LS).prod() - 1, (1 + 주기LS).prod() - 1])
판정표 = pd.DataFrame({
    "1. 4개 퀄리티 지표 모두 2~4분위 누적이 가장 높음": ("p.18", all(g in ["Q2", "Q3", "Q4"] for g in 최고.values()), "누적 최고 " + ", ".join(f"{n} {g}" for n, g in 최고.items())),
    "2. 4개 퀄리티 지표 모두 5분위 누적이 가장 낮음": ("p.18", all(g == "Q5" for g in 최저.values()), "누적 최저 " + ", ".join(f"{n} {g}" for n, g in 최저.items())),
    "3. 2008년 KOSPI < 0 이고 ROE L/S > 0(양호)": ("p.18·19", bool(표2008.loc["KOSPI", "2008년 누적"] < 0 and 표2008.loc["ROE L/S", "2008년 누적"] > 0),
                                           f"ROE L/S {표2008.loc['ROE L/S', '2008년 누적']:.4f}, KOSPI {표2008.loc['KOSPI', '2008년 누적']:.4f}"),
    "4. 로우볼 L/S 누적 > 0": ("p.23", bool(누적값(LS["로우볼"]) > 0), f"{누적값(LS['로우볼']):.4f}"),
    "5. 2008년 KOSPI < 0 이고 로우볼 L/S > 0(헤지)": ("p.23·24", bool(표2008.loc["KOSPI", "2008년 누적"] < 0 and 표2008.loc["로우볼 L/S", "2008년 누적"] > 0),
                                                 f"로우볼 L/S {표2008.loc['로우볼 L/S', '2008년 누적']:.4f}, KOSPI {표2008.loc['KOSPI', '2008년 누적']:.4f}"),
    "6. 2007~2014 로우볼 L/S 누적: 전체 > 300위 이내": ("p.25", bool(대형표.loc["KOSPI 전체 2007~2014 누적", "로우볼 L/S"] > 대형표.loc["300위 이내 2007~2014 누적", "로우볼 L/S"]),
                                                    f"전체 {대형표.loc['KOSPI 전체 2007~2014 누적', '로우볼 L/S']:.4f}, 300위 {대형표.loc['300위 이내 2007~2014 누적', '로우볼 L/S']:.4f}"),
    "7. 300위 이내 로우볼 L/S 누적 > 0": ("p.25·26", bool(대형표.loc["300위 이내 누적", "로우볼 L/S"] > 0), f"{대형표.loc['300위 이내 누적', '로우볼 L/S']:.4f}"),
    "8. 로우볼 L/S 누적: 6·12개월 리밸런싱 > 3개월": ("p.26", bool(min(볼누적["6개월 리밸런싱"], 볼누적["12개월 리밸런싱"]) > 볼누적["3개월 리밸런싱"]),
                                                  ", ".join(f"{k} {v:.4f}" for k, v in 볼누적.items())),
    "9. 변동성 12·24·36·60개월, 일·주·월간 모두 로우볼 L/S 누적 > 0": ("p.23·26", bool((변형누적 > 0).all()), ", ".join(f"{k} {v:.4f}" for k, v in 변형누적.items())),
}, index=["원문 쪽", "우리", "우리 숫자"]).T.rename_axis("원문 주장")
판정표.to_csv(평가.결과폴더 / "03_원문판정.csv", encoding="utf-8-sig")  # 결과요약 노트북이 읽음
판정표
```

### 10. 원문이 말하지 않아 기본 방법으로 정한 것
원문에 말이 없는데 빼면 계산이 안 되는 칸. 기본 방법 ① 자료가 없거나 계산이 안 되면 그 종목을 그 달에서 뺀다(우동호 p.512 '필요한 재무자료를 입수 가능한 기업' 원리) ② 나머지는 가장 단순한 쪽(회원 결정 2026-11-06).

| 항목 | 원문에 없음(확인한 곳) | 쓴 기본 방법 | 칸 |
|---|---|---|---|
| 리밸런싱 달 | p.18·23은 '3개월단위월말리밸런싱'과 '2002년3월~'뿐, 몇 월인지 없음(PDF 전체 검색) | ② 기간 첫 달 2002-03 월말부터 3개월마다 = 3·6·9·12월 말 | 3절 `리밸런싱=3`, 2절 `기간` |
| 기간 시작 해석 | '2002년3월'이 첫 리밸런싱 달인지 첫 수익률 달인지 없음(그림 가로축 '02.3부터) | ② 2002-03 월말 첫 리밸런싱(첫 수익률 달 2002-04) | 2절 |
| 6·12개월 리밸런싱 시작 달 | p.21·26 주기만 | ② 3개월과 같은 2002-03 월말부터 | 7절 |
| 로우볼 가중 | p.23 시험 조건 상자에 가중 없음 | ② 동일가중(퀄리티 p.18과 같음) | 3절 `비중` |
| 변동성 최소 관측 | p.23 '일간수익률의12개월변동성하위'뿐 | ① 직전 12개월 창 안 일간수익률이 2개 미만이라 표준편차가 안 나오는 종목만 뺌. 창 안 빈칸은 빼고 표준편차. '직전 12개월 내내 상장' 조건은 원문에 없어 걸지 않음(05:47 지시로 뺌) | `지표.변동성_원문` |
| 거래정지 날 수익률 | 원문 없음 | ② 가격 그대로(정지 날 가격이 그대로면 수익률 0이 들어감) | 공통 `포.재료` |
| ROE 정의 | p.18 이름만 | ② 지배주주순이익(M000908004) / 지배주주자본(M000903003), 같은 보고서 값 | 공통 `지표.ROE` |
| 영업이익률 정의 | p.18 이름만 | ② 영업이익(M000906001) / 매출액(M000904001) | 공통 `지표.영업이익률` |
| 부채비율 정의 | p.18 이름만(방향은 '로우퀄리티' = 5분위) | ② 부채총계(M000902001) / 자본총계(M000903001), 낮을수록 1분위 | 공통 `지표.부채비율` |
| 총자산회전율 정의 | p.18 이름만 | ② 매출액 / 자산총계(보고서 기간말), 같은 보고서 값 | 공통 `지표.총자산회전율` |
| 음수 자본 | 원문 없음 | ② 공식 그대로(자본이 음수면 ROE·부채비율 부호가 뒤집힘). 분모 0(±무한대)만 ①로 뺌 | 공통 `포.실행` |
| 발표일 없는 보고서 | 원문 없음 | ① 안 씀(그 보고서를 건너뛰고 그 전에 발표된 보고서를 씀, 없으면 그 종목을 뺌) | 공통 `포.발표행` |
| 묵은 재무 상한 | 원문 없음 | ② 상한 없음(가장 최근 발표 보고서) | 공통 `포.발표재무(상한=None)` |
| 같은 날 발표된 보고서 둘 | 원문 없음 | ② 기간 끝이 늦은 보고서 | 공통 `포.발표재무` |
| 리밸런싱 사이 비중 | 원문 없음 | ② 사서 그대로 둠(매수보유) | 공통 `포.보유` |
| 보유 중 상장폐지 | p.18 생존편향 문장뿐 | ② 마지막 상장 가격으로 판 것으로 봄 | 공통 `포.기간수익률` |
| 분위 나누기·같은 값 | p.18 '1분위에서5분위'만 | ② 순위('first')로 종목 수를 같게 5등분 | 공통 `포.고르기` |
| 거래비용 | 시험 조건 상자(p.18·23)에 없음 | ② 0 | 3절 `거래비용` |
| L/S 누적 | p.18 'Long…Short'만(계산법 글 없음) | ② 매달 Q1 − Q5 수익률을 복리로 쌓음. 원문사양 E절: 그림 27 끝값은 'Q1 누적 − Q5 누적'과 맞아 보임(눈대중), 그림 26은 그림 27과 서로 안 맞음 | 5·6·7·8절 |
| 시총 300위 순서·날짜 | p.20·25 '시가총액상위300위이내'만 | ② 리밸런싱일 시가총액, 다른 거르기 뒤 상위 300 | 6절 |

### 11. 차이 이유 후보
- **재무 금액**: 발표된 보고서 값 그대로라 분기 보고서(1·2·3분기) 행의 순이익·영업이익·매출은 3개월 값, 사업보고서 행은 1년 값이다. 같은 리밸런싱일에 종목마다 최근 보고서 종류가 다르면(예: 3월 말에 사업보고서를 아직 안 낸 종목) ROE·총자산회전율의 크기가 서로 다른 기간 값으로 비교된다. 원문이 어떻게 했는지는 원문에 없다.
- **리밸런싱 달**: 3·6·9·12월 말이라 5·8·11월에 나온 분기 보고서는 한 달 뒤 리밸런싱에 들어간다(10절).
- **ROE·부채비율 정의**: 원문에 분모(기말/평균, 지배/전체) 정의가 없다(10절).
- **음수 분모**: ROE·부채비율은 공식 그대로라 자본이 음수인 종목은 부호가 뒤집혀 좋은 쪽에 줄 설 수 있다(값은 고치지 않음).
- **변동성**: 원문은 12개월 일간 변동성뿐이라 창 안에 있는 값으로 표준편차를 낸다(상장 기간 조건 없음). 상장한 지 얼마 안 된 종목은 적은 관측으로 낸 변동성으로 줄 선다.
- **자료 판본**: 원문(2018)은 그때 FnGuide 자료, 우리는 DataGuide 2026-11 판(재무는 최근 수정치).
- **벤치마크**: 그림의 KOSPI는 가격지수(I.001, 배당 미포함), 포트폴리오 수익률도 수정주가(배당 미포함)라 같은 기준이다.

## 04 Novy-Marx(2013) 결합 전략 (GP/A + B/M)

### 0. 원문 정리
원문: Novy-Marx, R., "The Other Side of Value: Good Growth and the Gross Profitability Premium", NBER Working Paper 15940(2010년 4월판, 파일 `참고논문\2_수익성_NovyMarx_2013_TheOtherSideOfValue_NBERw15940.pdf`, 파일 이름은 2013이지만 표지는 April 2010. 학술지판 2013 JFE의 표 번호·숫자는 확인 안 함). 쪽은 논문에 찍힌 쪽. 원문 확인 기록 `_검증/1106/원문사양_04.md`.

> 회원 결정(2026-11-06): 04 따라하기 = Novy-Marx(2013) 결합 전략(원문 있음)만. 강환국 신마법공식 1.0·2.0은 책 원문을 확인 못 해 확장연구 `연구/따라하기에서_옮김_04.ipynb`로 옮김 — Novy-Marx 결합 전략을 국내 실무에서 바꾼 판, 책 원문 미확인, 2차 자료 intelliquant 573·578로 구현. 같은 노트북 설정으로 돌리던 우동호 IMF 나란히 보기, 시가총액가중 판, 연환산 등 원문에 없는 성과 칸, 이상값 칸, 로그 축 그림, 기록 표, 예전 판 → 원문 규칙 단계별 진단(부호 판정 규칙), 2차 자료 확인판 K1-K5도 같은 곳으로 옮김.
> 한국 근거(참고문헌 후보): 윤보현·유원석(Yoon·Liu, 2015), 유통과학연구 13(9), 37-45, GP/A × B/M 5×5(이 노트북에서 돌리지 않음).

원문(PDF에서 글자를 뽑아 2026-11-06 다시 확인, 쪽은 논문에 찍힌 쪽):
- p.21(4.3.1) "The strategy I consider is constructed within the 500 largest non-financial stocks for which gross profits-to-assets and book-to-market are both available. Each year I rank these stocks … At the end of each June the strategy buys one dollar of each of the 150 stocks with the highest average of the profitability and value ranks, and shorts one dollar of each of the 150 stocks with the lowest average ranks."
- p.21 "average excess returns of 0.64 percent per month, and has a realized annual Sharpe ratio of 0.76", "These large firms returned on average 0.41 percent per month …, so the strategy makes 58 percent of its profits on the long side, and 42 percent on the short side". p.23 "Only one-third of each side of the strategy turns over each year."
- p.22 표 7 설명 "the sum of the firms' GPA and BM ranks within the sample (Panel C) … time-series average portfolio characteristics (portfolio GPA, portfolio BM, average firm size (ME, in $10^6), and number of firms (n)). The sample covers July 1963 to December 2009."
- p.11 각주 3 "market equity is lagged six months … Book equity is shareholder equity, plus deferred taxes, minus preferred stock, when available." 각주 4 "I lag accounting data to the end of June of the following year". p.12 표 2 설명 GP = "revenues minus cost of goods sold, Compustat REVT - COGS … scaled by assets (AT)". 비금융 = "one-digit SIC codes of six"(p.8·p.11). SIC 첫 자리 6 = Division H 'Finance, Insurance, And Real Estate'(SIC 60~67). 그 안 Major Group 67 'Holding And Other Investment Offices'에 6798 'Real Estate Investment Trusts'(OSHA SIC 편람 https://www.osha.gov/sic-manual/6798), 6770 'Blank Checks'(스팩, SEC SIC 코드 목록)가 듦(원문 밖 자료, 2026-11-06 웹으로 확인). 무위험·FF3 요인의 출처는 4.3.1·표 7에 문장 없음(p.41 "Ken French's data library"는 원문 7절 표 15 요인 이야기, `_검증/1106/원문사양_04.md`).

우리(2026-11-06 원문대로 고침): 6월 마지막 거래일, **보통주만**(회원 결정) → 비금융(`"금융업"` 넓은 이름 규칙 + `"투자기구"`: 위 SIC 6 범위에 맞춰 금융 낱말·업종 빈 날 이름 규칙(리츠·스팩·'지주'로 끝남 포함)·WICS '금융' + 리츠·선박·부동산투자회사, 조정자 결정 2026-11-06 06:2x) → GP/A·B/M 둘 다 있음 → 시총 상위 500, 위 150 / 가운데 200 / 아래 150(`상하N`), 순위합산(평균과 순서 같음), 동일가중. 같은 500종목에서 GP/A 순위만(Panel A)·B/M 순위만(Panel B)으로도 같은 방법으로 고른다(p.21 "The table also shows, for comparison, the performance of similarly constructed strategies based on profitability and value individually", p.22 표 7 Panel A·B).
- GP/A = 매출총이익 M000904007 / 자산총계 M000901001(`지표.GPA`).
- B/M = **원문 장부가** `BP_장부원문_6개월`: (자본 + 이연법인세부채 M000911019 − 우선주자본금 M000903008) / 6개월 전 시가총액. 자본 = 자본총계(지배) M000903003, 빈칸이면 자산총계 − 부채총계(각주 3의 SEQ → … → AT − LT, 가운데 CEQ + PSTX는 대응 항목이 없어 건너뜀). 이연법인세·우선주가 빈 종목은 그 항만 뺌('when available'). 장부가가 음수인 종목도 뺄 문장이 원문에 없어 그대로 남김.
- 재무 = 결산 후 5개월이 지났고 결산일이 전 해인 회계연도(`재무시점=("결산후개월",5)`, `재무연도="직전년도"`: 각주 4 '다음 해 6월 말부터'. 전 해 결산이 없으면 그 종목은 그해 빠짐 = 기본 방법 ①).
- 수익률 가격 = 수정주가 S410000700(회원 결정, 배당 없음). 거래비용 0(원문에 없음, 기본 방법 ②), 무위험 = 콜금리(E11.02.003.001, 원문에 출처 없음 → 우리가 고른 값, 기존 기본설정 그대로), 요인 = DataGuide 한국 요인(3FM.2B3), t = OLS(회원 결정). 원문에 없어 기본 방법으로 정한 칸은 6절 표.
- 기간(3절 코드가 구함): 시작 = 시총 500 앞까지 거른 종목(직전년도 GP/A·6개월 전 시총 B/M이 있는 비금융 보통주)이 500개 이상인 첫 6월(기본 방법 ②), 끝 = 자료 끝(`읽기.공통끝()`) 안에서 1년 보유가 다 들어가는 마지막 6월(기본 방법 ②). 원문은 마지막 해가 불완전함(표본 1963-07~2009-12, 2009-06 편입분은 6개월만).
- 원문 표 7의 Panel A(GPA만)·Panel B(BM만)도 원문에 있어 같은 표본·방법으로 돌려 대조함(05:47 지시, 원문 p.21 'for comparison'). ME는 원문 단위가 백만 달러라 대조하지 않음.

### 3. 설정과 실행
원문 정의대로(0절). 기간은 이 칸이 자료로 구한다(0절 마지막 줄). 원문에 없어 기본 방법으로 정한 칸은 6절 표.


```python
%%절 6
nm = {**포.기본설정, "이름": "NovyMarx2013_표7C", "시장": ["KS", "KQ"], "기간": None, "기준일": "마지막거래일",
      "리밸런싱": 12, "재무시점": ("결산후개월", 5), "재무연도": "직전년도", "가격": 읽기.수정주가,
      "제외": ["보통주", "금융업", "투자기구", ("지표있음", {"지표들": ["GPA", "BP_장부원문_6개월"]}), ("시총상위", {"n": 500})],  # 비금융 = 원문 p.8 SIC 첫 자리 6(금융·보험·부동산 60~67: 리츠 6798·스팩 6770·투자회사 67 포함, OSHA SIC 편람) → 금융업 넓은 이름 규칙 + 투자기구
      "지표": [("GPA", "높게"), ("BP_장부원문_6개월", "높게")], "합산": "순위합산", "고르기": ("상하N", 150),
      "비중": ["동일"], "거래비용": dict(수수료=0, 거래세=0, 슬리피지=0),
      "전략들": {"A": dict(지표=[("GPA", "높게")], 합산="순위합산"), "B": dict(지표=[("BP_장부원문_6개월", "높게")], 합산="순위합산"),
               "C": dict(지표=[("GPA", "높게"), ("BP_장부원문_6개월", "높게")], 합산="순위합산")}}  # p.21 'similarly constructed strategies based on profitability and value individually'
# 기간(기본 방법 ②). 끝 = 자료 끝 안에서 6월 말 → 다음 해 6월 말 보유가 다 들어가는 마지막 해(원문은 마지막 해가 불완전함)
끝날 = 읽기.공통끝()
끝해 = max(y for y in range(1988, 끝날.year + 1) if pd.Timestamp(f"{y}-06-30") <= 끝날)
첫 = 읽기.일간(읽기.시총).index[0] + pd.DateOffset(months=6)  # B/M에 6개월 전 시가총액이 있어야 함
첫해 = 첫.year + (첫.month > 6)


def 표본수(y):  # 그해 6월 마지막 거래일에 시총 500 앞까지 거른 종목 수(직전년도 GP/A·B/M이 둘 다 있는 비금융 보통주)
    c = {**nm, "기간": (f"{y}-06", f"{y}-06")}
    return len(포.거르기(포.기준일(c)[0], {**c, "제외": nm["제외"][:-1]})[1])


수 = {}
for y in range(첫해, 끝해):
    수[y] = 표본수(y)
    if 수[y] >= 500:
        break
시작해 = y
assert 수[시작해] >= 500, f"500개 이상인 6월이 없음: {수}"
nm["기간"] = (f"{시작해}-06", f"{min(끝해, 시작해 + 2) if 시험 else 끝해}-06")
print("6월별 거른 종목 수:", 수, "→ 기간:", nm["기간"], "(자료 끝", 끝날.date(), ")")
assert {"M000911019", "M000903008"} <= set(읽기.재무항목()), "원문 장부가에 쓰는 이연법인세부채·우선주자본금 파일이 없음"
o전체 = 포.실행(nm)  # 표 7 Panel A(GPA만)·B(BM만)·C(순위합) = 같은 거르기 결과(같은 500종목)에서 한 번에(`전략들`, 열 = (판, 비중, 그룹)). 유니버스 시총 열은 안 씀


def 판(p):  # Panel p만 꺼내 전략 하나로 돌린 것과 같은 꼴로(열 (비중, 그룹), 기록 '편입:Q1' 등)
    return {**o전체, "수익률": o전체["수익률"][p], "회전율": o전체["회전율"][p],
            "보유": {k[1:]: w for k, w in o전체["보유"].items() if k[0] == p},
            "기록": o전체["기록"].rename(columns=lambda x: x.replace(f":{p}:", ":"))}


Fn = 평가.요인(포.기준일(nm), nm)
판들, Ns, 표들 = {}, {}, {}
for p in "ABC":
    판들[p] = 판(p)
    N = pd.DataFrame({f"{b}_{g}": 판들[p]["수익률"][(b, g)] for b, g in 판들[p]["수익률"]})
    N["동일_H-L"] = N["동일_Q1"] - N["동일_Q3"]
    Ns[p] = N.dropna(how="all")
    표들[p] = 평가.성과표(Ns[p], Fn, ["동일_H-L"])  # 대조표(4절)가 쓰는 칸: 초과평균·FF3 α와 t(OLS), 샤프
o, N, 표N = 판들["C"], Ns["C"], 표들["C"]  # 결합 전략(Panel C) = 아래 문장 대조·판정·저장의 본체
```

### 4. 원문 대조: 표 7 Panel A·B·C(p.22)와 p.21·23 문장 숫자
원문 = 미국 1963-07~2009-12, 우리 = 한국 2절 기간. 시장·기간·요인이 달라 크기를 그대로 견줄 수 없다(5절). 판정 = 원문이 말한 방향(부호)만 같음/다름, 크기는 원문값·우리값을 나란히 적음(회원 원칙 '원문 판정은 원문이 말한 방향·크기만').
판정표(`04_NovyMarx판정.csv`)의 '원문 문장 주장' 칸: H-L 월 초과·H-L FF3 α·High − Low·롱 쪽 비중 4줄 = '있음'(p.21·표 7 문장), Panel A·B의 H-L 월 초과·FF3 α 부호 4줄 = '없음(표 7 부호만 나란히, 참고)'(원문은 p.21 'for comparison'으로 표만 보이고 문장 주장은 없음).


```python
%%절 6
# 표 7(p.22, 미국 1963-07~2009-12) Panel A(GPA만)·B(BM만)·C(순위합)와 p.21 Sharpe(C만). Low = 아래 150(Q3), 2 = 가운데 200(Q2), High = 위 150(Q1), H-L = High − Low(동일가중)
묶음 = {"Low": "동일_Q3", "2": "동일_Q2", "High": "동일_Q1", "H-L": "동일_H-L"}
열 = ["월 초과수익(%)", "t", "FF3 α(%)", "FF3 t", "MKT", "MKT t", "SMB", "SMB t", "HML", "HML t", "GPA", "BM", "n"]
na = np.nan
원문표7 = {  # PDF 24쪽(논문 p.22)에서 옮김(2026-11-06 pymupdf 글자로 다시 확인). 줄 = Low·2·High·H-L, 열 = 위 `열`(re·[t]·α·[t]·MKT·[t]·SMB·[t]·HML·[t]·GPA·BM·n). ME는 단위($10^6)가 달라 뺌
    "A": [[0.34, 1.70, -0.17, -2.18, 1.02, 54.1, -0.03, -1.25, 0.23, 8.07, 0.13, 1.02, 150],
          [0.54, 2.37, -0.00, -0.01, 1.13, 72.7, 0.11, 5.12, 0.08, 3.48, 0.31, 0.85, 200],
          [0.63, 2.90, 0.25, 3.80, 1.03, 66.6, 0.08, 3.78, -0.17, -7.53, 0.64, 0.41, 150],
          [0.28, 2.38, 0.42, 3.90, 0.00, 0.05, 0.11, 3.20, -0.40, -10.5, na, na, na]],
    "B": [[0.35, 1.40, 0.07, 1.05, 1.10, 68.0, 0.04, 1.91, -0.49, -19.9, 0.51, 0.25, 150],
          [0.49, 2.33, -0.01, -0.22, 1.07, 78.7, 0.07, 3.66, 0.09, 4.33, 0.34, 0.58, 200],
          [0.68, 3.49, 0.02, 0.28, 1.02, 71.9, 0.06, 3.06, 0.53, 24.9, 0.22, 1.54, 150],
          [0.34, 2.22, -0.06, -0.72, -0.08, -4.49, 0.02, 0.68, 1.02, 37.0, na, na, na]],
    "C": [[0.15, 0.59, -0.22, -2.43, 1.13, 52.5, 0.01, 0.26, -0.27, -8.27, 0.22, 0.45, 150],
          [0.57, 2.89, 0.10, 1.90, 1.01, 85.3, 0.03, 1.53, 0.10, 5.31, 0.38, 0.68, 200],
          [0.78, 3.68, 0.16, 2.69, 1.08, 75.1, 0.15, 7.72, 0.30, 13.8, 0.45, 1.20, 150],
          [0.64, 5.21, 0.39, 3.77, -0.05, -1.92, 0.15, 4.38, 0.56, 15.6, na, na, na]]}
판이름 = {"A": "A GPA만", "B": "B BM만", "C": "C 순위합"}
특성값 = {}  # 포트폴리오 GPA·BM: 리밸런싱일마다 보유 종목 값의 평균 → 시점 평균(원문 'time-series average portfolio characteristics', 종목 평균인지 가중인지는 원문에 없음 → 동일 평균). 재료는 세 판이 같아 한 번만
for t in sorted(o["보유"][("동일", "Q1")]):
    s = 포.재료(t, o["설정"])
    특성값[t] = pd.DataFrame({"GPA": 지표.GPA(s), "BM": 지표.BP_장부원문_6개월(s)})
대조 = {}
for p in "ABC":
    op, Np, 표p = 판들[p], Ns[p], 표들[p]
    원문 = pd.DataFrame(원문표7[p], index=list(묶음), columns=열).assign(**{"연 Sharpe": [na, na, na, 0.76 if p == "C" else na]})
    ff3 = {k: 평가.회귀(Np[c] - (0 if k == "H-L" else Fn["RF"]), Fn[["MKT", "SMB", "HML"]])[2] for k, c in 묶음.items()}  # 성과()의 FF3 회귀와 같은 식, 적재값·t를 꺼내려고
    특성 = pd.DataFrame([{"묶음": k, **v.reindex(op["보유"][("동일", 묶음[k][3:])][t].index).mean()} for t, v in 특성값.items() for k in ("Low", "2", "High")]).groupby("묶음").mean()
    우리 = pd.DataFrame({k: {"월 초과수익(%)": 표p.loc[c, "초과평균 α(월)"] * 100, "t": 표p.loc[c, "초과평균 t(OLS)"],
                           "FF3 α(%)": 표p.loc[c, "FF3 α(월)"] * 100, "FF3 t": 표p.loc[c, "FF3 t(OLS)"],
                           **{x: ff3[k].params[x] for x in ("MKT", "SMB", "HML")}, **{f"{x} t": ff3[k].tvalues[x] for x in ("MKT", "SMB", "HML")},
                           "GPA": 특성["GPA"].get(k, np.nan), "BM": 특성["BM"].get(k, np.nan),
                           "n": op["기록"][f"편입:{c[3:]}"].mean() if k != "H-L" else np.nan,
                           "연 Sharpe": 표p.loc[c, "샤프"] if k == "H-L" else np.nan} for k, c in 묶음.items()}).T  # Sharpe = 월 평균/표준편차 × √12(원문 'realized annual' 계산법 없음 → 가장 단순)
    대조[판이름[p]] = pd.concat({"원문(미국 1963-07~2009-12)": 원문, "우리": 우리}, axis=1).swaplevel(axis=1)[list(원문)]
우리 = 대조["C 순위합"].xs("우리", axis=1, level=1)  # 아래 문장 대조·판정이 쓰는 결합 전략 값
NM대조 = pd.concat(대조, names=["판", "묶음"])
NM대조.loc["C 순위합"].rename_axis(None).to_csv(평가.결과폴더 / "04_NovyMarx대조.csv", encoding="utf-8-sig")  # 결과요약 노트북이 읽음(열 2단: 항목 × 원문/우리, 줄 = 묶음). 예전과 같은 꼴 = Panel C만
NM대조.to_csv(평가.결과폴더 / "04_NovyMarx대조_표7.csv", encoding="utf-8-sig")  # Panel A·B·C 다(줄 2단: 판 × 묶음)
NM대조
```

```python
%%절 6
# p.21·p.23 문장의 숫자('약 8%' = 원문 'almost eight percent per year', 0.64 × 12 = 7.68). '500종목 평균'은 세 묶음 월 초과수익률 평균을 종목 수(n)로 가중한 값(500종목 동일가중 포트폴리오를 따로 돌리지 않은 근사)
a = 우리.loc[["Low", "2", "High"]]
전체 = (a["월 초과수익(%)"] * a["n"]).sum() / a["n"].sum()
롱 = (우리.loc["High", "월 초과수익(%)"] - 전체) / 우리.loc["H-L", "월 초과수익(%)"]
회전 = o["회전율"][[("동일", "Q1"), ("동일", "Q3")]].iloc[1:].mean()  # 한쪽 회전율 = 리밸런싱 때 산 비중 합, 첫 매수(=1) 뺌
NM문장 = pd.DataFrame({"원문": [8, 0.41, 0.58, 0.42, 1 / 3, 1 / 3], "우리": [우리.loc["H-L", "월 초과수익(%)"] * 12, 전체, 롱, 1 - 롱, 회전.iloc[0], 회전.iloc[1]],
                     "원문 쪽": ["p.21 'almost eight percent per year'", "p.21", "p.21", "p.21", "p.23", "p.23"]},
                    index=["H-L 연 초과수익(%, 월 × 12)", "500종목 평균 월 초과수익(%)", "롱 쪽 이익 비중", "숏 쪽 이익 비중", "롱 쪽 한 해 회전율", "숏 쪽 한 해 회전율"])
NM문장.to_csv(평가.결과폴더 / "04_NovyMarx문장대조.csv", encoding="utf-8-sig")
NM문장
```

```python
%%절 6
# 원문이 말한 방향만 판정(부호), 크기는 나란히
HL, H, L = (우리.loc[k, "월 초과수익(%)"] for k in ("H-L", "High", "Low"))
a = 우리.loc["H-L", "FF3 α(%)"]
판정 = pd.DataFrame([("H-L 월 초과수익(%) > 0", "p.21·표 7", 0.64, HL, HL > 0),
                   ("H-L FF3 α(%) > 0", "표 7", 0.39, a, a > 0),
                   ("High 월 초과수익 − Low(%) > 0", "표 7(0.78 − 0.15)", 0.78 - 0.15, H - L, H > L),
                   ("롱 쪽 이익 비중 > 숏 쪽(0.5 초과)", "p.21(58% 대 42%)", 0.58, 롱, 롱 > 0.5)]
                  + [(f"Panel {p} H-L {x} {'>' if v > 0 else '<'} 0", f"표 7 Panel {p}", v, 대조[판이름[p]].loc["H-L", (x, "우리")], (대조[판이름[p]].loc["H-L", (x, "우리")] > 0) == (v > 0))
                     for p in "AB" for x in ("월 초과수익(%)", "FF3 α(%)") for v in [대조[판이름[p]].loc["H-L", (x, "원문(미국 1963-07~2009-12)")]]],  # 표 7 Panel A·B의 부호(p.21 'for comparison', 문장 주장은 없음. B의 α −0.06은 t −0.72)
                  columns=["원문 주장", "원문 쪽", "원문값", "우리값", "같은 방향"]).set_index("원문 주장")
판정["판정(방향)"] = np.where(판정.pop("같은 방향"), "같음", "다름")
판정["원문 문장 주장"] = np.where(판정.index.str.startswith("Panel "), "없음(표 7 부호만 나란히, 참고)", "있음")  # V04b: Panel A·B는 p.21 'for comparison'뿐
판정.to_csv(평가.결과폴더 / "04_NovyMarx판정.csv", encoding="utf-8-sig")
판정
```

```python
%%절 6
# 요약표용 저장: 동일가중 Low(Q3)·2(Q2)·High(Q1)·H-L 월수익률(표 7 Panel C 결합 + Panel A GPA만·B BM만), 요인 Fn, H-L은 롱숏(초과수익 = 그대로)
R = pd.DataFrame({(f"NovyMarx{'' if p == 'C' else '_' + {'A': 'GPA', 'B': 'BM'}[p]}_{k}", "동일"): Ns[p][c] for p in "CAB" for k, c in 묶음.items()})  # C = 예전 이름 그대로, Panel A·B = NovyMarx_GPA_*·NovyMarx_BM_*
평가.저장(R, Fn, "04_NovyMarx", 롱숏=[(f"NovyMarx{x}_H-L", "동일") for x in ("", "_GPA", "_BM")])
list(R)
```

### 5. 차이 이유 후보
- **Novy-Marx 결합 전략**: 시장·기간(미국 1963~2009 vs 한국), 비금융 판정(SIC 첫 자리 6 대신 거래소 업종 낱말·이름 규칙·WICS·투자기구 이름, SIC 6과 1:1 아님), 요인·무위험(원문 4.3.1·표 7에 출처 문장 없음, 미국 자료일 것)은 한국 자료(DataGuide 요인·콜금리)로 대신해 바꿀 수 없음. 장부가(각주 3)·보통주·Low·2 줄은 2026-11-06 원문대로 고침, B/M 6개월 전 시가·150/200/150·재무 시점은 2026-11-04 고침(단계별 진단은 `연구/따라하기에서_옮김_04.ipynb`로 옮김).
- **수익률 가격**: Novy-Marx 판은 수정주가(회원 결정; 원문 4.3.1·표 7에는 수익률 가격 문장 없음).

### 6. 원문과 다르게 한 것
| 항목 | 원문·정의 | 우리 | 이유 |

# 7. 확장 연구와 이로운 연구
## 7.1 확장 연구
저장소 `연구/확장연구_20261103.ipynb`의 칸(원래 import 칸과 7절 기록 표만 뺌). 리밸런싱 주기 비교도 결과를 지우지 않고 그대로 둔다.

## 확장 연구 (2026-11-03): 비중 × 리밸런싱 주기 × 지표 × 요인 순서 × 거래비용

우동호 외(2023) 유니버스에서 마법공식의 두 요인(가치·질)은 그대로 두고, **지표**, **요인을 쓰는 순서**, **비중**, **리밸런싱 주기**, **거래비용**을 바꿔 돌린다.
결과표는 전략 × 비중 × 주기 × 비용 × 기간 한 장(`연구/결과/확장연구_20261103.csv`)이고, 아래에서 그 표를 잘라 보인다.

줄임말: EW(동일가중), VW(시가총액가중), MDD(최대 낙폭), IR(정보비율: 자체 벤치마크 대비), CAPM·FF3·Carhart(요인모형), NW는 회원 결정(2026-11-06)으로 뺌, t는 OLS, TR(배당 포함 수정주가), EV(기업가치 = 시가총액 + 부채총계 − 현금).

### 1. 지표와 근거
요인마다 지표 4개. 기준(마법공식)은 EY와 ROC다. 근거 칸의 쪽수는 원문 PDF에서 확인한 것이고, 원문을 확인하지 못한 것은 '확인 안 함'이라고 적었다.
비율 지표는 수익률 꼴(이익/가격)로 높을수록 좋다(`공통/지표.py` 맨 위 설명).

```python
%%절 7.1
가치 = {"EY": "EBIT / EV. 우동호 외(2023) p.512, 그린블라트(2005)",
        "EBITDA_EV": "EBITDA / EV. NH 밸류(2018) p.14 'EV/EBITDA'",
        "EP": "지배순이익 / 시가총액. NH 밸류(2018) p.14 'P/E'",
        "영업이익_EV": "영업이익(발표기준) / EV. 인수자 배수(Acquirer's Multiple): 우동호 외(2023) 참고문헌의 Arif·Sumirat(2022), 칼라일 책. 두 원문 모두 확인 안 함"}
질 = {"ROC": "EBIT / (순운전자본 + 순고정자산). 우동호 외(2023) p.512",
      "GPA": "매출총이익 / 자산총계. 우동호 외(2023) p.513, Novy-Marx(2013)",
      "ROE": "지배순이익 / 지배자본. NH 퀄리티(2018) p.18",
      "영업CF자산": "영업활동현금흐름 / 자산총계. 근거 논문·리포트 원문 확인 안 함(가진 자료에서 못 찾음)"}
남길비율 = 0.5  # 단계별: 앞 요인으로 상위 50%를 남기고 뒤 요인으로 30종목. 근거 없음(예전 팩터순서 노트북에서 쓴 값)
pd.concat({"가치": pd.Series(가치), "질": pd.Series(질)}).rename("근거").to_frame()
```

### 2. 전략 목록과 설정
전략 = 가치 지표 × 질 지표 × 순서(순위합산 / 가치 먼저 → 질 / 질 먼저 → 가치). 지표에 드는 파일이 아직 없으면 그 지표가 든 전략은 빼고 이름을 출력한다(파일 오면 다시 실행).

```python
%%절 7.1
필요 = {"EY": ["M000907001", "M000906020", "M000902001", "M000901030"], "EBITDA_EV": ["M000701032"], "EP": ["M000908004"],
        "영업이익_EV": ["M000906002"], "ROC": ["M000901002", "M000902003", "M000901001", "M000901024"], "GPA": ["M000904007"],
        "ROE": ["M000908004", "M000903003"], "영업CF자산": ["M000909001"]}
없음 = {m for m, items in 필요.items() if not all(읽기.있음(i) for i in items)}
print("파일이 없어 뺀 지표(파일 오면 다시 실행):", sorted(없음) or "없음")
전략들 = {}
for v in 가치:
    for q in 질:
        if {v, q} & 없음:
            continue
        전략들[f"{v}+{q} 순위합산"] = dict(지표=[(v, "높게"), (q, "높게")], 합산="순위합산")
        전략들[f"{v}→{q}"] = dict(지표=[(v, "높게", 남길비율), (q, "높게", 1.0)], 합산="단계별")
        전략들[f"{q}→{v}"] = dict(지표=[(q, "높게", 남길비율), (v, "높게", 1.0)], 합산="단계별")
기간 = ("1990-04", "2026-09")
if 시험_기간: 기간 = ("2009-04", "2015-03")  # (제출본에서 더함) 0절 시험_기간일 때만 6년. 아래 표들이 24개월 넘는 줄만 남기므로: 코스피TR·코스닥TR 지수가 2011-01부터 있고, '공통' 기간은 최소분산(직전 36개월)에 수익률이 생긴 달부터라 둘 다 24개월 넘게 남도록. 기본은 위 기간
설정 = {**포.기본설정, "이름": "확장연구", "시장": ["KS", "KQ"], "기간": 기간, "기준일": "첫거래일",
        "재무시점": ("발표일", 150), "제외": 포.우동호2023["제외"], "가격": 읽기.TR, "고르기": ("상위N", 30),
        "비중": ["동일", "시총", "시총상한", "위험균형", "최소분산"], "비중상한": 0.1, "전량교체": False,
        "거래비용": dict(수수료=0.00015, 거래세=0.0025, 슬리피지=0.005), "무위험": "CD1m", "시장지수": "I.001", "전략들": 전략들}
주기들 = [1, 3, 6, 12]
비용배수 = {"비용0": 0, "비용기본": 1, "비용2배": 2}
print(f"전략 {len(전략들)}개 × 비중 {len(설정['비중'])} × 주기 {len(주기들)} × 비용 {len(비용배수)}")
pd.Series({k: str(v) for k, v in 설정.items() if k != "전략들"}).to_frame("설정")
```

### 3. 실행
주기마다 `포.실행`을 한 번 부른다. 기준일마다 거르기와 지표 계산은 한 번만 하고, 모든 전략이 같은 거른 종목에서 고른다. 비용 0배·2배는 같은 보유 비중으로 수익률만 다시 계산한다(`포.보유`).

```python
%%절 7.1
포.최소분산기록.clear()  # (제출본에서 더함) 다른 절이 남긴 기록을 지움: 아래 '최소분산 SLSQP' 횟수가 이 절 것만 되게(계산에는 안 씀)
결과, 수익 = {}, {}
for k in 주기들:
    t0 = time.time()
    o = 결과[k] = 포.실행({**설정, "리밸런싱": k})
    for 비용, x in 비용배수.items():
        c = {n: v * x for n, v in 설정["거래비용"].items()}
        for key, W in o["보유"].items():
            if key[0] != "유니버스":
                수익[(key[0], key[1], k, 비용)] = o["수익률"][key] if x == 1 else 포.보유(o["종목수익률"], W, c, 설정["전량교체"])
    print(f"{k}개월: 리밸런싱 {len(o['기록'])}회, {time.time() - t0:.0f}초")
이상 = {k: 포.이상값보유(o) for k, o in 결과.items()}  # (전략, 비중, 그룹) → +300% 종목-월 보유 건수(README 한계 '이상값 보유 칸')
수렴 = pd.DataFrame(포.최소분산기록)
print("최소분산 SLSQP:", len(수렴), "번 풂, 수렴 안 됨", int((~수렴["수렴"]).sum()), "번", 수렴.loc[~수렴["수렴"], "메시지"].value_counts().to_dict())
수익 = pd.DataFrame(수익).rename_axis(columns=["전략", "비중", "주기", "비용"])
grid = 포.기준일(설정)
벤치 = pd.DataFrame({"유니버스(시총가중)": 포.보유(결과[1]["종목수익률"], 결과[1]["보유"][("유니버스", "시총", "전체")], dict(수수료=0, 거래세=0, 슬리피지=0))})
지수 = {"코스피": 읽기.지수()["I.001"], "코스피TR": 읽기.지수()["I.030"], "코스닥": 읽기.지수()["I.201"], "코스닥TR": 읽기.지수()["I.273"], "KRX TMI": 읽기.KRX지수()["I.578"]}
for n, x in 지수.items():
    벤치[n] = x.reindex(grid).pct_change(fill_method=None)
F = 평가.요인(grid, 설정)
F주 = F.assign(시장=벤치["유니버스(시총가중)"])  # 정보비율은 유니버스 벤치마크 대비. 알파의 MKT는 코스피 − RF 그대로
수익.shape
```

### 4. 결과표: 전략 × 비중 × 주기 × 비용 × 기간
기간은 `평가.구간들`(전체, 1990-04~2004-03, 2004-04~2017-03, 2017-04~)과 '공통'(모든 전략·비중·주기에 수익률이 생긴 첫 보유 월부터 끝까지)이다. 최소분산·위험균형은 과거 수익률이 쌓인 뒤에 시작하므로, 비중 방식끼리 비교는 '공통' 기간으로 본다(감사 M3). 24개월보다 짧은 칸은 뺀다. 정보비율은 유니버스 시가총액 가중 벤치마크 대비다.
벤치마크 줄의 비중 칸은 '-', 주기 칸은 0이다.
칸: 개월·누적·연환산(기하)·표준편차(연, 월 표준편차 × √12)·벤치 대비 연환산(같은 달들의 유니버스 벤치마크 연환산과 차이)·샤프·MDD·정보비율(유니버스 대비)·CAPM(젠센)·FF3·Carhart α(월)와 각 t(OLS)·Carhart 적재값(β: MKT·SMB·HML·MOM)·adj R²·회전율(리밸런싱당 평균 한쪽, 전 기간). 타겟 페이퍼 지표: 우동호 표 2~7(p.514–516) = 누적·MDD·표준편차·샤프·정보비율·젠센·3요인·4요인 α(t), Novy-Marx(2013) 표 = 월 초과수익·α·적재값·adj R²(`공통/평가.py` 맨 위). α의 시장 요인(MKT)은 코스피 가격지수(I.001) − RF라, 배당 포함 가격(TR) 포트폴리오의 α에는 배당만큼이 섞인다(유니버스 시총가중 벤치마크도 Carhart α가 양수로 나옴).

```python
%%절 7.1
칸 = ["개월", "누적", "연환산", "표준편차(연)", "샤프", "MDD", "정보비율", "CAPM α(월)", "CAPM t(OLS)", "FF3 α(월)", "FF3 t(OLS)",
      "Carhart α(월)", "Carhart t(OLS)", "Carhart β:MKT", "Carhart β:SMB", "Carhart β:HML", "Carhart β:MOM", "Carhart adj R²"]  # 우동호 표 2~7(α와 t)·Novy-Marx 표(적재값·adj R²), 추가 지시 6
공통 = str((수익.apply(lambda r: r.first_valid_index()).max() - pd.Timedelta(days=15)).to_period("M"))  # 평가.구간과 같은 보유 월 계산
기간들 = {**평가.구간들, "공통": (공통, None)}
print("공통 기간 시작(보유 월):", 공통)
처음뺌 = lambda s: s.iloc[s.gt(0).to_numpy().argmax() + 1:].mean()  # 처음 종목을 산 리밸런싱(=1)과 그 전 빈 비중(=0)을 뺀 평균(감사 L4)
회전 = pd.concat({k: o["회전율"].drop(columns=[("유니버스", "시총", "전체")]).apply(처음뺌) for k, o in 결과.items()}, names=["주기"])  # 리밸런싱당 평균 한쪽 회전율
u = 벤치["유니버스(시총가중)"]
rows = {}
for key, r in list(수익.items()) + [(("벤치:" + n, "-", 0, "비용0"), r) for n, r in 벤치.items()]:
    for g, (a, b) in 기간들.items():
        x = 평가.구간(r.dropna(), a, b)
        if len(x) >= 24:
            o = {k: v for k, v in 평가.성과(x, F주).items() if k in 칸}
            o["보유 첫 달"], o["보유 끝 달"] = (f"{d - pd.Timedelta(days=15):%Y-%m}" for d in x.index[[0, -1]])  # 실제 기간(벤치 줄은 자료 시작이 늦어 기간 이름보다 짧을 수 있음, 한계 11번)
            o["벤치 대비 연환산"] = o["연환산"] - ((1 + u.reindex(x.index)).prod() ** (12 / len(x)) - 1)  # 같은 달들의 유니버스 벤치마크 연환산과 차이
            o["회전율(전 기간)"] = 회전.get((key[2], key[0], key[1], "상위30"), float("nan"))  # 리밸런싱당 평균 한쪽 회전율(기간 나누지 않음)
            o["이상값 보유(전 기간)"] = 이상[key[2]].get((key[0], key[1], "상위30"), float("nan")) if key[2] in 이상 else float("nan")  # 기간 나누지 않음
            rows[(*key, g)] = o
표 = pd.DataFrame(rows).T[["보유 첫 달", "보유 끝 달"] + 칸[:4] + ["벤치 대비 연환산"] + 칸[4:] + ["회전율(전 기간)", "이상값 보유(전 기간)"]].rename_axis(["전략", "비중", "주기", "비용", "기간"])
표.to_csv(Path.cwd() / "결과" / "확장연구_20261103.csv", encoding="utf-8-sig")
pd.to_pickle(dict(수익=수익, 벤치=벤치, F=F, F주=F주, 표=표, 회전=회전, 기록={k: o["기록"] for k, o in 결과.items()},
                  설정={k: v for k, v in 설정.items() if k != "전략들"}, 전략들=전략들),
             Path.cwd() / "결과" / "확장연구_20261103.pkl")  # 결과요약 노트북이 읽음(결과_월수익률/에 두면 평가.요약이 읽으려다 형식이 달라 멈춤)
print("저장:", len(표), "줄")
표.loc[[i for i in 표.index if i[0].startswith("벤치:") and i[4] == "전체"]]
```

### 6. 리포트용 표·그림 (`결과_전략수익률/`)
기준 = 마법공식(EY+ROC 순위합산), 동일가중, 12개월, 기본 비용. 비교마다 **한 가지 조건만** 바꾼 성과표(csv)와 누적 그림(png)을 저장한다. 벤치마크 줄(유니버스 시총가중·코스피)도 같은 기간으로 함께 둔다.
| 파일 | 바꾼 것 | 기간 |
|---|---|---|
| `1_비중` | 동일·시총·시총상한·위험균형·최소분산 | 공통(모든 비중이 수익률을 내는 첫 달부터) |
| `2_주기` | 1·3·6·12개월 | 전체 |
| `3_지표` | 가치 지표 4 × 질 지표 4 (순위합산) — 그림은 기준에서 지표 하나만 바꾼 7개 | 전체 |
| `4_순서` | 순위합산 / 가치 먼저 → 질 / 질 먼저 → 가치 | 전체 |
| `5_비용` | 비용 0·기본·2배 × 주기 1·12개월 | 전체 |
| `7_무위험_민감도` | 무위험 CD1m(본문) / 콜금리 E11.02.003.001(예전 기본): 기준 전략·유니버스·코스피의 샤프·CAPM·Carhart α·t(OLS) | 전체 |
| `6_기간` | 1990-04~2004-03 / 2004-04~2017-03 / 2017-04~ | 기간별(그림 없음: 기간마다 시작점이 달라 한 그림에 겹치지 않음. 전체 기간 누적은 `2_주기` 그림의 12개월 선) |

```python
%%절 7.1
리포트 = Path.cwd() / "결과" / "결과_전략수익률"
리포트.mkdir(exist_ok=True)
기준 = ("EY+ROC 순위합산", "동일", 12, "비용기본")
벤치줄 = lambda g: [("벤치:유니버스(시총가중)", "-", 0, "비용0", g), ("벤치:코스피", "-", 0, "비용0", g)]


def 비교(이름, 줄, 기간="전체", 그림열=None, 제목=""):
    """표 줄(전략, 비중, 주기, 비용) + 벤치마크 → csv, 그림열(수익 열) + 벤치마크 → 누적 png. 그 표를 돌려준다."""
    t = 표.reindex([(*k, 기간) for k in 줄] + 벤치줄(기간)).dropna(how="all")
    t.to_csv(리포트 / f"{이름}.csv", encoding="utf-8-sig")
    if 그림열:
        a = 공통 if 기간 == "공통" else None
        R = pd.concat([수익[k].rename(" ".join(map(str, k))) for k in 그림열] + [벤치[n] for n in ["유니버스(시총가중)", "코스피"]], axis=1)
        R = 평가.구간(R.loc[R[[" ".join(map(str, k)) for k in 그림열]].notna().any(axis=1).idxmax():], a, None)  # 첫 전략 수익률이 생긴 달부터
        평가.누적그림(R, 제목).savefig(리포트 / f"{이름}.png", dpi=150, bbox_inches="tight")
    return t


s0, b0, k0, c0 = 기준
비중들, 질들 = 설정["비중"], list(질)
t1 = 비교("1_비중", [(s0, b, k0, c0) for b in 비중들], "공통", [(s0, b, k0, c0) for b in 비중들], "비중 방식 비교: 마법공식, 12개월, 기본 비용('공통' 기간)")
t2 = 비교("2_주기", [(s0, b0, k, c0) for k in 주기들], "전체", [(s0, b0, k, c0) for k in 주기들], "리밸런싱 주기 비교: 마법공식, 동일가중, 기본 비용")
지표줄 = [(f"{v}+{q} 순위합산", b0, k0, c0) for v in 가치 for q in 질 if f"{v}+{q} 순위합산" in 전략들]
한개만 = [k for k in 지표줄 if k[0].startswith("EY+") or k[0].endswith("+ROC 순위합산")]
t3 = 비교("3_지표", 지표줄, "전체", 한개만, "지표 바꾸기(기준 EY+ROC에서 하나만): 동일가중, 12개월, 기본 비용")
순서줄 = [(n, b0, k0, c0) for n in ["EY+ROC 순위합산", "EY→ROC", "ROC→EY"]]
t4 = 비교("4_순서", 순서줄, "전체", 순서줄, "요인 순서 비교: EY·ROC, 동일가중, 12개월, 기본 비용")
비용줄 = [(s0, b0, k, c) for k in [1, 12] for c in 비용배수]
t5 = 비교("5_비용", 비용줄, "전체", 비용줄, "거래비용 민감도: 마법공식, 동일가중, 주기 1·12개월")
t6 = pd.concat({g: 표.reindex([(*기준, g)] + 벤치줄(g)).dropna(how="all") for g in list(평가.구간들)[1:]}, names=["기간칸"]).droplevel(-1)
t6.to_csv(리포트 / "6_기간.csv", encoding="utf-8-sig")
F콜 = 평가.요인(grid, {**설정, "무위험": "E11.02.003.001"}).assign(시장=벤치["유니버스(시총가중)"])  # 민감도: 예전 기본 콜금리(추가 지시 10)
t7 = pd.DataFrame({(n, 금리): {k: v for k, v in 평가.성과(r.dropna(), f).items() if k in ["개월", "샤프", "CAPM α(월)", "Carhart α(월)", "Carhart t(OLS)"]}
                   for n, r in {"기준 전략": 수익[기준], "벤치:유니버스(시총가중)": 벤치["유니버스(시총가중)"], "벤치:코스피": 벤치["코스피"]}.items()
                   for 금리, f in {"CD1m(본문)": F주, "콜 E11.02.003.001": F콜}.items()}).T.rename_axis(["줄", "무위험"])
t7.to_csv(리포트 / "7_무위험_민감도.csv", encoding="utf-8-sig")
display(t7)
print("저장:", sorted(f.name for f in 리포트.iterdir()))
pd.concat({"1_비중": t1, "2_주기": t2, "4_순서": t4, "5_비용": t5}, names=["비교"])[["개월", "연환산", "표준편차(연)", "샤프", "MDD", "회전율(전 기간)", "벤치 대비 연환산", "정보비율", "Carhart α(월)", "Carhart t(OLS)"]]
# (제출본에서 바꿈) 이 칸의 결과_전략수익률 폴더 → 결과/결과_전략수익률
```

> 원래 7절(리밸런싱마다 단계별 종목 수 표)은 옮기지 않았다 → 저장소 `연구/확장연구_20261103.ipynb` 7절 참고.

- 여러 조합 중 가장 좋은 것을 고르면 과최적화(데이터 마이닝) 위험이 있다. 조합 전체의 분포(중앙값)와 기간별 일관성으로 본다.
- 그 밖의 데이터 한계는 README '한계·미해결'.

## 7.2 이로운: 비중바꾸기
저장소 `연구/이로운_비중바꾸기.ipynb`의 칸(원래 1·2절 import 칸과 8절 기록 표만 뺌).

## 이로운: 비중 바꾸기 (마법공식 상위 30종목)
연구_틀(`연구_틀.ipynb`)을 복사해 **0번(실험 기록)과 3번(설정) 칸만** 바꾸고, 10번부터 실험 칸을 더했다. 공통 계산(`공통/`)은 이 노트북에서 바꾸지 않는다.

줄임말: EW(동일가중, Equal Weight), VW(시가총액가중, Value Weight), LS(롱숏: 좋은 분위를 사고 나쁜 분위를 판 수익률),
MDD(최대 낙폭, Maximum Drawdown), IR(정보비율, Information Ratio), CAPM(자본자산가격결정모형), FF3(파마-프렌치 3요인), Carhart(카하트 4요인: FF3 + 모멘텀),
OLS(보통최소제곱), NW는 회원 결정(2026-11-06)으로 뺌, t는 OLS, TR(배당 포함 수정주가).

### 0. 실험 기록 [바꾸는 칸]
누가, 언제, 무엇을 왜 바꾸는지 적는다. 저장 파일 이름(이름_날짜_실험명.csv)에 쓰인다.

```python
%%절 7.2
실험 = dict(
    이름="이로운",
    날짜="2026-11-02",
    실험명="비중바꾸기",
    무엇을="마법공식(ROC + EY) 상위 30종목을 동일·시총·시총상한 10%·30%·위험균형·최소분산으로 담아 비교",
    왜="같은 종목을 어떤 비중으로 담느냐에 따라 수익·위험·회전율·거래비용이 어떻게 달라지는지 보려고",
)
```

### 3. 설정 [바꾸는 칸]
`포.기본설정`에서 바꿀 것만 적는다. 쓸 수 있는 값:
- 시장: `"KS"`(유가증권), `"KQ"`(코스닥) / 기준일: `"첫거래일"`, `"마지막거래일"` / 리밸런싱: 1·3·12(달)
- 제외: `종목거르기.py`의 함수 이름 — 보통주, 스팩, 금융업, 투자기구, 결산월, 거래정지, 관리종목, 동전주, 자본잠식, 회전율, 거래대금, 시총상위, 시총분위, 종목빼기 (옵션은 `("이름", {옵션})`). `시장`과 `쓰지말것`(README_데이터 '쓰지 말 것' 구간)은 항상 먼저 적용된다
- 지표와 방향: `지표.py`의 함수 이름 + `"높게"`/`"낮게"` (단계별이면 셋째 값 = 남길 비율)
- 합산: `"순위합산"`, `"z합산"`, `"단계별"` / 고르기: `("상위N", n)`, `("분위", k)`
- 재무시점: `("결산후개월", k)`(결산일 + k개월, 발표일 안 봄) | `("발표일", k)`(발표일 다음 거래일, 없으면 결산일 + k일. k = 150 본문, 180 확인)
- 비중: `"동일"`, `"시총"`, `"시총상한"`(비중상한), `"시총상한10"`·`"시총상한30"`, `"위험균형"`(직전 12개월 변동성 역수), `"최소분산"`(공매도 없음, 직전 36개월 월수익률, Ledoit-Wolf 수축) / 거래비용: 수수료·거래세·슬리피지(·월비용) / 전량교체: `True`면 리밸런싱 때 전부 팔고 새로 삼(우동호 원문), `False`면 바뀐 만큼만

```python
%%절 7.2
설정 = {**포.기본설정,
    "이름": 실험["실험명"],
    "기간": ("1993-04", "2026-09"),  # 최소분산에 직전 36개월 월수익률이 필요. 배당 포함 수정주가가 1990-01부터 있어 1993-04부터
    "시장": ["KS", "KQ"],
    "기준일": "첫거래일",
    "리밸런싱": 12,
    "재무시점": ("발표일", 150),  # 1999년 이전은 발표일 기록이 거의 없어 결산일 + 150일(통합_확인기록 F8)
    "제외": ["보통주", "금융업", "투자기구", ("결산월", {"월": 12}), "거래정지"],
    "고르기": ("상위N", 30),
    "거래비용": dict(수수료=0.00015, 거래세=0.0025, 슬리피지=0.005),
    "지표": [("ROC", "높게"), ("EY", "높게")],
    "합산": "순위합산",
    "비중": ["동일", "시총", "시총상한10", "시총상한30", "위험균형", "최소분산"],  # 위험균형 = 직전 12개월 일간 변동성 역수, 최소분산 = 공매도 없음·36개월·Ledoit-Wolf
}
if 시험_기간: 설정["기간"] = ("1993-04", "1996-03")  # (제출본에서 더함) 0절 시험_기간일 때만 짧게. 기본은 위 기간
```

### 4. 새 지표 [바꾸는 칸]
이 실험은 새 지표를 만들지 않는다(`공통/지표.py`의 함수만 씀).

```python
%%절 7.2
# 새 지표 없음
```

### 5. 실행 [건드리지 않는 칸]

```python
%%절 7.2
포.최소분산기록.clear()  # (제출본에서 더함) 다른 절(7.1)이 남긴 기록을 지움: 11절 '최소분산 SLSQP' 횟수가 이 실험 것만 되게(계산에는 안 씀)
결과 = 포.실행(설정)
F = 평가.요인(포.기준일(결과["설정"]), 결과["설정"])
R = pd.DataFrame({f"{b}_{g}": 결과["수익률"][(b, g)] for b, g in 결과["수익률"]})
if 설정["고르기"][0] == "분위":  # LS는 거래비용 0으로 다시 돌려 만든다(비용 뺀 Q5를 빼면 숏 쪽 비용 부호가 거꾸로 됨)
    g = 포.실행({**설정, "거래비용": dict(수수료=0, 거래세=0, 슬리피지=0)})["수익률"]
    for b in 설정["비중"]:
        R[f"{b}_LS(비용0)"] = g[(b, "Q1")] - g[(b, f"Q{설정['고르기'][1]}")]
R["시장"] = F["시장"]
R = R.dropna(how="all")
```

### 6. 결과표·누적수익률 그림 [건드리지 않는 칸]

```python
%%절 7.2
표 = 평가.성과표(R, F, [c for c in R if "_LS" in c])
표.insert(0, "이상값 보유", pd.Series({f"{b}_{g}": n for (b, g), n in 포.이상값보유(결과).items()}))  # +300% 종목-월 보유 건수(README 한계 '이상값 보유 칸')
평가.누적그림(R, 설정["이름"]);
표
```

### 7. 같은 기간 기준과 자동 비교 [건드리지 않는 칸]
같은 기간으로 우동호(2023) 재현과 시장을 돌려 나란히 놓는다. 무위험이자율과 전량교체 여부는 3번 칸 설정 것을 쓴다.
기준은 두 줄이다. **본문 설정**은 우동호 그대로(배당 빠진 수정주가, 결산 후 3개월), **같은 가격·재무시점**은 가격 계열과 재무시점까지 3번 칸 설정에 맞춘 것. 두 기준의 차이는 배당 포함 여부와 재무시점 차이에서 나온다. 실험 효과는 '같은 가격·재무시점' 줄과 비교한다.
기간 안에서 개월 수가 다를 수 있다(개월 칸 확인. 첫 리밸런싱에 종목이 없으면 그 구간은 빈칸).

```python
%%절 7.2
기준 = {**포.우동호2023, "기간": 설정["기간"], "무위험": 설정["무위험"], "전량교체": 설정["전량교체"]}
기준들 = {f"기준:{기준['이름']}(본문 설정)": 기준, f"기준:{기준['이름']}(같은 가격·재무시점)": {**기준, "가격": 설정["가격"], "재무시점": 설정["재무시점"]}}
Fb = 평가.요인(포.기준일(기준), 기준)
기준표 = 평가.성과표(pd.DataFrame({k: 포.실행(v)["수익률"][("동일", "상위30")] for k, v in 기준들.items()} | {"기준:시장": Fb["시장"]}).dropna(how="all"), Fb)
비교 = pd.concat([표, 기준표])
비교[["개월", "누적", "연환산", "표준편차(연)", "샤프", "MDD", "정보비율", "Carhart α(월)", "Carhart t(OLS)"]]
```

> 원래 8절(리밸런싱마다 단계별 종목 수 표)은 옮기지 않았다 → 저장소 `연구/이로운_비중바꾸기.ipynb` 8절 참고.

### 9. 저장 [건드리지 않는 칸]
설정과 결과를 `연구/결과/이름_날짜_실험명.csv`에 남긴다. 같은 이름이 있으면 뒤에 번호를 붙인다(덮어쓰지 않음).

### 10. 회전율·거래비용 전후 [실험 칸]
- **회전율** = 리밸런싱 때 산 비중의 합(처음 매수 = 1이라 첫 리밸런싱은 평균에서 뺀다). 리밸런싱 사이 값이 변해 생긴 비중 차이도 들어간다.
- **비용 전** = 같은 설정에서 거래비용만 0. 둘의 연환산 차이가 비용으로 잃은 연수익이다.
- **편입 수** = 비중이 0.1%보다 큰 종목 수. 위험균형(12개월)·최소분산(36개월)은 이력이 모자란 종목이 빠지고, 최소분산은 비중 0인 종목이 생긴다.

```python
%%절 7.2
결과0 = 포.실행({**설정, "거래비용": dict(수수료=0, 거래세=0, 슬리피지=0)})
연 = lambda r: (1 + r.dropna()).prod() ** (12 / r.dropna().size) - 1
비용표 = pd.DataFrame({b: {"평균 회전율(리밸런싱당)": 결과["회전율"][(b, "상위30")].iloc[1:].mean(),
                         "편입 수 평균": pd.Series([(w > 0.001).sum() for w in 결과["보유"][(b, "상위30")].values()]).mean(),
                         "편입 수 최소": min((w > 0.001).sum() for w in 결과["보유"][(b, "상위30")].values()),
                         "최대 비중 평균": pd.Series([w.max() for w in 결과["보유"][(b, "상위30")].values()]).mean(),
                         "연환산(비용 전)": 연(결과0["수익률"][(b, "상위30")]), "연환산(비용 후)": 연(결과["수익률"][(b, "상위30")])} for b in 설정["비중"]}).T
비용표["비용으로 잃은 연수익"] = 비용표["연환산(비용 전)"] - 비용표["연환산(비용 후)"]
display(비용표)
평가.성과표(pd.DataFrame({f"{b}(비용 전)": 결과0["수익률"][(b, "상위30")] for b in 설정["비중"]}).dropna(how="all"), F)[["누적", "연환산", "표준편차(연)", "샤프", "MDD"]]
```

### 11. 요약표용 저장 [실험 칸]

```python
%%절 7.2
S = 결과["수익률"].copy()
S.columns = pd.MultiIndex.from_tuples([(f"비중바꾸기 MF {g}", b) for b, g in S])
평가.저장(S, F, "연구_이로운_비중바꾸기", 이상값={(f"비중바꾸기 MF {g}", b): n for (b, g), n in 포.이상값보유(결과).items()})
수렴 = pd.DataFrame(포.최소분산기록)
print("최소분산 SLSQP:", len(수렴), "번 풂, 수렴 안 됨", int((~수렴["수렴"]).sum()) if len(수렴) else 0, "번")
list(S)
```

### 3. 설정 [바꾸는 칸]
`포.기본설정`에서 바꿀 것만 적는다. 쓸 수 있는 값:
- 시장: `"KS"`(유가증권), `"KQ"`(코스닥) / 기준일: `"첫거래일"`, `"마지막거래일"` / 리밸런싱: 1·3·12(달)

- 지표와 방향: `지표.py`의 함수 이름 + `"높게"`/`"낮게"` (단계별이면 셋째 값 = 남길 비율)
- 합산: `"순위합산"`, `"z합산"`, `"단계별"` / 고르기: `("상위N", n)`, `("분위", k)`
- 재무시점: `("결산후개월", k)`(결산일 + k개월, 발표일 안 봄) | `("발표일", k)`(발표일 다음 거래일, 없으면 결산일 + k일. k = 150 본문, 180 확인)
- 비중: `"동일"`, `"시총"`, `"시총상한"`(비중상한), `"시총상한10"`·`"시총상한30"`, `"위험균형"`(직전 12개월 변동성 역수), `"최소분산"`(공매도 없음, 직전 36개월 월수익률, Ledoit-Wolf 수축) / 거래비용: 수수료·거래세·슬리피지(·월비용) / 전량교체: `True`면 리밸런싱 때 전부 팔고 새로 삼(우동호 원문), `False`면 바뀐 만큼만

```python
%%절 7.3
설정 = {**포.기본설정,
    "이름": 실험["실험명"],
    "기간": ("2004-04", "2026-09"),
    "시장": ["KS", "KQ"],
    "기준일": "첫거래일",
    "리밸런싱": 12,
    "재무시점": ("발표일", 150),  # 1999년 이전은 발표일 기록이 거의 없어 결산일 + 150일(통합_확인기록 F8)
    "제외": ["보통주", "금융업", "투자기구", ("결산월", {"월": 12}), "거래정지"],
    "고르기": ("상위N", 30),
    "거래비용": dict(수수료=0.00015, 거래세=0.0025, 슬리피지=0.005),
    "지표": [("EY", "높게"), ("ROC", "높게")],  # 기준(그린블라트 원래 공식). 조합은 아래 표
    "합산": "순위합산",
    "비중": ["동일", "시총"],
}
if 시험_기간: 설정["기간"] = ("2014-04", "2019-03")  # (제출본에서 더함) 0절 시험_기간일 때만 짧게(고르는 기간 ~2017-03 35개월, 검증 기간 2017-04~ 24개월이 들어가게). 기본은 위 기간
# 돌릴 조합(가치 함수, 퀄리티 함수, 경제적 이유). 함수는 공통/지표.py: EP=E/P, BP=B/M, SP=S/P, CFP=CF/P, EBITDA_EV=EBITDA/EV, ROA, GPA=GP/A, 영업CF자산=영업현금흐름/자산
조합 = pd.DataFrame([
    ("EY", "ROC", "기준: 그린블라트 원래 공식. 둘 다 기업 전체(부채 포함) 기준 — 이익/기업가치, 이익/투하자본"),
    ("EY", "GPA", "우동호 IMF. 매출총이익은 비용 분류·일회성 항목의 영향이 적은 수익성(Novy-Marx 2013)"),
    ("EY", "영업CF자산", "이익이 현금으로 뒷받침되는 회사 고르기(발생액이 큰 회사 피하기, Sloan 1996)"),
    ("EBITDA_EV", "ROC", "둘 다 기업 전체 기준. 가치 쪽을 감가상각 정책 차이를 덜 타는 EBITDA로"),
    ("EBITDA_EV", "GPA", "둘 다 감가상각·이자 전 이익 → 자본구조·감가상각 정책 영향 줄임"),
    ("EP", "ROE", "둘 다 주주 기준(순이익). P/B = P/E × ROE라 낮은 P/E와 높은 ROE를 함께 보면 '수익성 대비 싼 장부가'"),
    ("EP", "ROA", "순이익 기준끼리. 빚으로 부풀린 ROE 대신 자산 대비 순이익"),
    ("BP", "ROE", "Fama-French 5요인의 HML(B/M)·RMW(수익성) 결합. 분모가 같은 장부가"),
    ("BP", "GPA", "Novy-Marx(2013) 가치 + 수익성(04 신마법공식). 두 요인의 상관이 음수라 분산 효과"),
    ("SP", "영업이익률", "P/S가 낮아도 마진이 낮으면 싼 이유가 있음 → 마진으로 거름(P/E = P/S ÷ 순이익률)"),
    ("CFP", "영업CF자산", "둘 다 영업현금흐름 기준 → 회계 발생액의 영향 줄임"),
], columns=["가치", "퀄리티", "경제적 이유"])
고르는기간, 검증기간 = ("2004-04", "2017-03"), ("2017-04", None)  # 보유 월 기준
# 고르는 기준(돌리기 전에 정함): 고르는 기간 동일가중 샤프가 가장 높은 조합 1개 → 검증 기간 성과를 따로 본다
```

### 5. 실행 [건드리지 않는 칸]

```python
%%절 7.3
결과 = 포.실행(설정)
F = 평가.요인(포.기준일(결과["설정"]), 결과["설정"])
R = pd.DataFrame({f"{b}_{g}": 결과["수익률"][(b, g)] for b, g in 결과["수익률"]})
if 설정["고르기"][0] == "분위":  # LS는 거래비용 0으로 다시 돌려 만든다(비용 뺀 Q5를 빼면 숏 쪽 비용 부호가 거꾸로 됨)
    g = 포.실행({**설정, "거래비용": dict(수수료=0, 거래세=0, 슬리피지=0)})["수익률"]
    for b in 설정["비중"]:
        R[f"{b}_LS(비용0)"] = g[(b, "Q1")] - g[(b, f"Q{설정['고르기'][1]}")]
R["시장"] = F["시장"]
R = R.dropna(how="all")
```

### 6. 결과표·누적수익률 그림 [건드리지 않는 칸]

```python
%%절 7.3
표 = 평가.성과표(R, F, [c for c in R if "_LS" in c])
표.insert(0, "이상값 보유", pd.Series({f"{b}_{g}": n for (b, g), n in 포.이상값보유(결과).items()}))  # +300% 종목-월 보유 건수(README 한계 '이상값 보유 칸')
평가.누적그림(R, 설정["이름"]);
표
```

### 7. 같은 기간 기준과 자동 비교 [건드리지 않는 칸]
같은 기간으로 우동호(2023) 재현과 시장을 돌려 나란히 놓는다. 무위험이자율과 전량교체 여부는 3번 칸 설정 것을 쓴다.
기준은 두 줄이다. **본문 설정**은 우동호 그대로(배당 빠진 수정주가, 결산 후 3개월), **같은 가격·재무시점**은 가격 계열과 재무시점까지 3번 칸 설정에 맞춘 것. 두 기준의 차이는 배당 포함 여부와 재무시점 차이에서 나온다. 실험 효과는 '같은 가격·재무시점' 줄과 비교한다.
기간 안에서 개월 수가 다를 수 있다(개월 칸 확인. 첫 리밸런싱에 종목이 없으면 그 구간은 빈칸).

### 10. 조합 실행과 고르는 기간·검증 기간 성과 [실험 칸]
조합마다 같은 설정으로 돌린다(지표만 바꿈). 성과는 보유 월로 두 기간을 나눠 따로 계산한다: 고르는 기간 2004-04~2017-03, 검증 기간 2017-04~.
돌린 조합은 아래 표의 전부다(빼고 보고한 조합 없음).

```python
%%절 7.3
이름들 = 조합["가치"] + "+" + 조합["퀄리티"]
display(조합.set_index(이름들.rename("조합")))
조합결과 = {n: 결과 if (v, q) == ("EY", "ROC") else 포.실행({**설정, "이름": n, "지표": [(v, "높게"), (q, "높게")]})
           for n, v, q in zip(이름들, 조합["가치"], 조합["퀄리티"])}
열 = ["개월", "누적", "연환산", "표준편차(연)", "샤프", "MDD", "Carhart α(월)", "Carhart t(OLS)"]
표들 = {k: 평가.성과표(pd.DataFrame({(n, w): 평가.구간(o["수익률"][(w, "상위30")].dropna(), a, b) for n, o in 조합결과.items() for w in 설정["비중"]}), F)[열]
       for k, (a, b) in {"고르는 기간": 고르는기간, "검증 기간": 검증기간}.items()}
pd.concat(표들, axis=1)
```

### 11. 고르기와 검증 [실험 칸]
미리 정한 기준(고르는 기간 동일가중 샤프 1등)으로 하나를 고르고, 검증 기간에 기준(EY+ROC)과 비교한다.
조합을 여러 개 시험했으므로 고르는 기간 1등의 성과는 운이 섞여 부풀어 있다. 검증 기간 숫자가 덜 치우친 값이다.

```python
%%절 7.3
샤프 = {k: t.xs("동일", level=1)["샤프"].astype(float) for k, t in 표들.items()}
고른 = 샤프["고르는 기간"].idxmax()
print(f"고르는 기간 동일가중 샤프 1등: {고른} ({샤프['고르는 기간'][고른]:.4f}) / 기준 EY+ROC: {샤프['고르는 기간']['EY+ROC']:.4f}")
print(f"검증 기간 동일가중 샤프: {고른} {샤프['검증 기간'][고른]:.4f} (순위 {int(샤프['검증 기간'].rank(ascending=False)[고른])}/{len(샤프['검증 기간'])}) / 기준 EY+ROC {샤프['검증 기간']['EY+ROC']:.4f}")
print(f"두 기간 동일가중 샤프 순위의 스피어만 상관: {샤프['고르는 기간'].corr(샤프['검증 기간'], method='spearman'):.4f}")
표들["검증 기간"].loc[[(고른, "동일"), ("EY+ROC", "동일"), (고른, "시총"), ("EY+ROC", "시총")]]
```

## 7.4 이로운: 팩터순서
저장소 `연구/이로운_팩터순서.ipynb`의 칸(원래 1·2절 import 칸과 8절 기록 표만 뺌).

## 이로운: 팩터 순서 (순위 합산 vs 단계별 거르기)
연구_틀(`연구_틀.ipynb`)을 복사해 **0번(실험 기록)과 3번(설정) 칸만** 바꾸고, 10번부터 실험 칸을 더했다. 공통 계산(`공통/`)은 이 노트북에서 바꾸지 않는다.

줄임말: EW(동일가중, Equal Weight), VW(시가총액가중, Value Weight), LS(롱숏: 좋은 분위를 사고 나쁜 분위를 판 수익률),
MDD(최대 낙폭, Maximum Drawdown), IR(정보비율, Information Ratio), CAPM(자본자산가격결정모형), FF3(파마-프렌치 3요인), Carhart(카하트 4요인: FF3 + 모멘텀),
OLS(보통최소제곱), NW는 회원 결정(2026-11-06)으로 뺌, t는 OLS, TR(배당 포함 수정주가).

### 0. 실험 기록 [바꾸는 칸]
누가, 언제, 무엇을 왜 바꾸는지 적는다. 저장 파일 이름(이름_날짜_실험명.csv)에 쓰인다.

```python
%%절 7.4
실험 = dict(
    이름="이로운",
    날짜="2026-11-02",
    실험명="팩터순서",
    무엇을="ROC·EY 순위 합산(마법공식)과, 한 지표로 먼저 상위 50%·30%를 거른 뒤 다른 지표로 30종목을 고르는 단계별 방식을 비교",
    왜="두 지표를 어떤 순서로 쓰느냐(합산 vs 거르기, 무엇을 먼저)가 결과를 바꾸는지 보려고. 종목 수는 모두 30으로 같게",
)
```

### 3. 설정 [바꾸는 칸]
`포.기본설정`에서 바꿀 것만 적는다. 쓸 수 있는 값:
- 시장: `"KS"`(유가증권), `"KQ"`(코스닥) / 기준일: `"첫거래일"`, `"마지막거래일"` / 리밸런싱: 1·3·12(달)
- 제외: `종목거르기.py`의 함수 이름 — 보통주, 스팩, 금융업, 투자기구, 결산월, 거래정지, 관리종목, 동전주, 자본잠식, 회전율, 거래대금, 시총상위, 시총분위, 종목빼기 (옵션은 `("이름", {옵션})`). `시장`과 `쓰지말것`(README_데이터 '쓰지 말 것' 구간)은 항상 먼저 적용된다
- 지표와 방향: `지표.py`의 함수 이름 + `"높게"`/`"낮게"` (단계별이면 셋째 값 = 남길 비율)
- 합산: `"순위합산"`, `"z합산"`, `"단계별"` / 고르기: `("상위N", n)`, `("분위", k)`
- 재무시점: `("결산후개월", k)`(결산일 + k개월, 발표일 안 봄) | `("발표일", k)`(발표일 다음 거래일, 없으면 결산일 + k일. k = 150 본문, 180 확인)
- 비중: `"동일"`, `"시총"`, `"시총상한"`(비중상한), `"시총상한10"`·`"시총상한30"`, `"위험균형"`(직전 12개월 변동성 역수), `"최소분산"`(공매도 없음, 직전 36개월 월수익률, Ledoit-Wolf 수축) / 거래비용: 수수료·거래세·슬리피지(·월비용) / 전량교체: `True`면 리밸런싱 때 전부 팔고 새로 삼(우동호 원문), `False`면 바뀐 만큼만

```python
%%절 7.4
설정 = {**포.기본설정,
    "이름": 실험["실험명"],
    "기간": ("1990-04", "2026-09"),  # 백테스트 시작 1990(1988~89 배당 포함 수정주가 없음), 4월 리밸런싱
    "시장": ["KS", "KQ"],
    "기준일": "첫거래일",
    "리밸런싱": 12,
    "재무시점": ("발표일", 150),  # 1999년 이전은 발표일 기록이 거의 없어 결산일 + 150일(통합_확인기록 F8)
    "제외": ["보통주", "금융업", "투자기구", ("결산월", {"월": 12}), "거래정지"],
    "고르기": ("상위N", 30),
    "거래비용": dict(수수료=0.00015, 거래세=0.0025, 슬리피지=0.005),
    "지표": [("ROC", "높게"), ("EY", "높게")],  # 기준: 순위 합산
    "합산": "순위합산",
    "비중": ["동일", "시총"],
}
if 시험_기간: 설정["기간"] = ("1990-04", "1993-03")  # (제출본에서 더함) 0절 시험_기간일 때만 짧게. 기본은 위 기간
# 비교할 순서(지표, 합산). 단계별: (지표, 방향, 남길 비율) — 앞 지표로 상위 비율만 남기고 마지막 지표로 줄 세워 상위 30
순서 = {"순위합산(기준)": dict(지표=[("ROC", "높게"), ("EY", "높게")], 합산="순위합산"),
       "ROC 상위50% → EY": dict(지표=[("ROC", "높게", 0.5), ("EY", "높게", 1.0)], 합산="단계별"),
       "ROC 상위30% → EY": dict(지표=[("ROC", "높게", 0.3), ("EY", "높게", 1.0)], 합산="단계별"),
       "EY 상위50% → ROC": dict(지표=[("EY", "높게", 0.5), ("ROC", "높게", 1.0)], 합산="단계별"),
       "EY 상위30% → ROC": dict(지표=[("EY", "높게", 0.3), ("ROC", "높게", 1.0)], 합산="단계별")}
```

### 4. 새 지표 [바꾸는 칸]
이 실험은 새 지표를 만들지 않는다(`공통/지표.py`의 함수만 씀).

```python
%%절 7.4
# 새 지표 없음
```

### 5. 실행 [건드리지 않는 칸]

```python
%%절 7.4
결과 = 포.실행(설정)
F = 평가.요인(포.기준일(결과["설정"]), 결과["설정"])
R = pd.DataFrame({f"{b}_{g}": 결과["수익률"][(b, g)] for b, g in 결과["수익률"]})
if 설정["고르기"][0] == "분위":  # LS는 거래비용 0으로 다시 돌려 만든다(비용 뺀 Q5를 빼면 숏 쪽 비용 부호가 거꾸로 됨)
    g = 포.실행({**설정, "거래비용": dict(수수료=0, 거래세=0, 슬리피지=0)})["수익률"]
    for b in 설정["비중"]:
        R[f"{b}_LS(비용0)"] = g[(b, "Q1")] - g[(b, f"Q{설정['고르기'][1]}")]
R["시장"] = F["시장"]
R = R.dropna(how="all")
```

### 6. 결과표·누적수익률 그림 [건드리지 않는 칸]

```python
%%절 7.4
표 = 평가.성과표(R, F, [c for c in R if "_LS" in c])
표.insert(0, "이상값 보유", pd.Series({f"{b}_{g}": n for (b, g), n in 포.이상값보유(결과).items()}))  # +300% 종목-월 보유 건수(README 한계 '이상값 보유 칸')
평가.누적그림(R, 설정["이름"]);
표
```

### 7. 같은 기간 기준과 자동 비교 [건드리지 않는 칸]
같은 기간으로 우동호(2023) 재현과 시장을 돌려 나란히 놓는다. 무위험이자율과 전량교체 여부는 3번 칸 설정 것을 쓴다.
기준은 두 줄이다. **본문 설정**은 우동호 그대로(배당 빠진 수정주가, 결산 후 3개월), **같은 가격·재무시점**은 가격 계열과 재무시점까지 3번 칸 설정에 맞춘 것. 두 기준의 차이는 배당 포함 여부와 재무시점 차이에서 나온다. 실험 효과는 '같은 가격·재무시점' 줄과 비교한다.
기간 안에서 개월 수가 다를 수 있다(개월 칸 확인. 첫 리밸런싱에 종목이 없으면 그 구간은 빈칸).

```python
%%절 7.4
기준 = {**포.우동호2023, "기간": 설정["기간"], "무위험": 설정["무위험"], "전량교체": 설정["전량교체"]}
기준들 = {f"기준:{기준['이름']}(본문 설정)": 기준, f"기준:{기준['이름']}(같은 가격·재무시점)": {**기준, "가격": 설정["가격"], "재무시점": 설정["재무시점"]}}
Fb = 평가.요인(포.기준일(기준), 기준)
기준표 = 평가.성과표(pd.DataFrame({k: 포.실행(v)["수익률"][("동일", "상위30")] for k, v in 기준들.items()} | {"기준:시장": Fb["시장"]}).dropna(how="all"), Fb)
비교 = pd.concat([표, 기준표])
비교[["개월", "누적", "연환산", "표준편차(연)", "샤프", "MDD", "정보비율", "Carhart α(월)", "Carhart t(OLS)"]]
```

### 9. 저장 [건드리지 않는 칸]
설정과 결과를 `연구/결과/이름_날짜_실험명.csv`에 남긴다. 같은 이름이 있으면 뒤에 번호를 붙인다(덮어쓰지 않음).

```python
%%절 7.4
out = Path.cwd() / "결과"
f = out / f"{실험['이름']}_{실험['날짜']}_{실험['실험명']}.csv"
i = 1
while f.exists():
    f, i = out / f"{실험['이름']}_{실험['날짜']}_{실험['실험명']}_{i}.csv", i + 1
비교.assign(실험기록=json.dumps(실험, ensure_ascii=False), 설정=json.dumps(설정, ensure_ascii=False, default=str)).to_csv(f, encoding="utf-8-sig")
print("저장:", f)
```

### 10. 순서별 실행 [실험 칸]
첫 표: 리밸런싱마다 편입 종목 수의 최솟값·최댓값(모두 30이어야 같은 조건). 둘째 표: 순위합산 포트폴리오와 겹치는 종목 수. 셋째 표: 성과.

```python
%%절 7.4
순서결과 = {k: 결과 if k == "순위합산(기준)" else 포.실행({**설정, "이름": k, **v}) for k, v in 순서.items()}
display(pd.DataFrame({k: o["기록"]["편입:상위30"].agg(["min", "max"]) for k, o in 순서결과.items()}).T.rename_axis("편입 종목 수"))
기준W = 결과["보유"][("동일", "상위30")]
display(pd.DataFrame({k: pd.Series({t: len(w.index.intersection(기준W[t].index)) for t, w in o["보유"][("동일", "상위30")].items()}) for k, o in 순서결과.items()})
        .agg(["mean", "min", "max"]).T.rename_axis("순위합산과 겹치는 종목 수"))
X = pd.DataFrame({(k, w): o["수익률"][(w, "상위30")] for k, o in 순서결과.items() for w in 설정["비중"]}).dropna(how="all")
평가.누적그림(X.xs("동일", axis=1, level=1), "팩터 순서: 동일가중");
평가.성과표(X, F)
```

```python
%%절 7.4
S = X.copy()
S.columns = pd.MultiIndex.from_tuples([(f"팩터순서 {k}", w) for k, w in S])
평가.저장(S, F, "연구_이로운_팩터순서", 이상값={(f"팩터순서 {k}", w): 포.이상값보유(o)[(w, "상위30")] for k, o in 순서결과.items() for w in 설정["비중"]})
list(S)
```

### 1. 설정: 본문과 확인판
- **본문**: 새 따라하기와 같음(`포.우동호2023_원문`: 보통주만, 수정주가, 결산일 + 3개월, 직전년도 결산만, 금융업 원문 7업종, 투자기구 거르기 없음).
- **확인판**(원문에 없음): 재무시점만 `("발표일", 150)`로 바꿈. 발표일(FP56000900)이 기록된 보고서는 그 다음 거래일부터, 기록이 없으면 결산일 + 150일부터(150일은 근거 없음, 02 노트북 발표지연 표는 늦은 발표 비율만 보여 줌). 재무연도는 본문처럼 직전년도만이라, 그때까지 t−1 결산이 안 나온 종목은 그해 빠진다(예전 확인판은 FY t−2를 썼음).

```python
%%절 7.5.1
IMF = {"이름": "우동호2023_IMF", "지표": [("GPA", "높게"), ("EY", "높게")]}
설정 = {"MF": {**포.우동호2023_원문, "기간": 기간, "끝": "마지막거래일"}, "IMF": {**포.우동호2023_원문, **IMF, "기간": 기간, "끝": "마지막거래일"}}
설정 |= {f"{k}_확인판": {**v, "이름": v["이름"] + "_확인판", "재무시점": ("발표일", 150)} for k, v in list(설정.items())}
for k, v in 설정.items():  # 새 따라하기와 같은 기준(원문 p.512 '직전년도'·'수정주가', 보통주만 회원 결정 2026-11-06)
    assert v["재무연도"] == "직전년도" and v["가격"] == 읽기.수정주가 and "보통주" in v["제외"], k
    assert ("금융업", {"이름": "원문"}) in v["제외"] and "투자기구" not in v["제외"], k
pd.DataFrame(설정).astype(str)
```

### 2. 상위 30종목 전체 성과표: 본문·확인판 × EW·VW (원문은 본문 EW만)
월수익률 성과표 전체 칸(누적·연환산·표준편차(연)·샤프·MDD·정보비율(월 × √12), 초과평균·CAPM·FF3·Carhart α와 t(OLS), Carhart β·adj R²)과 '이상값 보유'(보유한 달 가운데 종목 수익률이 한 달 +300%를 넘은 종목-월 수, 7절). 요인은 새 따라하기처럼 DataGuide 요인(`평가.요인`, 과제 공지).
둘째 표는 같은 보유 비중을 일간 수익률에 넣어 원문 단위(`평가.일간정의`: 일 × √N)로 잰 SD·Sharpe·IR·MDD. KOSPI 행의 알파·t는 뜻이 없다.

```python
%%절 7.5.1
결과 = {k: 포.실행({**v, "비중": ["동일", "시총"]}) for k, v in 설정.items()}
g = 포.기준일(설정["MF"])
F = 평가.요인(g, 설정["MF"])  # SMB·HML·MOM = DataGuide 요인(과제 공지, 새 따라하기와 같음)
R = pd.DataFrame({f"{k}_{b}": o["수익률"][(b, "상위30")] for k, o in 결과.items() for b in ["동일", "시총"]})
R["KOSPI"] = F["시장"]
R = R.dropna(how="all")
표 = 평가.성과표(R, F)
표.insert(0, "이상값 보유", pd.Series({f"{k}_{b}": 포.이상값보유(o)[(b, "상위30")] for k, o in 결과.items() for b in ["동일", "시총"]}))
display(표)

# 일간(새 따라하기 4절과 같은 방법)
가격 = 설정["MF"]["가격"]
날 = 읽기.상장가격(가격).loc[g[0]:g[-1]].index
Rd = 포.쓰지말것칸(포.기간수익률(읽기.상장가격(가격), 날), 가격)
국고 = 읽기.국고3년일간()
무위험일 = 국고.reindex(날.union(국고.index)).ffill().reindex(날).shift(1) / 100 / 365
시장일 = 읽기.지수()[설정["MF"]["시장지수"]].reindex(날).pct_change(fill_method=None).iloc[1:]
일 = {f"{k}_{b}": 포.보유(Rd, o["보유"][(b, "상위30")], 설정[k]["거래비용"], 설정[k]["전량교체"]) for k, o in 결과.items() for b in ["동일", "시총"]} | {"KOSPI": 시장일}
일표 = pd.DataFrame({k: 평가.일간정의(r, 시장일, 무위험일) for k, r in 일.items()}).T
일표
```

원문 Fig. 1(선형 %, MF·IMF·KOSPI 셋)은 따라하기에 있다. 이 그림은 로그 축에 확인판·VW 선을 더한 것.

### 4. 본문 vs 확인판: 종목 수와 결과
첫 표는 리밸런싱마다 순위를 매긴 종목 수(거르기 뒤 − 지표 빈칸). 둘째 표는 첫 리밸런싱일에 종목마다 쓴 결산자료의 회계연도(행 = 본문, 열 = 확인판, 0 = 확인판에 행 없음). 셋째 표는 확인판이 쓴 행이 발표일 기록인지 결산일 + 150일인지. 마지막 표는 성과(누적·연환산·알파는 월, SD·Sharpe·MDD는 일간 정의).

```python
%%절 7.5.1
display(pd.DataFrame({k: (o["기록"]["거래정지"] - o["기록"]["빈칸으로 빠짐"]).set_axis(o["기록"].index.year) for k, o in 결과.items()}).rename_axis("순위 매긴 종목 수"))
t0 = g[0]
a, b = 포.재무표(t0, 설정["MF"]), 포.재무표(t0, 설정["MF_확인판"])
display(pd.crosstab(a["회계연도"], b["회계연도"].reindex(a.index).fillna(0).astype(int), rownames=["본문 회계연도"], colnames=["확인판 회계연도(0=없음)"]))
display(pd.crosstab(b["회계연도"], b["발표일"].notna().map({True: "발표일 기록 있음", False: "기록 없음 → 결산일+150일"}), rownames=["확인판 회계연도"], colnames=["확인판이 쓴 행"]))
_, c0, _ = 포.거르기(t0, 설정["MF"])
print(f"{t0:%Y-%m-%d} 거르기 뒤 {len(c0)}종목 중 본문이 쓴 결산자료의 실제 발표일이 그날 이후인 종목(원문 규칙의 미래 정보): {int((a['발표일'].reindex(c0) >= t0).sum())}")
k4 = ["MF_동일", "MF_확인판_동일", "IMF_동일", "IMF_확인판_동일"]
pd.concat([표.loc[k4, ["누적", "연환산", "CAPM α(월)", "Carhart α(월)", "Carhart t(OLS)"]], 일표.loc[k4, ["SD", "Sharpe", "MDD"]].add_suffix("(일간)")], axis=1)
```

### 5. 수업 형식: 5분위·LS(Q1 − Q5), EW·VW, 비용 0
같은 순위로 종목을 5묶음으로 나눈다. Q1이 총 순위가 가장 좋은 묶음. LS = Q1 − Q5. 분위·LS는 수업 형식대로 거래비용 0(비용을 뺀 Q5를 빼면 숏 쪽 비용 부호가 거꾸로 됨). 본문 설정만. Carhart α·t(OLS)는 DataGuide 요인으로 낸다(과제 공지 ④: 각 분위와 롱숏의 Carhart 4요인 회귀). '이상값 영향' = 그 묶음이 +300% 초과 종목-월(7절)을 보유했는지.

```python
%%절 7.5.1
비용0 = dict(수수료=0, 거래세=0, 슬리피지=0)
분위 = {k: 포.실행({**설정[k], "고르기": ("분위", 5), "비중": ["동일", "시총"], "거래비용": 비용0}) for k in ["MF", "IMF"]}
Q = {}
for k, o in 분위.items():
    for bw in ["동일", "시총"]:
        r = o["수익률"][bw]
        Q.update({f"{k}_{bw}_{q}": r[q] for q in r})
        Q[f"{k}_{bw}_LS"] = r["Q1"] - r["Q5"]
Q = pd.DataFrame(Q)
분위표 = 평가.성과표(Q, F, [c for c in Q if c.endswith("LS")])
영향 = {f"{k}_{bw}_{q}": n for k, o in 분위.items() for (bw, q), n in 포.이상값보유(o).items()}
영향 |= {f"{k}_{bw}_LS": 영향[f"{k}_{bw}_Q1"] + 영향[f"{k}_{bw}_Q5"] for k in 분위 for bw in ["동일", "시총"]}
분위표.insert(0, "이상값 보유", pd.Series(영향))
분위표.insert(1, "이상값 영향", 분위표["이상값 보유"].gt(0).map({True: "이상값 영향", False: ""}))
분위표
```

### 6. 첫 리밸런싱일 거르기 단계별 남은 수·재료별 빈칸·이자비용 빈칸(연도별)
원문 표 1 Number(18칸) 대조는 따라하기 10절에 남는다. 여기는 원문에 없는 단계별 수다.
'재료 전부 있음' = ROC·EY에 드는 세전이익·이자비용·유동자산·유동부채·자산총계·무형자산·현금·부채총계·시가총액이 모두 있는 종목. 우리 설정에는 원문에 없는 `보통주` 단계와 `쓰지말것` 단계도 있다. 마지막 표는 연도별 투자기구(선박투자회사·부동산투자·리츠) 이름에 걸리는 종목 수(새 기준은 투자기구를 거르지 않음, 회원 05:57: 거른다면 빠질 수)와 이자비용만 빈칸이라 빠진 수(빈칸은 채우지 않음).

```python
%%절 7.5.1
s, c, n = 포.거르기(t0, 설정["MF"])
재료 = {"세전이익": 지표.세전이익, "이자비용": 지표.이자비용, "유동자산": 지표.유동자산, "유동부채": 지표.유동부채, "자산총계": 지표.자산,
       "무형자산": 지표.무형자산, "현금": 지표.현금, "부채총계": 지표.부채}
x = s["fin"].loc[c, list(재료.values())].set_axis(list(재료), axis=1).assign(시가총액=s["mcap"][c])
n["재료 전부 있음"] = int(x.notna().all(axis=1).sum())
display(pd.Series(n, name=f"{t0:%Y-%m-%d} 남은 종목 수").rename({"가격있음": "전체(그날 가격 있음)"}).to_frame())
display(x.isna().sum().rename("거래정지 단계 뒤 재료별 빈칸 수").to_frame())
이자 = 읽기.연간재무()[지표.이자비용]
print(f"연간재무 전체 이자비용: 값 있는 칸 {int(이자.notna().sum())}, 그중 0인 칸 {int((이자 == 0).sum())}")
연도별 = {}
for t in 결과["MF"]["기록"].index:  # 리밸런싱일만
    st, ct, nt = 포.거르기(t, 설정["MF"])
    xt = st["fin"].reindex(ct)[list(재료.values())].set_axis(list(재료), axis=1).assign(시가총액=st["mcap"].reindex(ct))
    빈 = xt["이자비용"].isna()
    나머지 = xt.drop(columns="이자비용").notna().all(axis=1)
    연도별[t.year] = {"투자기구 이름에 걸림(안 거름)": len(ct) - len(포.종목거르기.투자기구(st, ct)), "거래정지 단계 뒤": len(ct), "이자비용 빈칸": int(빈.sum()),
                   "이자비용만 빈칸(나머지 재료 다 있음)": int((빈 & 나머지).sum()), "재료 전부 있음": int((~빈 & 나머지).sum())}
pd.DataFrame(연도별).T.rename_axis("리밸런싱 연도")
```

### 7. 이상값: 한 달 수익률 +300% 초과 종목-월 (원본 값 공개)
원본 수정주가(상장 여부와 상관없이 값 있는 날 전부)로 기준일 간 종목 수익률을 계산해 +300%를 넘는 종목-월을 모두 보인다. 값은 고치지 않는다.
- **그날 시장구분 / 상장 구간 수익률**: 실제 계산에 쓰는 값(시장구분이 코스피·코스닥인 날의 가격만). 빈칸이면 장외·상장 전 가격이라 계산에서 빠짐.
- **쓰지 말 것**: README_데이터의 '쓰지 말 것' 구간에 걸리면 근거(D-05 등).
- **MF·IMF 보유**: 2절 본문 EW 상위 30이 그 종목-월을 보유했는지.
- A008080·A070480은 재조회(R28)와 통합 파일의 차이가 0인 DataGuide 원래 값(회원 확인 2026-11-02).
- 끝에 코드 재사용 종목의 경계 수익률이 끊기는지 세고, A070480(신비앤텍) 보유 횟수가 0인지 확인한다(0이 아니면 멈춤).

```python
%%절 7.5.1
원시R = 포.기간수익률(읽기.일간(설정["MF"]["가격"]), g)
큰 = 포.큰수익률(원시R).to_frame("원본 가격 수익률")
상장R, 앞, 시장 = 결과["MF"]["종목수익률"], dict(zip(g[1:], g[:-1])), 읽기.일간(읽기.시장구분)
bb = 읽기.쓰지말것(설정["MF"]["가격"])
큰["이름"] = 읽기.종목정보()["코드명"].reindex(큰.index.get_level_values(1)).to_numpy()
큰["그날 시장구분"] = [시장.at[d, c] if c in 시장 else None for d, c in 큰.index]
큰["상장 구간 수익률"] = [상장R.at[d, c] for d, c in 큰.index]
큰["쓰지 말 것"] = [", ".join(bb.loc[(bb["코드"] == c) & (bb["시작"] <= d) & (bb["끝"] >= 앞[d]), "근거"]) for d, c in 큰.index]
for k in ["MF", "IMF"]:
    W = 결과[k]["보유"][("동일", "상위30")]
    d0 = pd.DatetimeIndex(sorted(W))
    큰[f"{k} 보유"] = [bool((d0 < d).any()) and c in W[d0[d0 < d][-1]].index for d, c in 큰.index]
print(f"원본 가격 +300% 초과 {len(큰)}건 / 상장 구간 수익률도 +300% 초과 {int((큰['상장 구간 수익률'] > 3).sum())}건 / "
      f"'쓰지 말 것'에 걸림 {int(큰['쓰지 말 것'].ne('').sum())}건 / MF 보유 {int(큰['MF 보유'].sum())}건 / IMF 보유 {int(큰['IMF 보유'].sum())}건")
display(큰)
gg = 포.기준일({**설정["MF"], "기간": ("1990-01", "2026-09"), "끝": None})
RR = 포.기간수익률(읽기.상장가격(설정["MF"]["가격"]), gg)
u = pd.read_csv(읽기.새목록 / "코드재사용.csv", encoding="utf-8-sig", parse_dates=["새회사_첫날짜"])
u = u[u["새회사_첫날짜"] < gg[-1]]
끊김 = [bool(pd.isna(RR.at[gg[gg > x][0], c])) if c in RR else True for c, x in zip(u["코드"], u["새회사_첫날짜"])]
print(f"코드 재사용 {len(u)}줄({u['코드'].nunique()}종목) 중 새 회사 첫 기준일 수익률이 빈칸(경계에서 끊김): {sum(끊김)}")
A070480 = {f"{k}_{bw}": sum("A070480" in w.index for w in o["보유"][(bw, "상위30")].values()) for k, o in 결과.items() for bw in ["동일", "시총"]}
print("A070480(신비앤텍)을 보유한 리밸런싱 횟수:", A070480)
assert not any(A070480.values()), "A070480을 보유함 → 이 종목을 뺀 확인판을 다시 넣을 것"
```

### 10. 원문 차이 진단 A1: 거르기 단계를 하나씩 끈 판·유가증권만 판의 순위 매긴 종목 수 대 원문 표 1 Number
(예전 `_검증/01_진단.py` A를 새 기준 설정으로 옮김. 표 1 평균수익률·표준편차 대조(A2)는 원문 대조라 옮기지 않음.) 새 기준은 투자기구를 거르지 않으므로(회원 05:57) '투자기구 안 거름' 판 대신 '투자기구 거름' 판(금융업 다음에 투자기구 거르기를 더함)을 둔다.

```python
%%절 7.5.1
기본, IMF설정 = 설정["MF"], 설정["IMF"]
표1 = pd.read_csv(읽기.새목록.parent / "저장소_검증" / "01_원문표1.csv", index_col="연도")
판들 = {"기본": 기본["제외"], "보통주 안 거름": [f for f in 기본["제외"] if f != "보통주"], "투자기구 거름": [x for f in 기본["제외"] for x in ([f, "투자기구"] if isinstance(f, tuple) and f[0] == "금융업" else [f])],
        "거래정지 안 거름": [f for f in 기본["제외"] if f != "거래정지"], "결산월 안 거름": [f for f in 기본["제외"] if not (isinstance(f, tuple) and f[0] == "결산월")]}
rows = []
for t in g[:-1][::12]:
    for 이름, ex in 판들.items():
        for 시장들 in (["KS", "KQ"], ["KS"]) if 이름 == "기본" else (["KS", "KQ"],):
            s, c, _ = 포.거르기(t, {**기본, "제외": ex, "시장": 시장들})
            v = pd.DataFrame({m: getattr(지표, m)(s).reindex(c) for m in ["ROC", "EY"]}).replace([np.inf, -np.inf], np.nan).dropna()
            rows.append({"연도": t.year, "판": 이름 + ("" if 시장들 == ["KS", "KQ"] else " (유가증권만)"), "순위 매김": len(v)})
A = pd.DataFrame(rows).pivot(index="연도", columns="판", values="순위 매김")
A.insert(0, "원문 Number", 표1["Number"].reindex(A.index))
A.to_csv(저장 / "01_진단_표본수.csv", encoding="utf-8-sig")
print("원문 Number − 우리 순위 매김(기본): 평균", round((A["원문 Number"] - A["기본"]).mean(), 1))
A
# (제출본에서 바꿈) 이 칸의 저장소 _검증 자료 파일 → data/저장소_검증/
```

### 11. 원문 차이 진단 B1~B3: 시총 1분위 모집단 3가지, 1분위 MF 연도별·종목 기여
분위 모집단: 남은 종목 / 지표 자료 있는 종목(따라하기 본문) / 상장 전 종목 경계. B2·B3은 본문 모집단의 1분위 MF.

```python
%%절 7.5.1
cum = lambda r: float((1 + r.dropna()).prod() - 1)
원문Q1 = {"MF": 44.2427, "IMF": 101.3052}  # 원문 표 3(p.514)
brow, 결과B = {}, {}
for 이름, cfg in {"MF": 기본, "IMF": IMF설정}.items():
    for 판, ex in {"남은 종목으로 나눔": [("시총분위", {"분위": 1})],
                  "지표 자료 있는 종목으로 나눔(따라하기 본문)": [("지표있음", {"지표들": [m[0] for m in cfg["지표"]]}), ("시총분위", {"분위": 1})],
                  "상장 전 종목 경계로 나눔": [("시총분위", {"분위": 1, "기준": "상장전체"})]}.items():
        o = 포.실행({**cfg, "제외": cfg["제외"] + ex})
        결과B[(이름, 판)] = o
        brow[(이름, 판)] = {"누적": cum(o["수익률"][("동일", "상위30")]), "원문 누적": 원문Q1[이름], "종목 수(첫 해 편입)": int(o["기록"]["편입:상위30"].iloc[0])}
B = pd.DataFrame(brow).T
B.to_csv(저장 / "01_진단_1분위모집단.csv", encoding="utf-8-sig")
display(B)
o = 결과B[("MF", "지표 자료 있는 종목으로 나눔(따라하기 본문)")]
W, Rs = o["보유"][("동일", "상위30")], o["종목수익률"]
dates = sorted(W)
yr_rows, contrib = [], []
for a0, a1 in zip(dates, dates[1:] + [Rs.index[-1]]):
    r = Rs.loc[(Rs.index > a0) & (Rs.index <= a1), W[a0].index]
    gr = (1 + r.fillna(0)).prod() - 1
    contrib.append((gr * W[a0]).rename(a0.year))
    yr_rows.append({"연도": a0.year, "포트 1년(비용 전, 동일가중 매수보유)": float((W[a0] * gr).sum()), "+100% 넘은 종목": int((gr > 1).sum()), "가장 큰 종목 수익률": float(gr.max())})
Y = pd.DataFrame(yr_rows).set_index("연도")
Y.to_csv(저장 / "01_진단_1분위연도.csv", encoding="utf-8-sig")
display(Y)
Cb = pd.concat(contrib, axis=1).stack().rename("기여").reset_index()
Cb.columns = ["코드", "연도", "기여"]
Cb["이름"] = 읽기.종목정보()["코드명"].reindex(Cb["코드"]).to_numpy()
Cb.nlargest(15, "기여")
```

### 12. 원문 차이 진단 C: 설정 판별 전체 기간 누적(배)
하루 비보유(3월 말 → 4월 첫 거래일 하루 현금), 거래세 매수에도, TR(배당 포함) 가격, 발표일 확인판, 비용 0, 세전이익 잠정치(FP56000300 파일이 있을 때, 잠정치 있는 종목만: 회원 결정 2026-11-04 15:13 확인판으로만). 원문 p.512 문장으로 본문을 바꾸지 않은 이유(예전 따라하기 15절 설명 그대로): '분석대상의 기간은 매년 4월 첫 거래일부터 다음연도 3월 마지막 거래일까지로 구분'은 표 1의 해마다 기간을 나누는 문장이고, 같은 쪽 '4월 첫 거래일에 기존 포트폴리오를 전량 매도하고, 새로운 재무데이터를 활용하여 투자 포트폴리오를 매입'은 같은 날 팔고 사는 것이라 3월 말~4월 첫 거래일 하루도 보유한다. 거래비용 문장('증권거래수수료 0.015%, 증권거래세 0.25%, 매수매도 거래 시마다 추가 거래비용(Slippage) 0.5%')은 '매수매도 거래 시마다'가 슬리피지에만 걸리는지 거래세에도 걸리는지 문장만으로 알 수 없다. 두 판 모두 진단으로만 둔다.

```python
%%절 7.5.1
P = 읽기.상장가격(기본["가격"])
def 하루비보유(cfg):
    """4월 첫 거래일 앞 마지막 거래일을 기준일에 더하고 그 하루 구간 수익률을 0(현금)으로."""
    o = 포.실행(cfg)
    gc = 포.기준일(cfg)
    rb = pd.DatetimeIndex(sorted(o["보유"][("동일", "상위30")]))
    앞날 = pd.DatetimeIndex([P.index[P.index.get_loc(t) - 1] for t in rb[1:]])
    Ra = 포.쓰지말것칸(포.기간수익률(읽기.상장가격(cfg["가격"]), gc.union(앞날)), cfg["가격"])
    Ra.loc[rb[1:]] = 0.0
    return 포.보유(Ra, o["보유"][("동일", "상위30")], cfg["거래비용"], cfg["전량교체"])
상위 = lambda c: 포.실행(c)["수익률"][("동일", "상위30")]
판C = {"기본(따라하기 본문)": 상위,
       "3월 말→4월 첫 거래일 하루 비보유": 하루비보유,
       "거래세 매수에도": lambda c: 상위({**c, "거래비용": {**c["거래비용"], "수수료": c["거래비용"]["수수료"] + c["거래비용"]["거래세"], "거래세": 0}}),
       "TR(배당 포함) 가격": lambda c: 상위({**c, "가격": 읽기.TR}),
       "재무 = 발표일 다음 날(확인판)": lambda c: 상위({**c, "재무시점": ("발표일", 150)}),
       "거래비용 0": lambda c: 상위({**c, "거래비용": dict(수수료=0, 거래세=0, 슬리피지=0)})}
if 읽기.있음("FP56000300"):
    잠정 = {"EY": "EY_잠정", "ROC": "ROC_잠정"}
    판C["세전이익 = 잠정치(FP56000300, 있는 종목만)"] = lambda c: 상위({**c, "지표": [(잠정.get(m, m), *r) for m, *r in c["지표"]]})
Cc = pd.Series({(이름, 판): cum(f(cfg)) for 이름, cfg in {"MF": 기본, "IMF": IMF설정}.items() for 판, f in 판C.items()}).unstack(0)
Cc.loc["원문 누적(표 2)"] = pd.Series({"MF": 9.8035, "IMF": 10.2791})
Cc.to_csv(저장 / "01_진단_설정.csv", encoding="utf-8-sig")
Cc
```

### 13. 원인별 요약
숫자는 위 9~12절 출력에서 가져온다. 원문 숫자: 표 1(p.512), 표 2·3(p.514).

```python
%%절 7.5.1
판 = [c for c in A.columns if c != "원문 Number"]
가까운판 = (A[판].sub(A["원문 Number"], axis=0)).abs().mean().idxmin()
f = lambda v: f"{v:.4f}"
근거C = {"3월 말→4월 첫 거래일 하루 비보유": "p.512 '4월 첫 거래일부터 다음연도 3월 마지막 거래일까지로 구분'", "거래세 매수에도": "p.512 '매수매도 거래 시마다'가 어디까지 걸리는지 문장만으로 모름",
       "세전이익 = 잠정치(FP56000300, 있는 종목만)": "p.512 '연간 사업보고서가 공시되는 3월 말까지 데이터를 수집': 원문은 확정치 쪽, 잠정치는 확인판"}
안거름 = [c for c in 판 if "안 거름" in c]
요약 = pd.DataFrame([
    ["표준편차·샤프 정의", "원문에 정의 없음. KOSPI 원문 숫자(표 2 p.514)가 일간 √N 정의에서 나옴(역산)",
     f"MF SD {f(정의표.loc['MF', ('우리(일간 정의)', 'SD')])}·샤프 {f(정의표.loc['MF', ('우리(일간 정의)', 'Sharpe')])}, IMF SD {f(정의표.loc['IMF', ('우리(일간 정의)', 'SD')])}·샤프 {f(정의표.loc['IMF', ('우리(일간 정의)', 'Sharpe')])}",
     "MF 0.9479·2.42, IMF 0.8857·2.58", "KOSPI 일간 정의 값은 9절 표"],
    ["표본 수(표 1 Number)", "p.512 표 1, 3.1절 거르기 조건",
     f"첫 해 순위 매긴 수: 본문 {A['기본'].iloc[0]}, 유가증권만 {A['기본 (유가증권만)'].iloc[0]}, 단계 하나씩 끔 {A[안거름].iloc[0].min()}~{A[안거름].iloc[0].max()}. 해마다 차이 절대값 평균이 가장 작은 판: {가까운판}",
     "2004년 850 (2021년 1,955)", f"본문 해마다 차이 평균 {(A['원문 Number'] - A['기본']).mean():.1f}. 원문이 더 적은 해 {', '.join(str(y) for y in A.index[A['원문 Number'] < A['기본']])}, 원문이 더 많은 해 {', '.join(str(y) for y in A.index[A['원문 Number'] > A['기본']])}"],
    ["시총 1분위 모집단", "p.514 표 3: 시총 5분위로 나눈 뒤 분위마다 상위 30. 모집단은 원문에 없음",
     ", ".join(f"{k[1]} {v:.1f}배" for k, v in B["누적"].items() if k[0] == "MF"),
     "MF 44.2427배", "1분위 MF 해마다 +100% 넘은 종목 수 최대: " + ", ".join(f"{t}년 {int(n)}개" for t, n in Y["+100% 넘은 종목"].nlargest(2).items())],
] + [[f"설정: {k}", 근거C.get(k, "원문 p.512(가격 = 수정주가, 재무 = 3월 말까지 공시, 비용)"),
      f"MF {Cc.loc[k, 'MF']:.4f}, IMF {Cc.loc[k, 'IMF']:.4f}", "MF 9.8035, IMF 10.2791",
      f"MF {Cc.loc[k, 'MF'] - 9.8035:+.4f}, IMF {Cc.loc[k, 'IMF'] - 10.2791:+.4f}"] for k in Cc.index if k != "원문 누적(표 2)"],
    columns=["원인", "근거", "바꾼 뒤 숫자", "원문", "남은 차이"])
요약.to_csv(저장 / "01_진단_요약.csv", encoding="utf-8-sig")
요약
```

### 15. 월 누적 대 일간 누적 (따라하기 6절에서 옮김, 원문에 없음)
따라하기 표 2~7의 Return(월수익률 누적)과 SD·Sharpe·IR·MDD에 쓴 일간 수익률 누적이 같은지 보인다. 같은 보유 비중이라도 구간 안에 값이 빈 날이 있는 종목(상장폐지·거래정지·'쓰지 말 것' 칸)은 월간은 그 달 전체를, 일간은 그날부터 빼고 남은 종목에 다시 넣으므로 경로가 달라 조금 다를 수 있다(예전 판 출력 `_검증/01_정의역산_결과.txt`: 일간을 월로 묶은 것과 월수익률 차이 최대 MF 1.96e-03, IMF 3.41e-04). 차이는 고치지 않고 표로만 보인다. 시총 5분위는 따라하기 6절과 같은 방법(그 전략 지표가 있는 종목을 4월 첫 거래일 시총으로 5분위, 분위마다 상위 30)으로 다시 돌린다.

```python
%%절 7.5.1
# 따라하기 6절과 같은 시총 5분위 실행(분위 모집단 = 그 전략 지표 자료가 있는 종목)
규모결과 = {f"{k}_시총{q}분위": 포.실행({**설정[k], "제외": 설정[k]["제외"] + [("지표있음", {"지표들": [m[0] for m in 설정[k]["지표"]]}), ("시총분위", {"분위": q})]}) for q in range(1, 6) for k in ["MF", "IMF"]}
규모 = pd.DataFrame({n: o["수익률"][("동일", "상위30")] for n, o in 규모결과.items()})
규모일 = {n: 포.보유(Rd, o["보유"][("동일", "상위30")], o["설정"]["거래비용"], o["설정"]["전량교체"]) for n, o in 규모결과.items()}
# 월 누적(Return 칸)과 같은 보유의 일간 누적이 같은지(차이가 크면 위 설명대로 경로 차이, 값은 고치지 않음). 전체 = 2절 본문 EW
누적차 = pd.DataFrame({n: {"월 누적": float((1 + r.dropna()).prod() - 1), "일간 누적": float((1 + d.dropna()).prod() - 1)}
                    for n, r, d in [(k, R[c], 일[c]) for k, c in [("MF", "MF_동일"), ("IMF", "IMF_동일"), ("KOSPI", "KOSPI")]] + [(k, 규모[k], 규모일[k]) for k in 규모]}).T
누적차["차이(월 − 일)"] = 누적차["월 누적"] - 누적차["일간 누적"]
누적차
```

### 1. 무위험·시장지수 (#20·#21)
본 실험(그림 16~19)은 원문에 무위험이자율·시장지수가 없다. 아래 성과표의 샤프·정보비율·α를 내려고 우리가 고른 것: 무위험 = 콜금리(1일물 중개거래, E11.02.003.001), 시장지수 = 코스피 I.001.
현실판 벤치마크 KOSPI 200 동일가중(I.401)은 원문(p.26)에 있어 따라하기에 남지만, 그 요인표(F200)는 여기 현실판 성과표에도 쓴다.

```python
%%절 7.5.2
F = 평가.요인(포.기준일({**기본, "기간": (기간[0], max(기간[1], 현실기간[1]))}), 기본)  # 시험 때 현실판(2009~) 칸에도 I.001이 있게; 본 실행은 기간 그대로
F200 = 평가.요인(포.기준일({**기본, "기간": 현실기간}), {**기본, "시장지수": "I.401"})
pd.Series({"무위험": 기본["무위험"], "시장지수": 기본["시장지수"], "F 개월": len(F), "F200 개월": len(F200)}).to_frame("값")
```

### 5. 현실판 덧붙인 판·성과표 (#57·#59·#62·#63)
원문(p.26)은 비용 월 0.2% 판과 KOSPI 200 동일가중(I.401)만 말한다. 비용 없는 판, 2002-03부터 판('확장'), 코스피(I.001) 참고 줄, 성과표는 우리 추가.
- 2009-01~ 표: 본(비용 월 0.2%)·비용 없음·I.401·I.001. 정보비율·CAPM α는 I.401 기준(F200).
- '확장' 판은 따로 자기 기간 전체로 잰다(HEAD 판은 `loc["2009-02":]`로 잘려 본 줄과 같은 숫자였음 = 감사표 결함 7). I.401은 2009-01-02부터라 확장 판의 기준은 I.001(F).
- 확장 판: 5개 지표가 다 있는 종목이 시총 300위·거래대금 거르기 뒤 5보다 적은 달은 묶음이 비어 빈칸 → 실제 첫 수익률 달이 늦다(아래 출력).

```python
%%절 7.5.2
현실설정 = {**기본, "이름": "NH밸류_현실판", "기간": 현실기간, "지표": [(m, "높게") for m in 지표명.values()], "합산": "z합산",
           "제외": 기본["제외"] + [("시총상위", {"n": 300}), ("거래대금", {"창": "월"})]}
월비용 = dict(수수료=0, 거래세=0, 슬리피지=0, 월비용=0.002)  # 원문 p.26 '매월 전체 포트폴리오의 0.2% 매매비용'
현실 = {"현실판_비용월0.2%(본)": 포.실행({**현실설정, "거래비용": 월비용}),
        "현실판_비용없음": 포.실행(현실설정),
        "확장: 현실판_2002-03부터_비용월0.2%": 포.실행({**현실설정, "기간": 기간, "거래비용": 월비용})}
q = lambda o: o["수익률"][("동일", "Q1")]
현실표 = pd.DataFrame({k: q(현실[k]) for k in ("현실판_비용월0.2%(본)", "현실판_비용없음")}
                    | {"KOSPI200 동일가중(I.401)": F200["시장"], "KOSPI(I.001, 참고)": F["시장"]}).reindex(F200.index[1:])
확장 = pd.DataFrame({"확장: 현실판_2002-03부터_비용월0.2%": q(현실["확장: 현실판_2002-03부터_비용월0.2%"]), "KOSPI(I.001)": F["시장"]})
print("'확장' 판 실제 첫 수익률 달:", f"{확장.iloc[:, 0].first_valid_index():%Y-%m}", "(기간 설정은", 기간[0], "부터)")
display(남김(평가.성과표(현실표, F200), "현실판_성과표_2009부터"))
남김(평가.성과표(확장, F), "현실판_확장_성과표")
```

### 6. 그림 19 대조의 '개월 수'·'첫 달'·'같은 달만' 열 (#66)
원문 그림 19는 지표별 숫자 하나뿐이다. 지표마다 1분위 시작 달이 다를 수 있어, 여섯 줄 모두 값 있는 달만 쓴 판을 덧붙였다(우리 추가). 본 열(Q1 월평균/월표준편차)은 따라하기에 남는다.

```python
%%절 7.5.2
우리19 = (Q1.mean() / Q1.std())[그림19.index]
같은달 = Q1[그림19.index].dropna()
같은19 = (같은달.mean() / 같은달.std())[그림19.index]
display(남김(pd.DataFrame({"원문 그림19": 그림19, "우리(본 열)": 우리19, "개월 수": Q1[그림19.index].count(),
                          "첫 달": [Q1[n].first_valid_index().strftime("%Y-%m") for n in 그림19.index],
                          f"우리, 같은 달만({같은달.index[0]:%Y-%m}~{같은달.index[-1]:%Y-%m}, {len(같은달)}개월)": 같은19}), "그림19_같은달"))
누적LS = (1 + LS).prod() - 1
같은LS = (1 + LS.loc[같은달.index]).prod() - 1
남김(pd.DataFrame({"원문 주장": ["모든 지표에서 LS 누적이 양(+)", "P/B·P/E가 가장 효과적(LS 누적 상위 2)", "Multi-Metric 1분위가 단일 지표 1분위보다 위험조정수익률 높음"],
                  "우리, 같은 달만": [bool((같은LS > 0).all()), sorted(같은LS.nlargest(2).index) == ["P/B", "P/E"], bool(같은19.idxmax() == "Multi-Metric")]}), "주장_같은달")
```

### 7. 확인판: 연간 재무 150·180일 (#70), 무위험이자율 (#71)
- **연간 판**: 재무를 연간(결산) 값으로, 발표일 다음 거래일부터(발표일 없으면 결산일 + 150일, 확인판 180일). 150·180일은 근거 없음(2절). 본문(발표값)과 LS 누적 비교.
- **무위험**: 본 성과표는 콜금리(E11.02.003.001). 무담보콜(E11.02.003.002), 통안채 364일(E11.02.003.009, 2000년 이후). 포트폴리오는 그대로, 샤프·α만 바뀐다(1분위).

```python
%%절 7.5.2
연간설정 = {**기본, "재무주기": ("연간",), "재무시점": ("발표일", 150)}
연간 = {"Multi-Metric": 포.실행({**연간설정, "지표": [(m, "높게") for m in 지표명.values()], "합산": "z합산"})}
연간 |= {n: 포.실행({**연간설정, "지표": [(m, "높게")]}) for n, m in 지표명.items()}
늦게 = {n: 포.실행({**연간설정, "재무시점": ("발표일", 180), "지표": [(m, "높게")]}) for n, m in 지표명.items()}
lsof = lambda d: (1 + pd.DataFrame({n: o["수익률"]["동일"]["Q1"] - o["수익률"]["동일"][숏.get(n, "Q5")] for n, o in d.items() if n != "Multi-Metric"})).prod() - 1
display(남김(pd.DataFrame({"본문(발표값)": 누적LS, "연간 150일": lsof(연간), "연간 180일": lsof(늦게)}).rename_axis("LS 누적"), "연간150_180"))
금리 = {"콜(본문)": F, "무담보콜": 평가.요인(포.기준일(기본), {**기본, "무위험": "E11.02.003.002"}),
       "통안364(2000년 이후)": 평가.요인(포.기준일(기본), {**기본, "무위험": "E11.02.003.009"}).loc["2000":]}
남김(pd.concat({k: 평가.성과표(q1.loc[f.index[0]:], f)[["샤프", "CAPM α(월)", "Carhart α(월)"]] for k, f in 금리.items()}, axis=1), "무위험_확인판")
```

### 8. 차이 검정 (#73·#74)
원문은 순위·크기만 말하고 검정이 없다. 우리 추가:
- 샤프 차이: Multi-Metric 대 단일 지표 1분위의 (월평균/월표준편차) 차이, Jobson·Korkie(1981) 식을 Memmel(2003)이 고친 식(`평가.샤프차이`, 두 논문 원문 확인 안 함). p > 0.05면 '경계'.
- 평균 차이: LS 누적 상위 2와 3위의 월 LS 차이 평균, t(OLS)·p(`평가.평균차이`).

```python
%%절 7.5.2
qq = q1
경계 = pd.DataFrame({f"Multi-Metric 대 {n}": 평가.샤프차이(qq["Multi-Metric"], qq[n]) for n in 지표명}).T
경계["판정"] = np.where(경계["p"] > 0.05, "경계", np.where(경계["차이"] > 0, "Multi가 높음", "Multi가 낮음"))
display(남김(경계, "샤프차이").style.set_caption("Multi-Metric 대 단일 지표 1분위: 월평균/월표준편차 차이 검정").format(precision=4))
순 = 누적LS.sort_values(ascending=False)
ls경계 = pd.DataFrame({f"{a} 대 {순.index[2]}": 평가.평균차이(LS[a], LS[순.index[2]]) for a in 순.index[:2]}).T
ls경계["판정"] = np.where(ls경계["p"] > 0.05, "경계", "차이 있음")
남김(ls경계, "LS상위2_평균차이").style.set_caption(f"LS 누적 상위 2({', '.join(순.index[:2])}) 대 3위: 월 LS 차이 평균 검정").format(precision=4)
```

### 9. p ≤ 0.05·'경계' 판정 (#80)
따라하기 02는 원문 결론을 방향·크기로만 판정한다(원문은 크기 비교 문장). 여기는 HEAD 판에 있던 p 기준 판정: 방향이 맞고 p ≤ 0.05면 '같음', 방향만 맞으면 '경계', 방향이 틀리면 '다름'. p = 월 차이 평균의 t(OLS)에서(양측 정규).
- 감사표 461줄: p 기준을 빼면 그림 36 줄이 '경계' → '같음'(방향만). 그림 18 줄은 '다름' 그대로.

```python
%%절 7.5.2
누적Q1 = (1 + qq).prod() - 1
최고단일 = 누적Q1.drop("Multi-Metric").idxmax()
월등 = 평가.평균차이(qq["Multi-Metric"], qq[최고단일])
본현실 = 현실표["현실판_비용월0.2%(본)"].dropna()
벤치 = F200["시장"].reindex(본현실.index)
초과 = 평가.평균차이(본현실, 벤치)
p판정 = lambda ok, p: "같음" if ok and p <= 0.05 else ("경계" if ok else "다름")
방향 = lambda ok: "같음" if ok else "다름"
ok18, ok36 = 누적Q1.idxmax() == "Multi-Metric", (1 + 본현실).prod() > (1 + 벤치).prod()
남김(pd.DataFrame([
    ["p.15 그림 18", "Multi-Metric 1분위 누적이 단일 지표 1분위보다 '월등하게 우수'",
     f"Multi {누적Q1['Multi-Metric']:.3f}, 단일 최고 {최고단일} {누적Q1[최고단일]:.3f}", 월등["t(OLS)"], 월등["p"], 방향(ok18), p판정(ok18, 월등["p"])],
    ["p.26 그림 36", "현실판(비용 월 0.2%)이 KOSPI 200 동일가중지수를 Outperform",
     f"현실판 {(1 + 본현실).prod() - 1:.3f}, I.401 {(1 + 벤치).prod() - 1:.3f}", 초과["t(OLS)"], 초과["p"], 방향(ok36), p판정(ok36, 초과["p"])],
], columns=["원문 쪽·그림", "원문 결론", "우리", "t(OLS)", "p", "방향만(따라하기)", "p ≤ 0.05 기준(우리 추가)"]), "p경계_판정")
```

### 10. 원인 바꾸기 판 (#86~#90)
원인 후보 하나만 바꿔 돌린 그림 19 숫자(Q1 월평균/월표준편차)·LS 누적을 원문 그림 19와 나란히. 값을 원문에 맞추려고 고르지 않는다.
- 연간(예전 본문, 7절과 같은 판) / 분기, 최근 연속 4분기 합(HEAD 판 본문) / 분기, 최근 분기 × 4 / 배수 그대로(P/B 등 낮을수록 1분위, 적자·자본잠식이 '싼 쪽'에 섞임) / 연간 + EV = 시총 + 단기차입금 + 유동성장기부채 − 현금(회원 결정 2026-11-04 15:51 확인판, 두 항목이 결산 파일뿐이라 연간).
- 흐름 항목 4분기 합·×4, 정렬 꼴, EV 정의 모두 원문에 없음.

```python
%%절 7.5.2
for n, m in 지표명.items():  # 원문 꼴 '배수 그대로'(진단용): P/B = 1 / (B/P), 낮을수록 1분위
    setattr(지표, m + "_배수", lambda s, m=m: 1 / getattr(지표, m)(s))
def 판돌리기(d, 꼴):
    res = {n: 포.실행({**기본, **d, "지표": [(m + 꼴[0], 꼴[1])]}) for n, m in 지표명.items()}
    res["Multi-Metric"] = 포.실행({**기본, **d, "지표": [(m + 꼴[0], 꼴[1]) for m in 지표명.values()], "합산": "z합산"})
    return res
판 = {"본문(발표값)": 결과, "연간(예전 본문)": 연간,
     "분기, 최근 연속 4분기 합(HEAD 판 본문)": 판돌리기(dict(재무주기=("분기", "4분기합")), ("", "높게")),
     "분기, 최근 분기 × 4": 판돌리기(dict(재무주기=("분기", "최근분기x4")), ("", "높게")),
     "배수 그대로": 판돌리기({}, ("_배수", "낮게"))}
if all(읽기.있음(i) for i in ("M000902009", "M000902044")):
    차입금 = lambda ms: [("EBITDA_EV_차입금" if m == "EBITDA_EV" else m, "높게") for m in ms]
    판["연간, EV = 시총 + 단기차입금 + 유동성장기부채 − 현금"] = {**연간, "EV/EBITDA": 포.실행({**연간설정, "지표": 차입금(["EBITDA_EV"])}),
        "Multi-Metric": 포.실행({**연간설정, "지표": 차입금(지표명.values()), "합산": "z합산"})}
else:
    print("M000902009·M000902044 파일이 없어 EV 차입금 판을 못 돌림")
qd = lambda o: o["수익률"]["동일"]["Q1"]
rows = {}
for k, res in 판.items():
    r19 = pd.Series({n: qd(o).mean() / qd(o).std() for n, o in res.items()})[그림19.index]
    ls = pd.Series({n: (1 + qd(o) - o["수익률"]["동일"][숏.get(n, "Q5")]).prod() - 1 for n, o in res.items() if n != "Multi-Metric"})
    rows[k] = {("그림19", n): v for n, v in r19.items()} | {("LS 누적", n): v for n, v in ls.items()} | {("Q1 개월 수", n): int(qd(o).count()) for n, o in res.items()} | {
        ("요약", "그림19와 차이 절대값 평균"): (r19 - 그림19).abs().mean(), ("요약", "Multi가 1위"): r19.idxmax() == "Multi-Metric",
        ("요약", "LS 상위 2 = P/B·P/E"): set(ls.nlargest(2).index) == {"P/B", "P/E"}}
원인표 = pd.DataFrame(rows)
원인표["원문"] = pd.Series({("그림19", n): v for n, v in 그림19.items()})
남김(원인표, "원인표")
```

```python
%%절 7.5.3
# 새 따라하기 03과 같은 기준 설정(보통주만, 수정주가, 발표된 재무 값 그대로)
기본 = {**포.기본설정, "이름": "NH퀄리티로우볼", "시장": ["KS"], "기간": 기간, "기준일": "마지막거래일", "리밸런싱": 3, "가격": 읽기.수정주가,
       "재무시점": ("발표월말", 150), "재무주기": ("발표값",), "제외": ["보통주", ("거래정지", {"아이템": 읽기.정지여부}), "관리종목"],
       "고르기": ("분위", 5), "비중": ["동일"], "거래비용": dict(수수료=0, 거래세=0, 슬리피지=0)}
지표들 = {"ROE": ("ROE", "높게"), "영업이익률": ("영업이익률", "높게"), "부채비율": ("부채비율", "낮게"),
         "총자산회전율": ("총자산회전율", "높게"), "로우볼": ("변동성_원문", "낮게")}  # 따라하기 03과 같음(상장 기간 조건 없음)
결과 = {n: 포.실행({**기본, "지표": [m]}) for n, m in 지표들.items()}
누적 = pd.DataFrame({n: (1 + o["수익률"]["동일"]).prod() - 1 for n, o in 결과.items()}).T
LS = pd.DataFrame({n: o["수익률"]["동일"]["Q1"] - o["수익률"]["동일"]["Q5"] for n, o in 결과.items()})
누적
```

### 2. 요인 만들기 (옮김 #1, 원래 칸 8)
성과표용 요인. DataGuide 2×3 요인지수(SMB·HML·MOM)와 시장(코스피 I.001), 무위험(콜금리 E11.02.003.001). 원문에는 요인 회귀가 없다.

```python
%%절 7.5.3
F = 평가.요인(포.기준일(기본), 기본)
F.describe()
```

### 3. LS 성과표와 퀄리티 LS 3개 (옮김 #2·#3, 원래 칸 10·28)
원문 LS는 ROE(그림 26·27)와 로우볼(그림 32~37)뿐이다. 영업이익률·부채비율·총자산회전율 LS와 다섯 LS의 성과 지표는 우리 추가. t는 OLS.

```python
%%절 7.5.3
평가.누적그림(LS[["영업이익률", "부채비율", "총자산회전율"]], "NH 퀄리티: 원문에 없는 LS 3개(Q1 − Q5)");
LS성과 = 평가.성과표(LS, F, list(LS))
LS성과
```

### 4. 분위별 성과표 (옮김 #4, 원래 칸 12)

### 5. 리밸런싱 주기: 총자산회전율 12개월과 LS (옮김 #5, 원래 칸 14) — 보고서에서 뺌(회원 결정)
원문 그림 29(p.21)는 총자산회전율 3·6개월의 Q1·Q5뿐이다. 12개월과 LS는 우리 추가. 로우볼 3·6·12개월(원문 그림 37)은 6·11절 판정·검정에 쓰려고 같이 돌린다.

```python
%%절 7.5.3
주기, 주기LS = {}, {}
for n in ["총자산회전율", "로우볼"]:
    for k in [3, 6, 12]:
        r = 결과[n]["수익률"]["동일"] if k == 3 else 포.실행({**기본, "지표": [지표들[n]], "리밸런싱": k})["수익률"]["동일"]
        주기LS[(n, k)] = r["Q1"] - r["Q5"]
        주기[(n, k)] = {"Q1 누적": (1 + r["Q1"]).prod() - 1, "Q5 누적": (1 + r["Q5"]).prod() - 1, "LS 누적": (1 + 주기LS[(n, k)]).prod() - 1}
주기표 = pd.DataFrame(주기).T
주기표
```

### 6. 확인판: 자본 > 0만, 우선주 포함 (옮김 #6·#7, 원래 칸 16)
- 자본 > 0만: ROE·부채비율을 자본이 0보다 큰 종목만으로 다시 계산(자본이 음수면 두 지표 부호가 뒤집혀 Q1에 들어갈 수 있음). 원문에 정의 없음.
- 우선주 포함: 예전 본문(우선주 포함)과 같은 판. 지금 본문은 보통주만(회원 결정)이라 예전 '확인판(보통주만)'이 본문이 됐고, 비교 대상을 우선주 포함 쪽으로 바꿨다.
- 판정 기준은 따라하기 8절과 같은 네 문장(원문 p.18·23·26).

```python
%%절 7.5.3
판정 = lambda 누적, 주기표: {
    "4개 퀄리티 지표 모두 2~4분위 누적이 가장 높음": all(누적.loc[n].idxmax() in ["Q2", "Q3", "Q4"] for n in list(지표들)[:4]),
    "4개 퀄리티 지표 모두 5분위 누적이 가장 낮음": all(누적.loc[n].idxmin() == "Q5" for n in list(지표들)[:4]),
    "로우볼 LS 누적이 양(+)": bool((1 + LS["로우볼"]).prod() - 1 > 0),
    "로우볼: 6·12개월 리밸런싱 LS 누적 > 3개월": bool(min(주기표.loc[("로우볼", 6), "LS 누적"], 주기표.loc[("로우볼", 12), "LS 누적"]) > 주기표.loc[("로우볼", 3), "LS 누적"])}
누적2 = 누적.copy()
for n in ["ROE", "부채비율"]:
    누적2.loc[n] = (1 + 포.실행({**기본, "지표": [(f"{n}_양자본", 지표들[n][1])]})["수익률"]["동일"]).prod() - 1
display(pd.concat({"본문": 누적.loc[["ROE", "부채비율"]], "확인판(자본>0만)": 누적2.loc[["ROE", "부채비율"]]}))
우선 = 포.실행({**기본, "지표": [지표들["로우볼"]], "제외": [e for e in 기본["제외"] if e != "보통주"]})["수익률"]["동일"]
우선LS = (1 + 우선["Q1"] - 우선["Q5"]).prod() - 1
print(f"로우볼 LS 누적: 본문(보통주만) {(1 + LS['로우볼']).prod() - 1:.4f}, 우선주 포함 판 {우선LS:.4f}")
확인판정 = pd.DataFrame({"본문": 판정(누적, 주기표), "확인판(자본>0만)": 판정(누적2, 주기표)}).rename_axis("원문 주장")
확인판정["우선주 포함 판"] = pd.Series({"로우볼 LS 누적이 양(+)": bool(우선LS > 0)})
확인판정
```

### 7. 연간 재무 150·180일 판 (옮김 #8, 원래 칸 18)
원문 p.18은 '분기/연간 재무제표가 발표되는 월의 월말'. 연간 재무만 쓰고 발표일이 없으면 결산일 + 150·180일로 대신하는 판은 우리 추가(150·180은 원문에 없음).

```python
%%절 7.5.3
연간 = {n: 포.실행({**기본, "재무주기": ("연간",), "재무시점": ("발표일", 150), "지표": [m]}) for n, m in 지표들.items() if n != "로우볼"}
늦게 = {n: 포.실행({**기본, "재무주기": ("연간",), "재무시점": ("발표일", 180), "지표": [m]}) for n, m in 지표들.items() if n != "로우볼"}
lsof = lambda d: (1 + pd.DataFrame({n: o["수익률"]["동일"]["Q1"] - o["수익률"]["동일"]["Q5"] for n, o in d.items()})).prod() - 1
연간표 = pd.DataFrame({"발표값(본문)": (1 + LS.drop(columns="로우볼")).prod() - 1, "연간 150일": lsof(연간), "연간 180일": lsof(늦게)}).rename_axis("LS 누적")
연간표
```

### 8. 무위험이자율 확인판 (옮김 #9, 원래 칸 18)
원문에 무위험이자율이 없다. 기본은 콜금리(E11.02.003.001). 확인판은 무담보콜(E11.02.003.002), 2000년 이후는 통안채 364일(E11.02.003.009). 포트폴리오는 안 바뀌고 샤프·알파만 바뀐다(Q1으로 본다).

```python
%%절 7.5.3
Q1x = pd.DataFrame({n: o["수익률"]["동일"]["Q1"] for n, o in 결과.items()}).dropna(how="all")
금리 = {"콜(기본)": F, "무담보콜": 평가.요인(포.기준일(기본), {**기본, "무위험": "E11.02.003.002"}),
       "통안364(2000년 이후)": 평가.요인(포.기준일(기본), {**기본, "무위험": "E11.02.003.009"}).loc["2000":]}
금리표 = pd.concat({k: 평가.성과표(Q1x.loc[f.index[0]:], f)[["샤프", "CAPM α(월)", "Carhart α(월)"]] for k, f in 금리.items()}, axis=1)
금리표
```

### 10. 주장 1·2 p '경계' 판정과 원인 바꾸기 판 (옮김 #11·#12, 원래 칸 23·24·26)
원문에 검정이 없다. 판정 기준은 따라하기 13절(HEAD 판)과 같다. t는 OLS(`평가.평균차이`, 전에는 NW).
- 주장 1(p.18, 그림 22~25): 지표마다 Q2~Q4 중 누적 최고가 Q1·Q5 중 누적 최고보다 크면 '같음'. 그 두 분위 월수익률 차이 평균의 p > 0.05이면 '경계'.
- 주장 2(p.18 '로우 퀄리티 현저히 낮음'): Q5 누적이 나머지 네 분위 중 최저보다 낮으면 '같음'. Q5와 그 분위 월 차이 평균 p > 0.05이면 '경계'.
- 원인 바꾸기 판: 분기 4분기 합(예전 본문)·분기 최근 분기 × 4·연간 150일·연간 180일·연간 발표 당일부터·자본 > 0만. 모두 원문에 없는 선택.

```python
%%절 7.5.3
def 지표판정(r):
    누 = (1 + r).prod() - 1
    중, 끝 = 누[["Q2", "Q3", "Q4"]].idxmax(), 누[["Q1", "Q5"]].idxmax()
    아래 = 누.drop("Q5").idxmin()
    p1, p2 = 평가.평균차이(r[중], r[끝])["p"], 평가.평균차이(r["Q5"], r[아래])["p"]
    판 = lambda ok, p: "경계" if p > 0.05 else ("같음" if ok else "다름")
    return {**{f"누적 {g}": 누[g] for g in 누.index}, "주장1: 중간 최고 대 끝 최고": f"{중} 대 {끝}", "주장1 p": p1, "주장1 판정": 판(누[중] > 누[끝], p1),
            "주장2: Q5 대 나머지 최저": f"Q5 대 {아래}", "주장2 p": p2, "주장2 판정": 판(누["Q5"] < 누[아래], p2)}
퀄 = [n for n in 지표들 if n != "로우볼"]
분기4 = {n: 포.실행({**기본, "재무주기": ("분기", "4분기합"), "지표": [지표들[n]]}) for n in 퀄}
곱4 = {n: 포.실행({**기본, "재무주기": ("분기", "최근분기x4"), "지표": [지표들[n]]}) for n in 퀄}
당일 = {n: 포.실행({**기본, "재무주기": ("연간",), "재무시점": ("발표당일", 150), "지표": [지표들[n]]}) for n in 퀄}
양자본 = {n: 포.실행({**기본, "지표": [(f"{n}_양자본", 지표들[n][1])]}) for n in ["ROE", "부채비율"]}
판들 = {"본문(발표값)": 결과, "분기, 4분기 합(예전 본문)": 분기4, "분기, 최근 분기 × 4": 곱4, "연간 150일": 연간, "연간 180일": 늦게, "연간 발표 당일부터": 당일, "자본>0만": 양자본}
지표별 = pd.concat({k: pd.DataFrame({n: 지표판정(o["수익률"]["동일"]) for n, o in res.items() if n != "로우볼"}).T for k, res in 판들.items()})
지표별
```

### 11. 로우볼 리밸런싱 검정: 6·12 대 3개월, 12 대 6개월 (옮김 #13, 원래 칸 25) — 보고서에서 뺌(회원 결정, 리밸런싱 주기)
원문 p.26 '6개월과 12개월 단위 리밸런싱 포트폴리오는 유사한 성과'. 원문에 검정이 없다. p > 0.05이면 '경계'. t는 OLS.

```python
%%절 7.5.3
볼LS = {k: 주기LS[("로우볼", k)] for k in (3, 6, 12)}
주장4 = pd.DataFrame({f"{k}개월 대 3개월": 평가.평균차이(볼LS[k], 볼LS[3]) for k in (6, 12)} | {"12개월 대 6개월": 평가.평균차이(볼LS[12], 볼LS[6])}).T
주장4["누적(앞)"] = [(1 + 볼LS[k]).prod() - 1 for k in (6, 12, 12)]
주장4["누적(뒤)"] = [(1 + 볼LS[k]).prod() - 1 for k in (3, 3, 6)]
주장4["판정"] = np.where(주장4["p"] > 0.05, "경계", np.where(주장4["누적(앞)"] > 주장4["누적(뒤)"], "앞이 큼", "뒤가 큼"))
주장4
```

### 13. 요약표용 요인 F·이상값 (옮김 #14, 원래 따라하기 13절 칸 23)
따라하기 03은 `평가.저장(S, None, "03_NH퀄리티로우볼", …)`로 월수익률만 남긴다. 요약표(`평가.요약`)는 F가 None이면 `결과_월수익률/요인/03_NH퀄리티로우볼.pkl`을 읽으므로, 2절의 같은 요인 F(DataGuide, 기준 설정이 따라하기 03과 같음)를 거기에 쓴다. 이상값(+300% 종목-월 보유 건수, L/S는 Q1 + Q5)은 이 노트북 폴더 csv로만 남긴다(따라하기 pkl에 없어 요약표의 03 '이상값 보유' 칸은 빈칸). 원문에는 요인·이상값이 없다.

```python
%%절 7.5.3
요인폴더 = 평가.결과폴더 / "요인"
요인폴더.mkdir(parents=True, exist_ok=True)
pd.to_pickle(F, 요인폴더 / "03_NH퀄리티로우볼.pkl")  # 따라하기 03 평가.저장 이름과 같게
이상 = {n: 포.이상값보유(o) for n, o in 결과.items()}
이상값 = pd.Series({f"NH퀄리티로우볼 {n} Q1": int(이상[n][("동일", "Q1")]) for n in 결과} | {f"NH퀄리티로우볼 {n} LS": int(이상[n][("동일", "Q1")] + 이상[n][("동일", "Q5")]) for n in ["ROE", "로우볼"]}, name="이상값 보유 건수")
이상값.to_csv(출력 / "03_이상값보유.csv", encoding="utf-8-sig")
이상값
```

### 0. 신마법공식 원문 정리 (따라하기 0절에서 옮김)
> **책 원문 미확인(책을 못 구함) — 정의는 책을 인용한 2차 자료, 2차 자료에 없는 설정은 우동호 설정을 따름.**
> 2차 자료(2026-11-05 2차 진단 7에서 웹 글자를 직접 받아 확인, 책 쪽수는 아무도 확인 못 함):
> - intelliquant.ai/article/573(신마법공식 1.0): '신마법공식 1.0 (저PBR + 고GP/A) 기대 CAGR(Compound Annual Growth Rate), 연복리수익률 -> 약 15 ~ 20% 이상 모든 주식의 PBR, GP/A 순위 계산 2개 순위를 더한 통합 순위가 가장 높은 종목 30개 매수 연 1회 리밸런싱 테스트기간: 2002년 7월 ~ 2016년 6월', '할수있다 퀀트투자에서 테스트한 결과인 24.46%', '할수있다 퀀트투자에서는 Top 50의 수익률이 25.44%로 가장 높게'.
> - intelliquant.ai/article/578(신마법공식 2.0): '신마법공식 2.0은 신마법공식 1.0에 소형주 팩터만 추가한 것', '소형주(시가총액 하위 20%) 기업만 매수'.

> - 책 목차(교보 E000002997763, YES24 45504859): 10장 '신마법공식', GP/A는 9장.
> 책의 숫자(연 24.46%, Top 50 25.44%)는 **2차 인용, 원문 쪽 미확인**이다. 기간 '2002년 7월 ~ 2016년 6월'이 책의 기간인지 글쓴이 시험 기간인지, 결과 수익률이 복리인지는 글에 직접 적혀 있지 않다(같은 목록 맨 앞에 '기대 CAGR …, 연복리수익률'이 있어 책 요약일 수 있음 → 12절 K4).

### 3. 설정 (따라하기 3절에서 옮김)
신마법공식(1.0), 신마법공식 2.0(시가총액 하위 20%), 같은 설정의 우동호 IMF(GP/A + EY)를 나란히 돌린다(우동호IMF 나란히 = 원문에 없는 추가, 감사표 60).

```python
%%절 7.5.4
def 전체순위(s):  # 2.0: 소형주로 거르기 전(우동호 거르기 뒤) 전체에서 PBR·GP/A 순위 합 = 인텔리퀀트 578 '모든 주식의 PBR, GP/A 순위 계산 … 30개 매수 소형주(시가총액 하위 20%) 기업만 매수'
    c0 = 종목거르기.적용(s, [("시장", {"시장들": 포.우동호2023["시장"]}), "쓰지말것"] + 포.우동호2023["제외"])[0]
    v = pd.DataFrame({m: getattr(지표, m)(s).reindex(c0) for m in ["BP", "GPA"]}).replace([np.inf, -np.inf], np.nan)
    return 포.점수(v, [("BP", "높게"), ("GPA", "높게")], "순위합산")
지표.신마법전체순위 = 전체순위
설정 = {"신마법공식": {**포.우동호2023, "이름": "신마법공식", "기간": 기간, "끝": "마지막거래일", "지표": [("BP", "높게"), ("GPA", "높게")]},
        "신마법공식2.0": {**포.우동호2023, "이름": "신마법공식2.0", "기간": 기간, "끝": "마지막거래일", "지표": [("신마법전체순위", "높게")],
                         "제외": 포.우동호2023["제외"] + [("지표있음", {"지표들": ["BP", "GPA"]}), ("시총분위", {"분위": 1})]},
        "우동호IMF": {**포.우동호2023, "이름": "우동호2023_IMF", "기간": 기간, "끝": "마지막거래일", "지표": [("GPA", "높게"), ("EY", "높게")]}}
pd.DataFrame(설정).astype(str)
```

### 4. 실행: 상위 30종목, EW·VW (따라하기 4절에서 옮김)
VW 판·연환산 칸·CAPM·FF3·Carhart 회귀·'이상값 보유' 칸은 원문(책·우동호)에 없는 추가다. t는 OLS. 요인 = DataGuide SMB·HML·MOM(`평가.요인`, 과제 공지 05:35: 요인은 DataGuide). KOSPI 행의 알파·t는 시장을 자기 자신에 회귀한 값이라 뜻이 없다. '이상값 보유' = 보유한 달 중 종목 수익률 +300% 초과 종목-월 수(0보다 크면 이상값 영향).

```python
%%절 7.5.4
결과 = {k: 포.실행({**v, "비중": ["동일", "시총"]}) for k, v in 설정.items()}
F = 평가.요인(포.기준일(설정["신마법공식"]), 설정["신마법공식"])
R = pd.DataFrame({f"{k}_{b}": o["수익률"][(b, "상위30")] for k, o in 결과.items() for b in ["동일", "시총"]})
R["KOSPI"] = F["시장"]
R = R.dropna(how="all")
표 = 평가.성과표(R, F)
표.insert(0, "이상값 보유", pd.Series({f"{k}_{b}": 포.이상값보유(o)[(b, "상위30")] for k, o in 결과.items() for b in ["동일", "시총"]}))
표.to_csv(저장폴더 / "옮김04_4절_성과표.csv", encoding="utf-8-sig")
표
```

### 8. 신마법 차이 이유 후보·원문과 다르게 한 것 (따라하기 8절·9절에서 옮김)
- **신마법공식**: 책 원문이 없어 원문 숫자와 비교할 수 없다(책 숫자는 2차 인용뿐, 12절). 우동호 IMF와의 차이는 EY 대신 PBR을 쓴 것뿐이다.
- **벤치마크**: 신마법공식은 수정주가(배당 미포함)라 코스피 가격지수와 정의가 맞다.

| 항목 | 원문·정의 | 우리 | 이유 |
|---|---|---|---|
| 책 원문(강환국 2017 10장) | 미확인 | 정의·30종목·연 1회·2.0은 2차 자료(0절), 나머지(리밸런싱 달·재무 시점·제외 업종·가중·2.0 분위 모집단)는 우동호 IMF 설정·01 본문 방식 | 책 숫자를 지어내지 않기. 2차 자료에 없는 칸은 `_검증/임의값_근거.md`, 확인판은 12절 |
| PBR 순위 | PBR 낮을수록 좋음 | BP(PBR 역수) 높을수록 좋음 | 자본이 음수인 종목은 PBR이 음수라 '가장 싼 쪽'에 줄 서는 것 막기. 자본이 양수인 종목끼리는 순서가 같음 |

데이터 규칙(상장 구간·'쓰지 말 것' 칸·호텔신라 FY2023·끝 날짜·코드 재사용·재무 값 뽑기·두 결산월)은 따라하기 04 9절과 같다.

### 10. Novy-Marx 판: VW·연환산·기록 표 (따라하기 7절에서 옮김)
원문 표 7 Panel C는 동일가중(p.21 'buys one dollar of each')만 있다. 시총가중(VW) 판, 성과표의 연환산 칸, 리밸런싱일별 기록 표는 원문에 없는 추가다(감사표 76·53·62). 원문 대조표(EW, `04_NovyMarx대조.csv`)는 따라하기 04 7절에 남는다.
설정은 따라하기 04 7절과 같게 둔다(6월 마지막 거래일, 비금융(`"금융업"` 넓은 이름 규칙 + `"투자기구"` = 원문 p.8 SIC 첫 자리 6 범위) → GP/A·B/M 둘 다 있음 → 시총 상위 500, 150/200/150, B/M = 장부가 원문 정의 ÷ 6개월 전 시총, 결산 후 5개월, 비용 0) + 회원 결정(보통주만, 수정주가). **따라하기 7절 설정이 바뀌면 이 칸의 `nm`도 같이 바꾼다.**

```python
%%절 7.5.4
nm = {**포.기본설정, "이름": "NovyMarx2013_표7C", "시장": ["KS", "KQ"], "기간": None, "기준일": "마지막거래일",
      "리밸런싱": 12, "재무시점": ("결산후개월", 5), "재무연도": "직전년도", "가격": 읽기.수정주가,
      "제외": ["보통주", "금융업", "투자기구", ("지표있음", {"지표들": ["GPA", "BP_장부원문_6개월"]}), ("시총상위", {"n": 500})],  # 비금융 = 원문 p.8 SIC 첫 자리 6(금융·보험·부동산 60~67: 리츠 6798·스팩 6770·투자회사 67 포함, OSHA SIC 편람) → 금융업 넓은 이름 규칙 + 투자기구
      "지표": [("GPA", "높게"), ("BP_장부원문_6개월", "높게")], "합산": "순위합산", "고르기": ("상하N", 150),
      "비중": ["동일", "시총"], "거래비용": dict(수수료=0, 거래세=0, 슬리피지=0)}
# 기간(기본 방법 ②, 원문 '쓸 수 있는 전 기간'처럼). 끝 = 자료 끝 안에서 6월 말 → 다음 해 6월 말 보유가 다 들어가는 마지막 해
끝날 = 읽기.공통끝()
끝해 = max(y for y in range(1988, 끝날.year + 1) if pd.Timestamp(f"{y}-06-30") <= 끝날)
첫 = 읽기.일간(읽기.시총).index[0] + pd.DateOffset(months=6)  # B/M에 6개월 전 시가총액이 있어야 함
첫해 = 첫.year + (첫.month > 6)


def 표본수(y):  # 그해 6월 마지막 거래일에 시총 500 앞까지 거른 종목 수(직전년도 GP/A·B/M이 둘 다 있는 비금융 보통주)
    c = {**nm, "기간": (f"{y}-06", f"{y}-06")}
    return len(포.거르기(포.기준일(c)[0], {**c, "제외": nm["제외"][:-1]})[1])


수 = {}
for y in range(첫해, 끝해):
    수[y] = 표본수(y)
    if 수[y] >= 500:
        break
시작해 = y
assert 수[시작해] >= 500, f"500개 이상인 6월이 없음: {수}"
nm["기간"] = (f"{시작해}-06", f"{min(끝해, 시작해 + 2) if 시험 else 끝해}-06")
print("6월별 거른 종목 수:", 수, "→ 기간:", nm["기간"], "(자료 끝", 끝날.date(), ")")
assert {"M000911019", "M000903008"} <= set(읽기.재무항목()), "원문 장부가에 쓰는 이연법인세부채·우선주자본금 파일이 없음"
o = 포.실행(nm)
Fn = 평가.요인(포.기준일(nm), nm)
N = pd.DataFrame({f"{b}_{g}": o["수익률"][(b, g)] for b, g in o["수익률"]})
N["동일_H-L"], N["시총_H-L"] = N["동일_Q1"] - N["동일_Q3"], N["시총_Q1"] - N["시총_Q3"]
N = N.dropna(how="all")
표N = 평가.성과표(N, Fn, ["동일_H-L", "시총_H-L"])
표N.to_csv(저장폴더 / "옮김04_10절_NovyMarx_EW_VW.csv", encoding="utf-8-sig")
display(표N)
o["기록"].T
```

### 11. Novy-Marx 원문 차이 진단 (따라하기 11절에서 옮김)
- **비교할 수 있는 것은 방향뿐이다.** 원문 표 7 Panel C는 미국 1963-07~2009-12, 우리는 한국 10절 설정. 시장·기간·요인(Ken French 미국 요인 대 DataGuide 한국 요인)·무위험이 모두 달라 숫자 크기는 비교하지 않는다. 판정 기준(원문에 없음, 돌리기 전에 정함 = 감사표 87): High 월 초과수익 > 0, H−L > 0, H−L FF3 α > 0의 **부호**가 원문과 같은지.
- 정의 대조:

| 항목 | 원문(쪽) | 우리 본문(따라하기 7절) |
|---|---|---|
| GP/A | (REVT − COGS)/AT(표 2 설명, p.12) | 매출총이익(M000904007)/자산총계(기말) |
| B/M | 장부가 = 자본 + 이연법인세 − 우선주, 시가는 6개월 전(각주 3, p.11) | (자본총계(지배) + 이연법인세부채 M000911019 − 우선주자본금 M000903008)/6개월 전 시가총액 |
| 대형주 500 | 비금융, GP/A·B/M 둘 다 있는 종목 중 시총 상위 500(p.21·22) | 보통주·금융업(넓은 이름 규칙)·투자기구 제외 → 둘 다 있음 → 시총 상위 500(유가+코스닥) |
| 비금융 | SIC 첫 자리 6 제외(p.8) = 금융·보험·부동산 60~67(리츠 6798·스팩 6770·지주/투자회사 67 포함, OSHA SIC 편람·SEC SIC 목록) | `"금융업"`(거래소 업종 낱말, 업종 빈 날 이름 규칙에 리츠·스팩·'지주'로 끝남 포함, WICS '금융') + `"투자기구"`(리츠·선박·부동산투자, 조정자 결정 2026-11-06 06:2x). SIC 6과 1:1 아님(한계) |
| 결합·묶음 | 두 순위 평균(p.21), 150/200/150(표 7 Panel C n) | 순위합산(평균과 순서 같음), `상하N` 150 |
| 가중 | 동일(one dollar each) | 동일(본문), 시총(10절) |
| 리밸런싱·재무 | 6월 말 연 1회, 다음 해 6월 말부터(각주 4) | 6월 마지막 거래일, 결산 후 5개월 |

아래 표는 예전 판에서 원문 규칙을 하나씩 더해 간 숫자다(값을 맞추려고 고르지 않음, 규칙은 모두 원문 문장). 마지막 '+ 장부가' 줄 = 따라하기 7절 본문.

```python
%%절 7.5.4
단계 = {"예전 판(3등분·그날 시가·발표일)": dict(고르기=("분위", 3), 재무시점=("발표일", 150), bp="BP"),
       "+ 150/200/150": dict(고르기=("상하N", 150), 재무시점=("발표일", 150), bp="BP"),
       "+ B/M 6개월 전 시가": dict(고르기=("상하N", 150), 재무시점=("발표일", 150), bp="BP_6개월"),
       "+ 재무 결산 후 5개월": dict(고르기=("상하N", 150), 재무시점=("결산후개월", 5), bp="BP_6개월"),
       "+ 장부가 = 자본 + 이연법인세 − 우선주(각주 3, = 7절 본문)": dict(고르기=("상하N", 150), 재무시점=("결산후개월", 5), bp="BP_장부원문_6개월")}
rows = {}
for k, d in 단계.items():
    c = {**nm, "비중": ["동일"], "고르기": d["고르기"], "재무시점": d["재무시점"], "지표": [("GPA", "높게"), (d["bp"], "높게")],
         "제외": ["보통주", "금융업", "투자기구", ("지표있음", {"지표들": ["GPA", d["bp"]]}), ("시총상위", {"n": 500})]}  # 비금융 = 원문 p.8 SIC 첫 자리 6(금융·보험·부동산 60~67: 리츠 6798·스팩 6770·투자회사 67 포함, OSHA SIC 편람) → 금융업 넓은 이름 규칙 + 투자기구
    r = 포.실행(c)["수익률"]["동일"]
    x = pd.DataFrame({"High": r["Q1"], "H-L": r["Q1"] - r["Q3"]}).dropna(how="all")
    t = 평가.성과표(x, Fn, ["H-L"])
    rows[k] = {"High 월 초과(%)": t.loc["High", "초과평균 α(월)"] * 100, "High t": t.loc["High", "초과평균 t(OLS)"],
               "H-L 월(%)": t.loc["H-L", "초과평균 α(월)"] * 100, "H-L t": t.loc["H-L", "초과평균 t(OLS)"],
               "H-L FF3 α(%)": t.loc["H-L", "FF3 α(월)"] * 100, "H-L FF3 t": t.loc["H-L", "FF3 t(OLS)"]}
rows["원문 표 7 Panel C(미국)"] = {"High 월 초과(%)": 0.78, "High t": 3.68, "H-L 월(%)": 0.64, "H-L t": 5.21, "H-L FF3 α(%)": 0.39, "H-L FF3 t": 3.77}
진단 = pd.DataFrame(rows).T
부호 = np.sign(진단[["High 월 초과(%)", "H-L 월(%)", "H-L FF3 α(%)"]])
진단["부호가 원문과 같은 칸(3칸 중)"] = (부호 == 부호.iloc[-1]).sum(axis=1)
진단.to_csv(저장폴더 / "옮김04_11절_NovyMarx진단.csv", encoding="utf-8-sig")
진단
```

### 12. 2차 자료 대조와 확인판 K1-K5 (따라하기 12절에서 옮김)
2차 자료에 없는 칸은 본문을 바꾸지 않고 확인판으로만 돌린다. 판 목록은 `_검증/2차_07_신마법.py`(2026-11-05, 진행기록 02:32 절에 돌리기 전에 정함)와 같다. 따라하기 HEAD 판은 그 스크립트의 옛 출력(`_검증/2차_07_신마법.csv`, 옛 공통 코드)을 읽기만 했는데, 여기서는 지금 기준 설정(3절)으로 다시 계산한다(옛 csv는 덮어쓰지 않음). 가중은 동일가중.
- K1 상위 50(책 'Top 50 25.44%' 2차 인용) / K2 GP/A > 0·자본 > 0만(github screener `(gp/a > 0) & (pbr > 0)`) / K3 2.0 분위 모집단 = 남은 종목(지표있음 없이) / K4 기간 2002-07~2016-06·7월 첫 거래일(573 '테스트기간', 책 기간인지 글쓴이 기간인지 불분명) / K5 2.0 순위를 소형주 안에서(처음 본문).

```python
%%절 7.5.4
def GPA양수(s, c):  # github screener `df['gp/a'] > 0`(K2에서만)
    v = 지표.GPA(s).reindex(c)
    return c[(v > 0).to_numpy()]
종목거르기.GPA양수 = GPA양수
기본 = 설정["신마법공식"]
소형 = [("지표있음", {"지표들": ["BP", "GPA"]}), ("시총분위", {"분위": 1})]
양수 = ["자본잠식", "GPA양수"]  # 자본잠식 = 자본총계 ≤ 0 제외 = PBR > 0
def 전체순위_양수(s):  # K2 2.0: 양수 조건까지 거른 전체에서 순위
    c0 = 종목거르기.적용(s, [("시장", {"시장들": 기본["시장"]}), "쓰지말것"] + 기본["제외"] + 양수)[0]
    v = pd.DataFrame({m: getattr(지표, m)(s).reindex(c0) for m in ["BP", "GPA"]}).replace([np.inf, -np.inf], np.nan)
    return 포.점수(v, [("BP", "높게"), ("GPA", "높게")], "순위합산")
지표.신마법전체순위_양수 = 전체순위_양수
전체 = [("신마법전체순위", "높게")]
K4 = {"기간": ("2002-07", "2004-07") if 시험 else ("2002-07", "2016-06")}  # 끝 = 끝 달 마지막 거래일(기본의 "끝")
판 = {"K1 1.0 상위 50": {**기본, "고르기": ("상위N", 50)}, "K1 2.0 상위 50": {**설정["신마법공식2.0"], "고르기": ("상위N", 50)},
     "K2 1.0 GP/A>0·자본>0": {**기본, "제외": 기본["제외"] + 양수},
     "K2 2.0 GP/A>0·자본>0": {**기본, "제외": 기본["제외"] + 양수 + 소형, "지표": [("신마법전체순위_양수", "높게")]},
     "K3 2.0 분위 모집단 = 남은 종목": {**기본, "제외": 기본["제외"] + [("시총분위", {"분위": 1})], "지표": 전체},
     "K4 1.0 2002-07~2016-06 7월": {**기본, **K4}, "K4 2.0 2002-07~2016-06 7월": {**설정["신마법공식2.0"], **K4},
     "K5 2.0 순위는 소형주 안에서(처음 본문)": {**기본, "제외": 기본["제외"] + 소형}}
def 한줄(o, k):
    r, rec = o["수익률"][k].dropna(), o["기록"]
    n, w = len(r), (1 + r).prod()
    return {"기간": f"{r.index[0]:%Y-%m}~{r.index[-1]:%Y-%m}", "개월": n, "누적(배)": w - 1, "연복리(CAGR)": w ** (12 / n) - 1,
            "연 산술평균": r.mean() * 12, "편입 수 최소": int(rec[f"편입:{k[1]}"].min()), "편입 수 최대": int(rec[f"편입:{k[1]}"].max())}
판7 = {"본문 1.0": 한줄(결과["신마법공식"], ("동일", "상위30")), "본문 2.0(전체에서 순위 → 소형주 안 상위 30)": 한줄(결과["신마법공식2.0"], ("동일", "상위30"))}
for 이름, c in 판.items():
    판7[이름] = 한줄(포.실행({**c, "비중": ["동일"]}), ("동일", f"상위{c['고르기'][1]}"))
판7 = pd.DataFrame(판7).T
판7.to_csv(저장폴더 / "옮김04_12절_K1-K5.csv", encoding="utf-8-sig")
print("2차 인용 숫자(쪽 미확인): 책 1.0 연 24.46%, 책 Top 50 25.44%(인텔리퀀트 573이 '할수있다 퀀트투자에서 테스트한 결과'로 인용). "
      "인텔리퀀트 자체 시험: 1.0 28.83%, 2.0 37.29%(2002-07~2016-06), 2.0 2016-07 이후 19.66%, 1.0 2016 이후 -0.13%.")
판7
```

## 요약표: 전략 × 가중 × 기간
따라하기 01~04와 연구 노트북이 `결과_월수익률/`에 남긴 월수익률을 모아 한 표로 만든다. 숫자는 모두 `평가.요약()`이 계산한다.
- **기간**: 전체(그 노트북이 돌린 기간), 1990-04~2004-03, 2004-04~2017-03, 2017-04~ (보유 월 기준). 24개월 미만 구간은 뺀다.
- **평균(월 초과)**: 월평균 초과수익률(포트폴리오 − 무위험. 롱숏(LS)은 수익률 그대로). **평균 t(OLS)**: 그 평균의 OLS t.
- **샤프**: 연환산. **MDD**: 최대 낙폭. **Carhart α(월)**: 시장·SMB·HML·MOM 4요인 알파, **α t(OLS)**: 그 t.
- 무위험이자율은 노트북마다 원문 설정을 따른다(01·04 국고3년, 02·03·연구 콜금리). 시장지수는 코스피 가격지수(I.001)라 배당 포함 가격(TR)을 쓴 전략은 알파가 배당만큼 높게 나온다.
- 전략 이름 끝의 Q1·LS는 분위 포트폴리오, '상위30'은 30종목 포트폴리오다.
- **이상값 보유** 칸: `평가.저장`에 이상값 건수를 넘긴 노트북 줄만 숫자가 있다(옮김_01·옮김_04·이로운_비중바꾸기·이로운_지표바꾸기·이로운_팩터순서). 따라하기 01~04 줄(02·03 포함)은 원문에 없는 칸이라 저장하지 않아 빈칸이다. 03의 이상값 건수는 `연구/결과/따라하기에서_옮김_03/03_이상값보유.csv`(옮김_03 칸 27), 02는 이상값 건수를 낸 노트북이 없다.

줄임말: EW(동일가중, Equal Weight), VW(시가총액가중, Value Weight), LS(롱숏: 좋은 분위를 사고 나쁜 분위를 판 수익률),
MDD(최대 낙폭, Maximum Drawdown), IR(정보비율, Information Ratio), CAPM(자본자산가격결정모형), FF3(파마-프렌치 3요인), Carhart(카하트 4요인: FF3 + 모멘텀),
OLS(보통최소제곱), NW는 회원 결정(2026-11-06)으로 뺌, t는 OLS, TR(배당 포함 수정주가).

```python
%%절 8
표 = 평가.요약()
표.to_csv(평가.결과폴더 / "요약표.csv", encoding="utf-8-sig")
print("파일:", sorted(f.name for f in 평가.결과폴더.glob("*.pkl")), "/ 줄 수:", len(표))
표
```

### 6. 확장 연구 핵심 결과
우동호 유니버스, 배당 포함 가격, 1990-04~2026-09. 주 벤치마크 = 유니버스 시가총액 가중(비용 0). 무위험이자율 = 교수님 제공 엑셀(`08.[액셀실습]TSMom(K200).xlsx`)의 CD1m 한 계열(교수님 제공 파일의 값, 1990-01~1994-12는 14% 고정). 출처 대조(`_검증/1106/무위험_CD1m_출처.md`, 2026-11-06): 1990-01~1994-12 14.0 고정 = 교수 가정으로 보임(원 데이터 없음), 1995-01~2021-11 = (ECOS 817Y002 CD(91일) + 콜금리(1일, 전체거래)) ÷ 2의 그 달 마지막 영업일 값과 324/324개월 같음(±0.005), 2021-12~2022-12 = 그 값의 다음 달 값(한 달 밀림), 2023-01~2026-10 46개월 = 출처 못 찾음(KORIBOR 1개월 등, 가장 많이 맞은 것도 14/46). 노트북은 강의 엑셀 CD1m을 그대로 씀. ECOS 원자료 = `과제02\data\ECOS_202611\`(같은 파일 `제출_원문대로\data\ECOS_202611\`). 콜금리로 바꾼 결과는 7절 `7_무위험_민감도`. 자세한 표와 설정·근거는 `연구/확장연구_20261103.ipynb`.
- **α 읽을 때**: 시장 요인은 코스피 가격지수(배당 없음)라 배당 포함 포트폴리오의 α에 배당만큼이 섞인다. 전략 없이 유니버스를 시총가중으로 들고만 있어도 Carhart α가 양수로 나온다(확장연구 4절 벤치마크 줄). α는 유니버스 벤치마크 줄과 견주어 본다.
- **여러 조합에서 고름(과최적화)**: 수익률 열이 2,880개(전략 48 × 비중 5 × 주기 4 × 비용 3)다. 그중 '상위'를 고르면 우연히 좋은 것이 뽑힐 수 있고, 여러 번 검정한 것에 대한 t 보정은 하지 않았다. 단계별 순서 전략은 근거 없는 '남길 비율 0.5'에 달려 있다. 그래서 상위 목록보다 6-2(중앙값)·6-3(기간별 일관성)을 함께 본다.

```python
%%절 8
x = pd.read_pickle(Path.cwd() / "결과" / "확장연구_20261103.pkl")
표 = x["표"]
전체 = 표.xs("전체", level="기간")
기본 = 전체.xs("비용기본", level="비용")
print(f"전략 {len(x['전략들'])}개, 결과표 {len(표)}줄")
display(표.loc[[i for i in 표.index if i[0].startswith("벤치:")]].xs("전체", level="기간").droplevel([1, 2, 3]).style.set_caption("벤치마크(전체 기간)").format(precision=4))
기본.loc["EY+ROC 순위합산"]["샤프"].unstack("주기").style.set_caption("마법공식(EY+ROC 순위합산): 비중 × 주기 샤프, 전체 기간, 기본 비용").format(precision=3)
```

#### 6-1. 동일가중·12개월·기본 비용: 샤프 상위 10 전략과 기준(전체 기간)

```python
%%절 8
k = 기본.xs(("동일", 12), level=["비중", "주기"]).sort_values("샤프", ascending=False)
k = k.assign(순위=range(1, len(k) + 1))
pd.concat([k.head(10), k.loc[["EY+ROC 순위합산"]]]).pipe(lambda t: t[~t.index.duplicated()])  # 같은 전략이 두 번 나오면 하나만(감사 L11)
```

#### 6-2. 비중 방식 × 주기: 전 전략 샤프 중앙값 / 거래비용 민감도(연환산 중앙값)

```python
%%절 8
전략줄 = lambda t: t.drop(index=[i for i in t.index if i[0].startswith("벤치:")])
display(전략줄(표.xs(("공통", "비용기본"), level=["기간", "비용"]))["샤프"].groupby(["비중", "주기"]).median().unstack("주기").style.set_caption("샤프 중앙값('공통' 기간: 모든 비중 방식이 같은 달부터)").format(precision=3))
전략줄(전체)["연환산"].groupby(["주기", "비용"]).median().unstack("비용")[["비용0", "비용기본", "비용2배"]].style.set_caption("연환산 수익률 중앙값").format(precision=4)
```

#### 6-3. 기간별 일관성: 동일가중·12개월·기본 비용에서 유니버스 벤치마크보다 샤프가 높은 전략 수

```python
%%절 8
b = 표.xs(("벤치:유니버스(시총가중)", "-", 0, "비용0"), level=[0, 1, 2, 3])["샤프"]
s = 전략줄(표).xs(("동일", 12, "비용기본"), level=["비중", "주기", "비용"])["샤프"].unstack("기간")
pd.DataFrame({"전략 수": s.notna().sum(), "벤치마크보다 샤프 높음": (s.gt(b)).sum(), "벤치마크 샤프": b}).T
```

### 7. 리포트에 쓸 핵심 표·그림 (`결과_전략수익률/`, `연구/확장연구_20261103.ipynb` 6절이 만듦)
기준 = 마법공식(EY+ROC 순위합산), 동일가중, 12개월, 기본 비용. 비교마다 조건 하나만 바꿨다. 표 아래 문장은 그 표의 숫자에서 코드가 만든다(가장 높은·낮은 줄을 그대로 옮김, 해석 없음).
칸: 연환산 = 기하 연환산 수익률, 표준편차(연) = 월 표준편차 × √12, 샤프 = 월 초과수익(교수님 제공 CD1m 차감, 연 %/1200, 지난달 말 값) 평균/표준편차 × √12, MDD = 최대 낙폭, 회전율 = 리밸런싱당 평균 한쪽 회전율(전 기간), 벤치 대비 연환산 = 같은 달들의 유니버스 시총가중 벤치마크와 연환산 차이, 정보비율 = 유니버스 벤치마크 대비. α·t는 코스피 가격지수(배당 없음) 기준이라 배당 포함 포트폴리오에서는 배당만큼 높게 나온다(6절).

```python
%%절 8
from IPython.display import Image
리포트 = Path.cwd() / "결과" / "결과_전략수익률"
칸 = ["개월", "연환산", "표준편차(연)", "샤프", "MDD", "회전율(전 기간)", "벤치 대비 연환산", "정보비율", "Carhart α(월)", "Carhart t(OLS)"]
제목 = {"1_비중": "비중 방식 (공통 기간)", "2_주기": "리밸런싱 주기", "3_지표": "지표 바꾸기 (순위합산)", "4_순서": "요인 순서", "5_비용": "거래비용 민감도", "6_기간": "기간별 (기준 전략과 벤치마크)"}
for n, t in 제목.items():
    x = pd.read_csv(리포트 / f"{n}.csv", encoding="utf-8-sig", index_col=list(range(5)))
    s = x[~x.index.get_level_values("전략").str.startswith("벤치:")]
    이름 = lambda i: " / ".join(str(v) for v in i if v not in ("비용기본", "전체", "공통"))
    print(f"[{t}] 줄 {len(s)}개. 샤프 가장 높음: {이름(s['샤프'].idxmax())} ({s['샤프'].max():.3f}), 가장 낮음: {이름(s['샤프'].idxmin())} ({s['샤프'].min():.3f}). "
          f"연환산 가장 높음: {이름(s['연환산'].idxmax())} ({s['연환산'].max():.2%}). MDD 가장 얕음: {이름(s['MDD'].idxmax())} ({s['MDD'].max():.2%}).")
    g = "기간" if "기간" in x.index.names else "기간칸"
    u = x.xs("벤치:유니버스(시총가중)", level="전략")["샤프"].groupby(level=g).first()  # 같은 기간의 유니버스 벤치마크 샤프
    print(f"  같은 기간 유니버스 시총가중 벤치마크보다 샤프가 높은 줄: {int((s['샤프'].to_numpy() > u.reindex(s.index.get_level_values(g)).to_numpy()).sum())}/{len(s)}개 (벤치 샤프 {', '.join(f'{k} {v:.3f}' for k, v in u.items())})")
    display(x[칸].style.set_caption(t).format(precision=4))
    if (리포트 / f"{n}.png").exists():
        display(Image(filename=리포트 / f"{n}.png"))
display(pd.read_csv(리포트 / "7_무위험_민감도.csv", encoding="utf-8-sig", index_col=[0, 1]).style.set_caption("무위험이자율 민감도: CD1m(본문) vs 콜금리, 전체 기간").format(precision=4))
# (제출본에서 바꿈) 이 칸의 결과_전략수익률 폴더 → 결과/결과_전략수익률
```
